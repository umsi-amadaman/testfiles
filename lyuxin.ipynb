```python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    # create the various mapping and order lists/dictionaries
    # return: list of episodes (in order), episode->name map, list of episode names (in order)
    episodes = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    names = {
        'EI' : 'The Phantom Menace', 'EII' : 'Attack of the Clones', 'EIII' : 'Revenge of the Sith', 
        'EIV': 'A New Hope', 'EV': 'The Empire Strikes Back', 'EVI' : 'The Return of the Jedi'
    }

    # we're also going to use this order to sort, so names_l will now have our sort order
    return episodes, names, [names[ep] for ep in episodes]

episodes, namesDict, namesList = genEpisodeNamesDF()
```

```python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
```

```python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN
    salo = indf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    return(salo)
```

```python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
```

```python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total = len(inpf)
    percs = []

    # loop over each column and calculate the number of people who have seen the movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count them
    # and divide by the percent
    for seen_ep in ['seen_' + ep for ep in episodes]:
        perc = len(inpf[~ pd.isna(inpf[seen_ep])]) / total
        percs.append(perc)

    # at this point percs is holding our percentages

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes],percs))
    seen_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    return(seen_per_df)
```

```python
seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
```

```python
# let's see what's inside
seenPerMovieDF
```

```python
def genPercentVis(indf, names_l=namesList):
    # input: indf, the dataframe as seen_per_df
    # input: names_l - a list of all the movies in series order (default namesList)
    # output: simple altair bar chart
    
    # ok, time to make the chart... let's make a bar chart (use mark_bar)
    bars = alt.Chart(indf).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        )
    )

    # at this point we don't really have a great plot (it's missing the annotations, titles, etc.)
    return(bars)

seenPerMovieVis = genPercentVis(seenPerMovieDF)
```

```python
def augmentPercentVis(base):
    # input: base (the base vis, i.e., bars as above)
    # we're going to overlay the text with the percentages, so let's make another visualization
    # that's just text labels


    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    seen_movies = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=180
    ).properties(
        # add a title
        title="Which 'Star Wars' Movies Have you Seen?"
    )

    return(seen_movies)

# note that we are NOT formatting this in the Five Thirty Eight Style yet... we'll leave that to you to figure out
```

### 2.2 How people rate the 'Star Wars' movie? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genRateDf(inpf, eps, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    seen_df = inpf
    for seen_ep in ['seen_' + ep for ep in episodes]:
        seen_df = seen_df[seen_df[seen_ep].notna()]
        
    total = len(seen_df)
    percs = []

    for ep in eps:
        rank_ep = 'rank_' + ep
        perc = round(len(seen_df[(seen_df[rank_ep] == '1') | (seen_df[rank_ep] == '2')]) / total, 2)
        percs.append((names[ep], "Top third", perc))
        perc = round(len(seen_df[(seen_df[rank_ep] == '3') | (seen_df[rank_ep] == '4')]) / total, 2)
        percs.append((names[ep], "Middle third", perc))
        perc = round(len(seen_df[(seen_df[rank_ep] == '5') | (seen_df[rank_ep] == '6')]) / total, 2)
        percs.append((names[ep], "Bottom third", perc))
    
                                                              
    #tuples = list(zip([names[ep] for ep in eps],percs_top, percs_mid, percs
    rate_df = pd.DataFrame(percs, columns = ['Name', 'Tier', 'Percentage'])
    # print(rate_df)
    return(rate_df)

def augmentRateVis(inpf,eps, names=namesDict, names_l=namesList):

    tier_order = ['Top third', 'Middle third', 'Bottom third']
    inpf['Tier'] = pd.Categorical(inpf['Tier'], categories=tier_order) 
    bars = alt.Chart(inpf).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X( 'Percentage',
                axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis,
            'Name:N',
            axis=alt.Axis(tickCount=5, title=''),
            # we give the sorting order to avoid alphabetical order 
            sort=names_l
        ),
        color=alt.Color('Tier:N',
                       scale=alt.Scale(
                           domain=tier_order,
                           range=['#6d904f', '#008fd5', '#fc4f30']
                       ),
                       legend=None),
    )

    # Add text labels for percentages
    text = bars.mark_text(
        align='left',
        baseline='middle',
        font='Helvetica',
        fontSize=12,
        dx=3
    ).encode(
        text=alt.Text('Percentage:Q', format='.0%')
    )
    
    # Combine bars and text, then facet by Tier
    base = alt.layer(bars, text).properties(
        width=150,
        height=180
    )

    rate_movie = base.facet(
        column=alt.Column('Tier:N',
                         header=alt.Header(
                             title=None,
                             labelAngle=0,
                             labelAlign='center',
                             labelFont='Helvetica',
                             labelFontSize=14,
                             labelFontWeight='bold'
                         ),
                         sort=tier_order)
    ).resolve_scale(
        x='independent'
    ).configure(
        background='#f0f0f0', # Light gray background for whole chart (incl
    ).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        stroke=None,
        strokeWidth=0
    ).configure_axisY(
        domain=False,
        ticks=False,
        grid=False,
        labelFont='Helvetica',
        labelFontSize=14,
        labelColor='grey',
        title=None, # ensure no axis titles appear
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # add a title
        title={ "text": [
                "How People rate the Star Wars films"
            ],
            "subtitle": ["How often each film was rated in the top, middle and bottom third",
                         "(By 471 respondants who have seen all six movies)"],
            "font": "Helvetica",
            "fontSize": 24,
            "fontWeight": "bold",
            "anchor": "start",
            "subtitleFont": "Helvetica",
            "subtitleFontSize": 18,
            "subtitleColor": "grey"
        }
    ).configure_text(
        align='left',
        baseline='middle',
        font='Helvetica',
        fontSize=12,
        dx=3 # Nudges text to right so it doesn't appear on top of the bar
    )
    return rate_movie

def genRateVis(inpf,eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesD
    # input: names_l - a list of all the movies in series order (default: na
    # output: the Altair visualization as described above
    df = genRateDf(inpf, eps)
    rate_movies = augmentRateVis(df, eps)
    return rate_movies
```

### 2.3 Star Wars' Characters Favorability Ratings. Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genFavDf(inpf):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    characters = [
        "Luke Skywalker", "Han Solo", "Princess Leia Organa", "Obi Wan Kenobi", 
        "Yoda", "R2 D2", "C-3P0", "Anakin Skywalker", "Darth Vader",
        "Lando Calrissian", "Padme Amidala", "Boba Fett",
        "Emperor Palpatine", "Jar Jar Binks"
    ]
    
    percs = []
    for character in characters:
        # Get only respondents who answered about this character
        fav = inpf[inpf[character].notna()]
        total = len(fav)
        
        # Calculate favorable percentage
        perc = len(fav[(fav[character] == 'Somewhat favorably') | 
                       (fav[character] == 'Very favorably')]) / total
        percs.append((character, "Favorable", perc))
        
        # Calculate neutral percentage
        perc = len(fav[fav[character] == 'Neither favorably nor unfavorably (neutral)']) / total
        percs.append((character, "Neutral", perc))
        
        # Calculate unfavorable percentage
        perc = len(fav[(fav[character] == 'Somewhat unfavorably') | 
                       (fav[character] == 'Very unfavorably')]) / total
        percs.append((character, "Unfavorable", perc))
        
        # Calculate unfamiliar percentage
        perc = len(fav[fav[character] == 'Unfamiliar (N/A)']) / total
        percs.append((character, "Unfamiliar", perc))
    
    df = pd.DataFrame(percs, columns=['Name', 'Favorability', 'Percentage'])
    return df

def genFavorabilityVis(inpf):
    # input: inpf, the star wars dataset (as defined at top of file)
    # output: the Altair visualization as described above
    
    # YOUR CODE HERE
    # raise NotImplementedError()
    df = genFavDf(inpf)
    
    char_order = [
        "Luke Skywalker", "Han Solo", "Princess Leia Organa", "Obi Wan Kenobi",
        "Yoda", "R2 D2", "C-3P0", "Anakin Skywalker", "Darth Vader",
        "Lando Calrissian", "Padme Amidala", "Boba Fett",
        "Emperor Palpatine", "Jar Jar Binks"
    ]
    fav_order = ['Favorable', 'Neutral', 'Unfavorable', 'Unfamiliar']
    
    color_map = {
        'Favorable': '#6d904f',
        'Neutral': '#008fd5',
        'Unfavorable': '#fc4f30',
        'Unfamiliar': '#a0a0a0'
    }
    bars = alt.Chart(df).mark_bar(size=15).encode(
        x=alt.X('Percentage:Q', 
                axis=None),
        y=alt.Y('Name:N', 
                axis=alt.Axis(title=''),
                sort=char_order),
        color=alt.Color('Favorability:N',
                       scale=alt.Scale(
                           domain=list(color_map.keys()),
                           range=list(color_map.values())
                       ),
                       legend=None)
    )
    
    # Add text labels for percentages
    text = bars.mark_text(
        align='left',
        baseline='middle',
        font='Helvetica',
        fontSize=11,
        dx=3
    ).encode(
        text=alt.Text('Percentage:Q', format='.0%')
    )
    
    # Combine bars and text with height before faceting
    base = alt.layer(bars, text).properties(
        width=120,
        height=350,
        view=alt.ViewBackground(strokeWidth=0)
    )
    
    chart = base.facet(
        column=alt.Column('Favorability:N',
                         header=alt.Header(
                             title=None,
                             labelAngle=0,
                             labelAlign='center',
                             labelFont='Helvetica',
                             labelFontSize=12,
                             labelFontWeight='bold'
                         ),
                         sort=fav_order)
    ).resolve_scale(
        x='independent'
    ).properties(
        title={
            "text": "'Star Wars' Characters' Favorability Ratings",
            "subtitle": "By 834 respondents",
            "font": "Helvetica",
            "fontSize": 20,
            "fontWeight": "bold",
            "anchor": "start",
            "subtitleFont": "Helvetica",
            "subtitleFontSize": 14,
            "subtitleColor": "grey"
        }
    ).configure_view(
        strokeWidth=0
    ).configure(
        background='#f0f0f0'
    ).configure_facet(
        spacing=30
    ).configure_axisY(
        domain=False,
        ticks=False,
        labelFont='Helvetica',
        labelFontSize=11,
        labelColor='grey'
    )
    
    return chart
```

```python
# YOUR CODE HERE
# raise NotImplementedError()
# Good visualization
def genFanComparisonGood(inpf, eps, names=namesDict, names_l=namesList):
    """Compare how fans vs non-fans rank movies"""
    # Filter to people who have seen all 6 movies
    seen_all = inpf.copy()
    for ep in eps:
        seen_all = seen_all[seen_all[f'seen_{ep}'].notna()]
    
    # Calculate top-third rankings for fans vs non-fans
    data = []
    for ep in eps:
        for fan_status in ['Yes', 'No']:
            subset = seen_all[seen_all['fan'] == fan_status]
            if len(subset) > 0:
                count_top = len(subset[(subset[f'rank_{ep}'] == '1') | 
                                      (subset[f'rank_{ep}'] == '2')])
                perc = count_top / len(subset)
                data.append({
                    'Movie': names[ep],
                    'Fan Status': 'Fan' if fan_status == 'Yes' else 'Not a Fan',
                    'Percentage': perc
                })
    
    df = pd.DataFrame(data)
    
    # Create grouped bar chart
    chart = alt.Chart(df).mark_bar().encode(
        x=alt.X('Percentage:Q', 
                title='% Rating in Top Third',
                axis=alt.Axis(format='%')),
        y=alt.Y('Movie:N', 
                axis=alt.Axis(title=''),
                sort=names_l),
        color=alt.Color('Fan Status:N',
                       scale=alt.Scale(range=['#008fd5', '#fc4f30']),
                       legend=alt.Legend(title=None, orient='top')),
        xOffset='Fan Status:N'
    ).properties(
        width=500,
        height=300,
        title={
            "text": ["Movie Rankings: Fans vs Non-Fans"],
            "subtitle": ["Comparing top-third rankings between self-identified fans and non-fans"],
            "font": "Helvetica",
            "fontSize": 20,
            "fontWeight": "bold",
            "anchor": "start"
        }
    ).configure(
        background='#f0f0f0'
    ).configure_view(
        stroke=None
    ).configure_axisY(
        domain=False,
        ticks=False,
        labelFont='Helvetica',
        labelFontSize=12
    )
    
    return chart

genFanComparisonGood(sw, episodes)
```
