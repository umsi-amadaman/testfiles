When a pie chart is okay:
Use a pie chart for part to whole questions about one thing. For example, for one movie you can show Seen vs. Not seen. Keep the pie small: 2–4 slices, big differences, and the slices add to 100%. Use clear labels on each slice. A single pie gives a quick, simple picture. It is good when you want a gist, not exact comparison between many items.

Why a pie chart is not good here:
This chart compares six movies. We need to rank and compare the percentages. People read bar length and x-position more accurately than pie angles or areas. Also, the six percentages do not add to 100%, so one pie would be misleading. Many small pies (one per movie) make it hard to compare 68% vs 66%. Bars are better: they share a common baseline, sort easily, and are clearer for accessibility. Bottom line: keep bars for comparing categories; use a pie only for the composition of a single item.

```python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    # create the various mapping and order lists/dictionaries
    # return: list of episodes (in order), episode->name map, list of episode names (in order)
    episodes = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    names = {
        'EI' : 'The Phantom Menace', 'EII' : 'Attack of the Clones', 'EIII' : 'Revenge of the Sith', 
        'EIV': 'A New Hope', 'EV': 'The Empire Strikes Back', 'EVI' : 'The Return of the Jedi'
    }

    # we're also going to use this order to sort, so names_l will now have our sort order
    return episodes, names, [names[ep] for ep in episodes]

episodes, namesDict, namesList = genEpisodeNamesDF()
```

```python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
```

```python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN
    salo = indf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    return(salo)
```

```python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
```

```python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total = len(inpf)
    percs = []

    # loop over each column and calculate the number of people who have seen the movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count them
    # and divide by the percent
    for seen_ep in ['seen_' + ep for ep in episodes]:
        perc = len(inpf[~ pd.isna(inpf[seen_ep])]) / total
        percs.append(perc)

    # at this point percs is holding our percentages

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes],percs))
    seen_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    return(seen_per_df)
```

```python
seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
```

```python
def genPercentVis(indf, names_l=namesList):
    # input: indf, the dataframe as seen_per_df
    # input: names_l - a list of all the movies in series order (default namesList)
    # output: simple altair bar chart
    
    # ok, time to make the chart... let's make a bar chart (use mark_bar)
    bars = alt.Chart(indf).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        )
    )

    # at this point we don't really have a great plot (it's missing the annotations, titles, etc.)
    return(bars)

seenPerMovieVis = genPercentVis(seenPerMovieDF)
```

```python
def augmentPercentVis(base):
    # input: base (the base vis, i.e., bars as above)
    # we're going to overlay the text with the percentages, so let's make another visualization
    # that's just text labels


    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    seen_movies = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=180
    ).properties(
        # add a title
        title="Which 'Star Wars' Movies Have you Seen?"
    )

    return(seen_movies)

# note that we are NOT formatting this in the Five Thirty Eight Style yet... we'll leave that to you to figure out
```

### 2.2 How people rate the 'Star Wars' movie? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genRateVis(inpf,eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above

    df = inpf[inpf[[f'seen_{e}' for e in eps]].notna().all(axis=1)].copy()
    n = len(df)
    rank_cols = [f'rank_{e}' for e in eps]
    df[rank_cols] = df[rank_cols].apply(pd.to_numeric, errors='coerce')
    m = (df.melt(value_vars=rank_cols, var_name='ep', value_name='rank')
              .assign(Movie=lambda x: x['ep'].str.replace('rank_','',regex=False).map(names))
              .dropna(subset=['rank']))

    m['Group'] = pd.cut(m['rank'], bins=[0,2,4,6],
                           labels=['Top third','Middle third','Bottom third'],
                           right=True, include_lowest=True)

    pct = (m.groupby(['Movie','Group']).size().rename('n').reset_index())
    pct['Percent'] = pct['n'] / pct.groupby('Movie')['n'].transform('sum') * 100
    pct['PercentLabel'] = pct['Percent'].round().astype(int).astype(str) + '%'

    order = names_l

    base = alt.Chart(pct).encode(
        y=alt.Y('Movie:N', sort=order, title=None, axis=alt.Axis(labelColor='#6b6b6b', labelFontSize=12, labelPadding=0)),
        x=alt.X('Percent:Q', scale=alt.Scale(domain=[0,100]), axis=None)
    )
    bars = base.mark_bar(size=22).encode(
        color=alt.Color('Group:N', legend=None,
                        scale=alt.Scale(domain=['Top third','Middle third','Bottom third'],
                                        range=['#4caf50','#1e88e5','#e53935']))
    )
    labels = base.mark_text(align='left', baseline='middle', dx=4, color='black'
                           ).encode(text='PercentLabel:N')

    chart = (bars + labels).properties(width=100, height=190).facet(
        column=alt.Column('Group:N',title=None, sort=['Top third','Middle third','Bottom third'],
                header=alt.Header(    
                labelColor='black',
                titleColor='black',
                labelFontSize=15,
                titleFontSize=15))
    ).resolve_scale(x='independent').properties(
        background='white',
        title=alt.TitleParams(
            text="How People Rate the ‘Star Wars’ Movies",
        subtitle=[
            "How often each film was rated in the top, middle and bottom third",
            f"(by {n} respondents who have seen all six)"],
            anchor='start',
            fontSize=25,
            subtitleFontSize=17,
            color='black',      
            subtitleColor='black'   
        )
    ).configure_view(stroke=None).configure_axis(
        grid=False,
        domain=False,
        labelColor='black',     
        titleColor='black'        
    )

    return chart
```

```python
# Recreate this image using altair here (10 POINTS)

def genFavorabilityVis(inpf):
    # input: inpf, the star wars dataset (as defined at top of file)
    # output: the Altair visualization as described above
    
    FAV_COLS = [
        'Luke Skywalker','Han Solo','Princess Leia Organa','Obi Wan Kenobi','Yoda','R2 D2','C-3P0',
        'Anakin Skywalker','Darth Vader','Lando Calrissian','Padme Amidala','Boba Fett',
        'Emperor Palpatine','Jar Jar Binks'
    ]    

    VIEW_MAP = {
        'Very favorably': 'Favorable',
        'Somewhat favorably': 'Favorable',
        'Neither favorably nor unfavorably (neutral)': 'Neutral',
        'Somewhat unfavorably': 'Unfavorable',
        'Very unfavorably': 'Unfavorable',
        'Unfamiliar (N/A)': 'Unfamiliar'
    }

    GROUP_ORDER = ['Favorable','Neutral','Unfavorable','Unfamiliar']

    df = inpf[FAV_COLS].melt(var_name='Character', value_name='View')
    df = df.dropna(subset=['View']).copy()
    df['Group'] = df['View'].map(VIEW_MAP)
    df = df.dropna(subset=['Group'])

    count = (df.groupby(['Character','Group']).size()
           .rename('Count').reset_index())
    count['Percent'] = count['Count'] / count.groupby('Character')['Count'].transform('sum') * 100
    count['PercentLabel'] = count['Percent'].round().astype(int).astype(str) + '%'
    order = (count[count['Group']=='Favorable']
                 .sort_values('Percent', ascending=False)['Character'].tolist())

    base = alt.Chart(count).encode(
        y=alt.Y('Character:N', sort=order, title=None),
        x=alt.X('Percent:Q', title='Percent', scale=alt.Scale(domain=[0, 100]),
                axis=None)
    )

    bars = base.mark_bar().encode(
        color=alt.Color(
            'Group:N',
            scale=alt.Scale(domain=GROUP_ORDER,
                            range=['#4caf50','#1e88e5','#e53935','#9e9e9e']),
            legend=None
        )
    )
    labels = base.mark_text(align='left', baseline='middle', dx=3, color='black', fontWeight='bold').encode(
        text='PercentLabel:N'
    )

    layer = (bars + labels).properties(width=110, height=380)

    chart = layer.facet(
        column=alt.Column(
            'Group:N',
            title=None,
            sort=GROUP_ORDER,
            header=alt.Header(    
                labelColor='black',
                titleColor='black',
                labelFontSize=15,
                titleFontSize=15
            )
        )
    ).resolve_scale(
        x='independent').properties(background='white').properties(
        title=alt.TitleParams(
            text="‘Star Wars’ Characters Favorability Ratings",
            subtitle=f"By 834 respondents",
            anchor='start',
            fontSize=25,
            subtitleFontSize=17,
            color='black',      
            subtitleColor='black'   
        )
    ).configure_view(stroke=None).configure_scale(bandPaddingInner=0.2).configure_axis(
        grid=False,
        domain=False,
        labelColor='black',     
        titleColor='black'        
    )

    return chart
```

### 2.4 Who shot first? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genFirstShotVis(inpf):
    # input: inpf, the star wars dataset
    # output: the Altair visualization as described above
    
    df = inpf.copy()
    if 'seen_any_movie' in df.columns:
        df = df[df['seen_any_movie'].astype(str).str.strip().str.lower().eq('yes')]

    col = next(c for c in df.columns if 'shot first' in str(c).lower())
    order = ['Han', 'Greedo', "I don't understand this question"]

    m = (df[col].dropna()
         .value_counts()
         .reindex(order, fill_value=0)
         .rename_axis('answer')
         .reset_index(name='count'))
    n = int(m['count'].sum())
    m['pct'] = m['count'] / n * 100
    m['pct_label'] = m['pct'].round().astype(int).astype(str) + '%'

    bars = alt.Chart(m).mark_bar(size=26, color='#1e88e5').encode(
        y=alt.Y('answer:N', sort=order, title=None, axis=alt.Axis(labelFontSize=20, labelColor='black', labelPadding=6)),
        x=alt.X('pct:Q', scale=alt.Scale(domain=[0, 42], nice=False), axis=None)
    )

    labels = alt.Chart(m).mark_text(
        align='left', baseline='middle', dx=10,
        color='#000000', fontWeight='bold', fontSize=15
    ).encode(
        y=alt.Y('answer:N', sort=order),
        x='pct:Q',
        text='pct_label:N'
    )

    chart = (bars + labels).properties(
        width=600, height=200, background='white'
    ).properties(
        title=alt.TitleParams(
            text='Who Shot First?',
            subtitle=f'According to {n} respondents',
            anchor='start',
            fontSize=25,
            subtitleFontSize=17,
            color='black',      
            subtitleColor='black'   
        )
    ).configure_view(
        stroke=None
    ).configure_axis(
        grid=False,
        domain=False,
        labelColor='black',     
        titleColor='black'        
    )

    return chart
```

```python
def seen_by_gender_dumbbell(piv):
    piv2 = piv.copy()

    piv2[['Female','Male']] = piv2[['Female','Male']].astype(float)
    piv2 = piv2.dropna(subset=['Female','Male'])

    order = piv2['Movie'].tolist()
    piv2['Movie'] = pd.Categorical(piv2['Movie'], categories=order, ordered=True)
    piv2 = piv2.sort_values('Movie').reset_index(drop=True)

    y_enc = alt.Y(
        'Movie:N',
        title=None,
        sort=order,                        
        axis=alt.Axis(labelColor='black', labelFontSize=14, labelPadding=4)
    )
    x_f = alt.X('Female:Q', title=None, axis=None, scale=alt.Scale(domain=[50, 100]))
    x_m = alt.X('Male:Q',   title=None, axis=None, scale=alt.Scale(domain=[50, 100]))

    rule = (alt.Chart(piv2)
            .mark_rule(stroke='#AAA', strokeWidth=1.5)
            .encode(y=y_enc, x=x_f, x2=alt.X2('Male:Q')))

    female_pts = alt.Chart(piv2).mark_point(filled=True, size=100, color='#EF553B').encode(y=y_enc, x=x_f)
    male_pts   = alt.Chart(piv2).mark_point(filled=True, size=100, color='#636EFA').encode(y=y_enc, x=x_m)

    female_txt = (alt.Chart(piv2)
        .transform_calculate(lbl="isValid(datum.Female) ? format(datum.Female, '.0f') + '%' : ''")
        .mark_text(align='right', dx=-6, baseline='middle', color='#333', fontSize=15)
        .encode(y=y_enc, x=x_f, text='lbl:N'))

    male_txt = (alt.Chart(piv2)
        .transform_calculate(lbl="isValid(datum.Male) ? format(datum.Male, '.0f') + '%' : ''")
        .mark_text(align='left', dx=6, baseline='middle', color='#333', fontSize=15)
        .encode(y=y_enc, x=x_m, text='lbl:N'))

    gap_txt = (
        alt.Chart(piv2)
          .transform_calculate(
              gap="datum.Male - datum.Female",             
              mid="(datum.Male + datum.Female) / 2"         
          )
          .mark_text(dy=-8, color='#666', fontSize=12, fontWeight='bold')
          .encode(
              y=y_enc,
              x='mid:Q',
              text=alt.Text('gap:Q', format='+.0f')       
          )
    )

    legend = pd.DataFrame({'x':[94, 98], 'label':['Female', 'Male']})
    legend_pts = (alt.Chart(legend).mark_point(filled=True, size=80)
                  .encode(x=alt.X('x:Q', scale=alt.Scale(domain=[50, 100])),
                          y=alt.value(12),
                          color=alt.Color('label:N',
                                          scale=alt.Scale(domain=['Female','Male'],
                                                          range=['#EF553B','#636EFA']),
                                          legend=None)))
    legend_txt = (
        alt.Chart(legend)
          .mark_text(dy=-8, baseline='bottom', align='center', fontSize=12)
          .encode(
              x=alt.X('x:Q', scale=alt.Scale(domain=[50, 100])),
              y=alt.value(12),
              text='label:N',
              color=alt.Color('label:N',
                              scale=alt.Scale(domain=['Female','Male'], range=['#EF553B','#636EFA']),
                              legend=None)
          )
    )
    
    chart = (rule + female_pts + male_pts + female_txt + male_txt + gap_txt + legend_pts + legend_txt).properties(
        width=560, height=250,
        title=alt.TitleParams(
            text="Who has seen more ‘Star Wars’ film (by gender)",
            subtitle=(f"According to 828 respondents who have seen any film"),
            anchor='start', color='black', subtitleColor='black'
        ),
        background='white'
    ).configure_view(
        stroke=None
    ).configure_axis(
        grid=False, domain=False, labelColor='black', titleColor='black'
    ).configure_title(
        color='black',
        fontSize=22,
        subtitleColor='black',
        subtitleFontSize=14,
        anchor='start'
    )

    return chart


seen_by_gender_dumbbell(piv)
```
