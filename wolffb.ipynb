If I collapse the question into a single whole. Example: “Of all 835 people, what share have seen at least one Star Wars movie vs none?” A 2-slice pie works because slices sum to 100 and the message is composition, not ranking.

If I have a mutually exclusive set. Example: “Among respondents’ favorite films, what share picked each title?” if each person picks exactly one. A pie with a few slices can be fine for a quick gut check.

If I am only showing one snapshot and precision or close comparisons are not important. A small pie can communicate a big majority quickly.

```python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    # create the various mapping and order lists/dictionaries
    # return: list of episodes (in order), episode->name map, list of episode names (in order)
    episodes = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    names = {
        'EI' : 'The Phantom Menace', 'EII' : 'Attack of the Clones', 'EIII' : 'Revenge of the Sith', 
        'EIV': 'A New Hope', 'EV': 'The Empire Strikes Back', 'EVI' : 'The Return of the Jedi'
    }

    # we're also going to use this order to sort, so names_l will now have our sort order
    return episodes, names, [names[ep] for ep in episodes]

episodes, namesDict, namesList = genEpisodeNamesDF()
```

```python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
```

```python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN
    salo = indf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    return(salo)
```

```python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
```

```python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total = len(inpf)
    percs = []

    # loop over each column and calculate the number of people who have seen the movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count them
    # and divide by the percent
    for seen_ep in ['seen_' + ep for ep in episodes]:
        perc = len(inpf[~ pd.isna(inpf[seen_ep])]) / total
        percs.append(perc)

    # at this point percs is holding our percentages

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes],percs))
    seen_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    return(seen_per_df)
```

```python
seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
```

```python
# let's see what's inside
seenPerMovieDF
```

```python
def genPercentVis(indf, names_l=namesList):
    # input: indf, the dataframe as seen_per_df
    # input: names_l - a list of all the movies in series order (default namesList)
    # output: simple altair bar chart
    
    # ok, time to make the chart... let's make a bar chart (use mark_bar)
    bars = alt.Chart(indf).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        )
    )

    # at this point we don't really have a great plot (it's missing the annotations, titles, etc.)
    return(bars)

seenPerMovieVis = genPercentVis(seenPerMovieDF)
```

```python
def augmentPercentVis(base):
    # input: base (the base vis, i.e., bars as above)
    # we're going to overlay the text with the percentages, so let's make another visualization
    # that's just text labels


    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    seen_movies = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=180
    ).properties(
        # add a title
        title="Which 'Star Wars' Movies Have you Seen?"
    )

    return(seen_movies)

# note that we are NOT formatting this in the Five Thirty Eight Style yet... we'll leave that to you to figure out
```

```python
def genBestVis(inpf, eps, names=namesDict, names_l=namesList):
    seen_all = inpf.dropna(subset=[f"seen_{ep}" for ep in eps], how="any").copy()
    for ep in eps:
        col = f"rank_{ep}"
        if col in seen_all.columns:
            seen_all[col] = pd.to_numeric(seen_all[col], errors="coerce")
    denom = len(seen_all)
    rows = [(names[ep], (seen_all[f"rank_{ep}"] == 1).sum() / denom) for ep in eps]
    best_df = pd.DataFrame(rows, columns=["Name", "ShareFavorite"])
    best_df["Name"] = pd.Categorical(best_df["Name"], categories=names_l, ordered=True)
    best_df = best_df.sort_values("Name")
    max_pad = float(best_df["ShareFavorite"].max()) * 1.12
    base = alt.Chart(best_df).mark_bar(size=26).encode(
        x=alt.X("ShareFavorite:Q", axis=None, scale=alt.Scale(domain=[0, max_pad])),
        y=alt.Y("Name:N", sort=names_l, title="")
    )
    text = base.mark_text(align="left", baseline="middle", dx=6).encode(
        text=alt.Text("ShareFavorite:Q", format=".0%")
    )
    title_text = "What is the Best 'Star Wars' Movie?"
    subtitle_text = f"Of {denom} respondents who have seen all six films"
    chart = (base + text)\
        .configure_mark(color="#008fd5")\
        .configure_view(stroke=None, fill="#f0f0f0")\
        .configure(background="#f0f0f0")\
        .configure_axis(domain=False, grid=False, labelColor="#505050",
                        labelFont="Helvetica Neue", labelFontSize=13, title=None)\
        .configure_title(font="Helvetica Neue", fontSize=22, anchor="start",
                         color="#333333", subtitleColor="#666666",
                         subtitleFont="Helvetica Neue", subtitleFontSize=15)\
        .properties(width=640, height=260,
                    title=alt.TitleParams(text=title_text, subtitle=subtitle_text))
    return chart
```

### 2.2 How people rate the 'Star Wars' movie? Recreate the above image using altair (10 Points)

```python
def genRateVis(inpf, eps, names=namesDict, names_l=namesList):
    seen_all = inpf.dropna(subset=[f"seen_{ep}" for ep in eps], how="any").copy()
    for ep in eps:
        col = f"rank_{ep}"
        if col in seen_all.columns:
            seen_all[col] = pd.to_numeric(seen_all[col], errors="coerce")

    denom = len(seen_all)
    rows = []
    for ep in eps:
        r = seen_all[f"rank_{ep}"]
        rows.append((names[ep], "Top third", (r.isin([1, 2])).mean()))
        rows.append((names[ep], "Middle third", (r.isin([3, 4])).mean()))
        rows.append((names[ep], "Bottom third", (r.isin([5, 6])).mean()))
    df = pd.DataFrame(rows, columns=["Name", "Tier", "Share"])
    df["Name"] = pd.Categorical(df["Name"], categories=names_l, ordered=True)
    df["Tier"] = pd.Categorical(df["Tier"], categories=["Top third", "Middle third", "Bottom third"], ordered=True)

    def panel(tier, color, show_y):
        d = df[df["Tier"] == tier]
        mx = float(d["Share"].max()) * 1.12
        y_axis = alt.Axis(title="", domain=False, ticks=False, labelColor="#505050",
                          labelFont="Helvetica Neue", labelFontSize=13) if show_y else alt.Axis(labels=False, ticks=False, domain=False)
        base = alt.Chart(d).mark_bar(size=22, color=color).encode(
            x=alt.X("Share:Q", axis=None, scale=alt.Scale(domain=[0, mx])),
            y=alt.Y("Name:N", sort=names_l, axis=y_axis)
        )
        text = base.mark_text(align="left", baseline="middle", dx=6).encode(
            text=alt.Text("Share:Q", format=".0%")
        )
        return (base + text).properties(width=260, height=220, title=tier)

    left = panel("Top third", "#73b66b", True)
    mid = panel("Middle third", "#008fd5", False)
    right = panel("Bottom third", "#e45756", False)

    title_text = "How People Rate the 'Star Wars' Movies"
    subtitle_text = f"How often each film was rated in the top, middle and bottom third (by {denom} respondents who have seen all six films)"

    chart = alt.hconcat(left, mid, right, spacing=24)\
        .resolve_scale(y="shared")\
        .configure_view(stroke=None, fill="#f0f0f0")\
        .configure(background="#f0f0f0")\
        .configure_title(font="Helvetica Neue", fontSize=22, anchor="start",
                         color="#333333", subtitleColor="#666666",
                         subtitleFont="Helvetica Neue", subtitleFontSize=15)\
        .properties(title=alt.TitleParams(text=title_text, subtitle=subtitle_text))

    return chart
```

### 2.3 Star Wars' Characters Favorability Ratings. Recreate the above image using altair (10 Points)

```python
def genFavorabilityVis(inpf):
    favorable_vals = ["Very favorably", "Somewhat favorably"]
    neutral_vals = ["Neither favorably nor unfavorably (neutral)"]
    unfavorable_vals = ["Somewhat unfavorably", "Very unfavorably"]
    unfamiliar_vals = ["Unfamiliar (N/A)"]

    name_map = {
        "Luke Skywalker": "Luke Skywalker",
        "Han Solo": "Han Solo",
        "Princess Leia Organa": "Princess Leia Organa",
        "Obi Wan Kenobi": "Obi Wan Kenobi",
        "Yoda": "Yoda",
        "R2-D2": "R2 D2",
        "C-3PO": "C-3P0",  # dataset uses a zero
        "Anakin Skywalker": "Anakin Skywalker",
        "Darth Vader": "Darth Vader",
        "Lando Calrissian": "Lando Calrissian",
        "Padme Amidala": "Padme Amidala",
        "Boba Fett": "Boba Fett",
        "Emperor Palpatine": "Emperor Palpatine",
        "Jar Jar Binks": "Jar Jar Binks",
    }

    rows = []
    for disp, col in name_map.items():
        s = inpf[col].dropna()
        rows.append((disp, "Favorable",   s.isin(favorable_vals).mean()))
        rows.append((disp, "Neutral",     s.isin(neutral_vals).mean()))
        rows.append((disp, "Unfavorable", s.isin(unfavorable_vals).mean()))
        rows.append((disp, "Unfamiliar",  s.isin(unfamiliar_vals).mean()))

    df = pd.DataFrame(rows, columns=["Name", "Category", "Share"])

    fav_order = (
        df[df["Category"] == "Favorable"]
        .sort_values("Share", ascending=False)["Name"]
        .tolist()
    )
    df["Name"] = pd.Categorical(df["Name"], categories=fav_order, ordered=True)
    df["Category"] = pd.Categorical(
        df["Category"],
        categories=["Favorable", "Neutral", "Unfavorable", "Unfamiliar"],
        ordered=True
    )

    def panel(cat, color, show_y):
        d = df[df["Category"] == cat]
        mx = float(d["Share"].max()) * 1.12
        y_axis = alt.Axis(
            title="", domain=False, ticks=False,
            labelColor="#505050", labelFont="Helvetica Neue", labelFontSize=13
        ) if show_y else alt.Axis(title="", labels=False, ticks=False, domain=False)
        base = alt.Chart(d).mark_bar(size=18, color=color).encode(
            x=alt.X("Share:Q", axis=None, scale=alt.Scale(domain=[0, mx])),
            y=alt.Y("Name:N", sort=fav_order, axis=y_axis)
        )
        txt = base.mark_text(align="left", baseline="middle", dx=6).encode(
            text=alt.Text("Share:Q", format=".0%")
        )
        return (base + txt).properties(width=230, height=420, title=cat)

    left  = panel("Favorable",   "#73b66b", True)
    mid   = panel("Neutral",     "#008fd5", False)
    right = panel("Unfavorable", "#e45756", False)
    gray  = panel("Unfamiliar",  "#9e9e9e", False)

    title_text = "'Star Wars' Characters Favorability Ratings"
    subtitle_text = "Percent of respondents who answered each character item"

    chart = alt.hconcat(left, mid, right, gray, spacing=24)\
        .resolve_scale(y="shared")\
        .configure_view(stroke=None, fill="#f0f0f0")\
        .configure(background="#f0f0f0")\
        .configure_title(font="Helvetica Neue", fontSize=22, anchor="start",
                         color="#333333", subtitleColor="#666666",
                         subtitleFont="Helvetica Neue", subtitleFontSize=15)\
        .properties(title=alt.TitleParams(text=title_text, subtitle=subtitle_text))
    return chart
```

### 2.4 Who shot first? Recreate the above image using altair (10 Points)

```python
def genFirstShotVis(inpf):
    col = "Which character shot first?"
    s = inpf[col].dropna()
    denom = len(s)

    label_map = {
        "Han": "Han",
        "Greedo": "Greedo",
        "I don't understand this question": "I don't understand\nthis question",
    }

    rows = [(disp, (s == raw).mean()) for raw, disp in label_map.items()]
    df = pd.DataFrame(rows, columns=["Choice", "Share"])
    order = ["Han", "Greedo", "I don't understand\nthis question"]
    df["Choice"] = pd.Categorical(df["Choice"], categories=order, ordered=True)

    max_pad = float(df["Share"].max()) * 1.12
    base = alt.Chart(df).mark_bar(size=26, color="#008fd5").encode(
        x=alt.X("Share:Q", axis=None, scale=alt.Scale(domain=[0, max_pad])),
        y=alt.Y("Choice:N", sort=order, title="")
    )
    text = base.mark_text(align="left", baseline="middle", dx=6).encode(
        text=alt.Text("Share:Q", format=".0%")
    )

    title_text = "Who Shot First?"
    subtitle_text = f"According to {denom} respondents"

    chart = (base + text)\
        .configure_view(stroke=None, fill="#f0f0f0")\
        .configure(background="#f0f0f0")\
        .configure_axis(domain=False, grid=False, labelColor="#505050",
                        labelFont="Helvetica Neue", labelFontSize=13, title=None)\
        .configure_title(font="Helvetica Neue", fontSize=22, anchor="start",
                         color="#333333", subtitleColor="#666666",
                         subtitleFont="Helvetica Neue", subtitleFontSize=15)\
        .properties(width=700, height=220,
                    title=alt.TitleParams(text=title_text, subtitle=subtitle_text))
    return chart
```

Reading through the article, identify some aspect of the story that would benefit from an extra visualization (you can imagine extending the article if it helps). Note: The two visualizations should use different marks and at least some of the encodings should be different. If you're struggling with coming up with two really different representations, your visualization may simply be too simple. Try to add additional columns/comparisons. 

```python
def genGenderAgeEngagement_Good_v3(inpf):
    age_order = ["18-29","30-44","45-60","> 60"]
    df = inpf[inpf["Gender"].isin(["Male","Female"])].copy()
    df = df[df["Age"].isin(age_order)]

    seen = (
        df.dropna(subset=["seen_any_movie"])
          .groupby(["Age","Gender"], as_index=False)
          .agg(Share=("seen_any_movie", lambda s: (s=="Yes").mean()))
    )
    seen["Metric"] = "Seen any movie"

    fan_base = df[df["seen_any_movie"]=="Yes"]
    fan = (
        fan_base.dropna(subset=["fan"])
                .groupby(["Age","Gender"], as_index=False)
                .agg(Share=("fan", lambda s: (s=="Yes").mean()))
    )
    fan["Metric"] = "Percent fans among viewers"

    data = pd.concat([seen, fan], ignore_index=True)
    data["Age"] = pd.Categorical(data["Age"], categories=age_order, ordered=True)
    data["Gender"] = pd.Categorical(data["Gender"], categories=["Female","Male"], ordered=True)

    x_scale = alt.Scale(domain=[0,1])
    x_axis = alt.Axis(format="%", grid=True, tickCount=5, title=None)

    base = alt.Chart(data)

    connectors = base.transform_aggregate(
        minShare="min(Share)", maxShare="max(Share)", groupby=["Metric","Age"]
    ).mark_rule(stroke="#bdbdbd", strokeWidth=2).encode(
        y=alt.Y("Age:N", sort=age_order, title=""),
        x=alt.X("minShare:Q", scale=x_scale, axis=None),
        x2="maxShare:Q"
    )

    points = base.mark_point(filled=True, size=100).encode(
        y=alt.Y("Age:N", sort=age_order, title=""),
        x=alt.X("Share:Q", scale=x_scale, axis=x_axis),
        color=alt.Color("Gender:N",
                        scale=alt.Scale(domain=["Female","Male"], range=["#e45756","#008fd5"]),
                        legend=alt.Legend(orient="top", title="Gender")),
        tooltip=["Metric","Age","Gender",alt.Tooltip("Share:Q", format=".0%")]
    )

    male_text = base.transform_filter(alt.datum.Gender == "Male")\
        .mark_text(align="left", baseline="middle", dx=6).encode(
            y=alt.Y("Age:N", sort=age_order), x=alt.X("Share:Q", scale=x_scale),
            text=alt.Text("Share:Q", format=".0%"), color=alt.value("#404040")
        )

    female_text = base.transform_filter(alt.datum.Gender == "Female")\
        .mark_text(align="right", baseline="middle", dx=-6).encode(
            y=alt.Y("Age:N", sort=age_order), x=alt.X("Share:Q", scale=x_scale),
            text=alt.Text("Share:Q", format=".0%"), color=alt.value("#404040")
        )

    gap_label = base.transform_aggregate(
        minShare="min(Share)", maxShare="max(Share)", groupby=["Metric","Age"]
    ).transform_calculate(
        mid="(datum.minShare + datum.maxShare)/2",
        gap_pp="round((datum.maxShare - datum.minShare)*100)"
    ).mark_text(baseline="bottom", dy=-6, fontWeight="bold", color="#666666").encode(
        y=alt.Y("Age:N", sort=age_order, title=""),
        x=alt.X("mid:Q", scale=x_scale),
        text="gap_pp:N"
    )

    panel = (connectors + points + male_text + female_text + gap_label).properties(width=560, height=210)

    title = "Engagement gap by gender and age"
    subtitle = "Share who have seen any movie, and percent fans among viewers"

    chart = panel.facet(
        row=alt.Row("Metric:N",
                    sort=["Seen any movie","Percent fans among viewers"],
                    header=alt.Header(labelFont="Helvetica Neue", labelFontSize=16, labelColor="#333"))
    ).resolve_scale(x="shared", y="shared")\
     .configure_view(stroke=None, fill="#f0f0f0")\
     .configure(background="#f0f0f0")\
     .configure_axis(domain=False, gridColor="#e0e0e0",
                     labelColor="#505050", labelFont="Helvetica Neue", labelFontSize=13, title=None)\
     .configure_title(font="Helvetica Neue", fontSize=22, anchor="start",
                      color="#333333", subtitleColor="#666666",
                      subtitleFont="Helvetica Neue", subtitleFontSize=15)\
     .properties(title=alt.TitleParams(text=title, subtitle=subtitle))
    return chart
```

```python
# Version B (contrast): Small-multiple pies for "Seen any movie" by gender and age
# Uses angles/areas and many panels, which makes comparisons harder

def genGenderAgeSeen_Pies(inpf):
    age_order = ["18-29","30-44","45-60","> 60"]
    df = inpf.copy()
    df = df[df["Gender"].isin(["Male","Female"])]
    df = df[df["Age"].isin(age_order)]

    # build yes/no shares
    grp = (df.dropna(subset=["seen_any_movie"])
             .groupby(["Gender","Age"], as_index=False)["seen_any_movie"]
             .value_counts(normalize=True))
    grp.rename(columns={"proportion":"Share","seen_any_movie":"Response"}, inplace=True)
    grp["Age"] = pd.Categorical(grp["Age"], categories=age_order, ordered=True)
    grp["Response"] = pd.Categorical(grp["Response"], categories=["Yes","No"], ordered=True)

    pie = alt.Chart(grp).mark_arc(outerRadius=70).encode(
        theta=alt.Theta("Share:Q"),
        color=alt.Color("Response:N",
                        scale=alt.Scale(domain=["Yes","No"], range=["#008fd5","#9e9e9e"]),
                        legend=alt.Legend(orient="bottom", title=None)),
        tooltip=["Gender","Age","Response",alt.Tooltip("Share:Q", format=".0%")]
    ).properties(width=170, height=170)

    title = "Seen any Star Wars movie? by gender and age (pie multiples)"

    chart = pie.facet(
        column=alt.Column("Age:N", sort=age_order,
                          header=alt.Header(title="Age", labelFont="Helvetica Neue", labelFontSize=14)),
        row=alt.Row("Gender:N", header=alt.Header(title=None, labelFont="Helvetica Neue", labelFontSize=14))
    ).configure_view(stroke=None, fill="#f0f0f0")\
     .configure(background="#f0f0f0")\
     .configure_title(font="Helvetica Neue", fontSize=22, anchor="start", color="#333333")\
     .properties(title=title)
    return chart


genGenderAgeSeen_Pies(sw)
```

#### Justification
*Provide your justification here* to argue why your "better" version is actually better. Make sure to defend this based on expressiveness and effectiveness as well as perceptual arguments.

My dumbbell version beats the pie multiples on both expressiveness and effectiveness. The dumbbells use position on a shared 0–100% axis, the most accurate channel, and the connector literally encodes the gender gap as a length you can read fast. Color only maps gender, so the encoding matches the data and does not imply extra parts of a whole. The pies ask you to compare angles and areas across many small panels, which is slow and imprecise, and they suggest composition when the real task is comparing rates and gaps. With the dumbbells, a light grid and direct percent labels reduce guesswork, and the small multiples share scales so I can compare “Seen any movie” vs “Percent fans among viewers” without re-learning the axis. The result is cleaner, more comparable, and closer to best practice.
