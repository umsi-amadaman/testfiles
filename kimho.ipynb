```python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    # create the various mapping and order lists/dictionaries
    # return: list of episodes (in order), episode->name map, list of episode names (in order)
    episodes = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    names = {
        'EI' : 'The Phantom Menace', 'EII' : 'Attack of the Clones', 'EIII' : 'Revenge of the Sith', 
        'EIV': 'A New Hope', 'EV': 'The Empire Strikes Back', 'EVI' : 'The Return of the Jedi'
    }

    # we're also going to use this order to sort, so names_l will now have our sort order
    return episodes, names, [names[ep] for ep in episodes]

episodes, namesDict, namesList = genEpisodeNamesDF()
```

```python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
```

```python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN
    salo = indf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    return(salo)
```

```python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
```

```python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total = len(inpf)
    percs = []

    # loop over each column and calculate the number of people who have seen the movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count them
    # and divide by the percent
    for seen_ep in ['seen_' + ep for ep in episodes]:
        perc = len(inpf[~ pd.isna(inpf[seen_ep])]) / total
        percs.append(perc)

    # at this point percs is holding our percentages

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes],percs))
    seen_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    return(seen_per_df)
```

```python
seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
```

```python
# let's see what's inside
seenPerMovieDF
```

```python
def genPercentVis(indf, names_l=namesList):
    # input: indf, the dataframe as seen_per_df
    # input: names_l - a list of all the movies in series order (default namesList)
    # output: simple altair bar chart
    
    # ok, time to make the chart... let's make a bar chart (use mark_bar)
    bars = alt.Chart(indf).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        )
    )

    # at this point we don't really have a great plot (it's missing the annotations, titles, etc.)
    return(bars)

seenPerMovieVis = genPercentVis(seenPerMovieDF)
```

```python
def augmentPercentVis(base):
    # input: base (the base vis, i.e., bars as above)
    # we're going to overlay the text with the percentages, so let's make another visualization
    # that's just text labels


    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    seen_movies = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=180
    ).properties(
        # add a title
        title="Which 'Star Wars' Movies Have you Seen?"
    )

    return(seen_movies)

# note that we are NOT formatting this in the Five Thirty Eight Style yet... we'll leave that to you to figure out
```

### 2.2 How people rate the 'Star Wars' movie? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genRateVis(inpf,eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above

    #creating the dataset of only the people who saw all the movies
    seen_all = inpf.dropna(subset=['seen_' + ep for ep in eps],how='any')
    total = len(seen_all)
    tops = []
    middles = []
    bottoms = []

    # loop over each movie column and calculate the number of people who ranked it top, middle or bottom
    # count them and divide by the percent
    for fav_ep in ['rank_' + ep for ep in episodes]:
        top = len(seen_all[seen_all[fav_ep].isin(['1','2'])]) / total
        middle = len(seen_all[seen_all[fav_ep].isin(['3','4'])]) / total
        bottom = len(seen_all[seen_all[fav_ep].isin(['5','6'])]) / total
        tops.append(top)
        middles.append(middle)
        bottoms.append(bottom)

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    #adding an additional column to help facet the chart by rankings
    toples = list(zip([names[ep] for ep in episodes],tops))
    top = pd.DataFrame(toples, columns = ['Name', 'Percentage'])
    top['rank']='Top third'
    midples = list(zip([names[ep] for ep in episodes],middles))
    mid = pd.DataFrame(midples, columns = ['Name', 'Percentage'])
    mid['rank']='Middle third'
    botples = list(zip([names[ep] for ep in episodes],bottoms))
    bot = pd.DataFrame(botples, columns = ['Name', 'Percentage'])
    bot['rank']='Bottom third'
    #Concatenating all the dataframes created into one for the visualization
    ranking_df = pd.concat([top,mid,bot],axis=0)
    base = alt.Chart(
        ranking_df,
        ).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        #tweaking the size of the bars to look more similar to the original
        x=alt.X(
            'Percentage',
            axis=None,
            scale=alt.Scale(rangeMax=50)
        ),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Name:N',
             axis=alt.Axis(tickCount=5, title='',labelFontWeight='bold',labelColor='black',gridOpacity=0),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        ),
    )
    bars_rank= base.mark_bar().encode(
        color=alt.Color('rank:N',legend=None,title=None).scale(scheme='set1')
    )
    text_rank = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    ranked_movies = (text_rank + bars_rank).properties(
        # set the dimensions of the visualization
        width=160,
        height=180
    ).facet(
        #faceting the chart so each ranking is in a column but maintaining bar scales across
        column=alt.Column('rank:O',title=None,sort=['Top third','Middle third','Bottom third'],header=alt.Header(labelFontSize=20,labelFontWeight='bold'))
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).configure_axis(domainOpacity=0,tickOpacity=0).properties(title=alt.Title(
            "How People Rate the 'Star Wars' Movies",
            subtitle=["How often each film was rated in the top, middle, and bottom third", "(by 471 respondents who have seen all six films)"],
            subtitleFontSize=19
        ))
    alt.theme.enable('fivethirtyeight')
    return ranked_movies
```

```python
# Recreate this image using altair here (10 POINTS)

def genFavorabilityVis(inpf):
    # input: inpf, the star wars dataset (as defined at top of file)
    # output: the Altair visualization as described above

    # loop over each character column and calculate the number of people for each rating bucketing them into one of the four,
    # specifically, filter out the people who are *NaN* for a specific character, count  them
    # and divide by the percent
    favors = []
    neutrals = []
    unfavors = []
    unfam = []
    for favor in inpf.columns[-23:-9]:
        current_frame = inpf.dropna(subset = [favor])
        total = len(current_frame)
        favorable = len(current_frame[current_frame[favor].isin(['Very favorably','Somewhat favorably'])]) / total
        neutral = len(current_frame[current_frame[favor] == 'Neither favorably nor unfavorably (neutral)']) / total
        unfavorable = len(current_frame[current_frame[favor].isin(['Somewhat unfavorably','Very unfavorably'])]) / total
        unknown = len(current_frame[current_frame[favor] == 'Unfamiliar (N/A)']) / total
        favors.append(favorable)
        neutrals.append(neutral)
        unfavors.append(unfavorable)
        unfam.append(unknown)
        
    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    #additionally adding a column that marks which 'rating' this percentage is for
    tuples = list(zip(inpf.columns[-23:-9],favors))
    favor = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    favor['rating']='Favorable'
    #creating a master sort order based who people liked the most
    sort_order = favor.sort_values(by='Percentage',ascending=False)['Name']
    tuples = list(zip(inpf.columns[-23:-9],neutrals))
    neutral = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    neutral['rating']='Neutral'
    tuples = list(zip(inpf.columns[-23:-9],unfavors))
    unfavor = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    unfavor['rating']='Unfavorable'
    tuples = list(zip(inpf.columns[-23:-9],unfam))
    unfam = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    unfam['rating']='Unfamiliar'
    #Concatenating all the dataframes created into one for the visualization
    rating_df = pd.concat([favor,neutral,unfavor,unfam],axis=0)
    
    #Setting the Colors directly by facet
    domain = ['Favorable','Neutral','Unfavorable','Unfamiliar']
    range_ = ['#4da74a','#008fd5','#ff2600','#999999']
    
    base = alt.Chart(
        rating_df,
        ).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        #setting rangeMax to look closer to original
        x=alt.X(
            'Percentage',
            axis=None,
            scale=alt.Scale(rangeMax=90)
        ),
        y=alt.Y(
            # encode y using the name, use the character name to label the axis, sort using the sort order from above
            'Name:N',
             axis=alt.Axis(tickCount=5, title='',labelFontWeight='bold',labelColor='black',gridOpacity=0),
            sort=sort_order
        ),
    )
    #seperately adding the color to the bars so the labels aren't affected
    bars_rated= base.mark_bar().encode(
        color=alt.Color('rating:N',legend=None,title=None,scale=alt.Scale(domain=domain,range=range_))
    )
    text_rank = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    rated_chars = (text_rank + bars_rated).properties(
        # set the dimensions of the visualization
        width=100,
        height=300
    # faceting on the rating column and setting an explicit sort order
    #tweaking settings to look more like the original
    ).facet(
        column=alt.Column('rating:O',title=None,sort=['Favorable','Neutral','Unfavorable','Unfamiliar'],
                          header=alt.Header(labelFontSize=15,labelFontWeight='bold',titleAlign='left',titleAnchor='start'))
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).configure_axis(domainOpacity=0,tickOpacity=0).properties(title=alt.Title(
            "'Star Wars' Characters Favorability Ratings",
            subtitle=["By 834 respondents"],
            subtitleFontSize=15
        ))
    alt.theme.enable('fivethirtyeight')
    
    return rated_chars
```

### 2.4 Who shot first? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genFirstShotVis(inpf):
    # input: inpf, the star wars dataset
    # output: the Altair visualization as described above
    
    #Filtering for only people who answered the question
    df = inpf.dropna(subset='Which character shot first?')
    #Grouping by responses and then finding the percentage of the total by dividing by the total respondents
    table = (df.groupby('Which character shot first?').agg('size') / len(df)).reset_index(). \
    rename(columns={0:'Percentage'}).replace("I don't understand this question","I don't understand_this question") #tweaking the name of don't understand so that the line break happens
    base = alt.Chart(
        table,
        title=alt.Title(
            "Who Shot First?",
            subtitle="According to 834 respondents",
            subtitleFontSize=19
        )).mark_bar(size=20
                   ).encode(
                            # encode x as the percent, and hide the axis
                            x=alt.X(
                                'Percentage',
                                axis=None),
                            y=alt.Y(
                                # encode y using the response
                                'Which character shot first?:N',
                                 axis=alt.Axis(tickCount=5, title='',labelColor='black',gridOpacity=0,labelFontSize=14),
                                 sort=["Han","Greedo","I don't understand this question"]
                            )
    )
    shoot_text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    shoot = (shoot_text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=1
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height= 100
    ).configure_axisY(labelExpr='split(datum.label, "_")')
    shoot = shoot.configure_axis(domainOpacity=0,tickOpacity=0)
    alt.theme.enable('fivethirtyeight')
    return shoot
```
