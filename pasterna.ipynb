```python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    # create the various mapping and order lists/dictionaries
    # return: list of episodes (in order), episode->name map, list of episode names (in order)
    episodes = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    names = {
        'EI' : 'The Phantom Menace', 'EII' : 'Attack of the Clones', 'EIII' : 'Revenge of the Sith', 
        'EIV': 'A New Hope', 'EV': 'The Empire Strikes Back', 'EVI' : 'The Return of the Jedi'
    }

    # we're also going to use this order to sort, so names_l will now have our sort order
    return episodes, names, [names[ep] for ep in episodes]

episodes, namesDict, namesList = genEpisodeNamesDF()
```

```python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
```

```python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN
    salo = indf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    return(salo)
```

```python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
```

```python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total = len(inpf)
    percs = []

    # loop over each column and calculate the number of people who have seen the movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count them
    # and divide by the percent
    for seen_ep in ['seen_' + ep for ep in episodes]:
        perc = len(inpf[~ pd.isna(inpf[seen_ep])]) / total
        percs.append(perc)

    # at this point percs is holding our percentages

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes],percs))
    seen_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    return(seen_per_df)
```

```python
seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
```

```python
# let's see what's inside
seenPerMovieDF
```

```python
def genPercentVis(indf, names_l=namesList):
    # input: indf, the dataframe as seen_per_df
    # input: names_l - a list of all the movies in series order (default namesList)
    # output: simple altair bar chart
    
    # ok, time to make the chart... let's make a bar chart (use mark_bar)
    bars = alt.Chart(indf).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        )
    )

    # at this point we don't really have a great plot (it's missing the annotations, titles, etc.)
    return(bars)

seenPerMovieVis = genPercentVis(seenPerMovieDF)
```

```python
def augmentPercentVis(base):
    # input: base (the base vis, i.e., bars as above)
    # we're going to overlay the text with the percentages, so let's make another visualization
    # that's just text labels


    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    seen_movies = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=180
    ).properties(
        # add a title
        title="Which 'Star Wars' Movies Have you Seen?"
    )

    return(seen_movies)

# note that we are NOT formatting this in the Five Thirty Eight Style yet... we'll leave that to you to figure out
```

```python

def genBestVis(inpf, eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above

    seen_cols = ['seen_EI','seen_EII','seen_EIII','seen_EIV','seen_EV','seen_EVI']
    df = inpf.dropna(subset=seen_cols).copy()
    
    rank_cols = ['rank_EI','rank_EII','rank_EIII','rank_EIV','rank_EV','rank_EVI']
    df.loc[:, rank_cols] = df[rank_cols].apply(pd.to_numeric, errors='coerce')
    
    df = df.rename(columns={
        'rank_EI':'EI','rank_EII':'EII','rank_EIII':'EIII',
        'rank_EIV':'EIV','rank_EV':'EV','rank_EVI':'EVI'
    })
    
    long_df = df.melt(value_vars=eps, var_name='Episode', value_name='Rank')
    long_df['Movie'] = long_df['Episode'].map(names)
    
    top1_counts = long_df[long_df['Rank'] == 1].groupby('Movie').size()
    total_respondents = len(df)
    top1_pct = (top1_counts / total_respondents).reset_index(name='Percentage')
    
    base = alt.Chart(top1_pct).mark_bar().encode(
        y=alt.Y('Movie:N',
                sort=names_l, 
                axis=alt.Axis(title=None, labelFontSize=12, labelColor='black')),
        x=alt.X('Percentage:Q',
                title=None,
                axis=alt.Axis(grid=False, labels = False, ticks = False))
    )
    
    text = alt.Chart(top1_pct).mark_text(
        align='left',
        baseline='middle',
        dx=3,
        color='black'
    ).encode(
        y=alt.Y('Movie:N', sort=names_l), 
        x='Percentage:Q',
        text=alt.Text('Percentage:Q', format='.0%')
    )
    

    best_movies_chart = (base + text).configure_mark(
       color='#008fd5'
    ).configure_view(
        strokeWidth=0   
    ).configure_axis(
        domain=False,    
        grid=False 
    ).configure_scale(
        bandPaddingInner=0.2
    ).configure(background='#F2F2F2'
    ).properties(
        width=500,
        height=180,
        title=alt.TitleParams(
            text="What's the best 'Star Wars' movie?",
            subtitle="Of 471 respondents who have seen all six films",
            color='black',
            subtitleColor='black',
            fontSize=20,
            subtitleFontSize=15,
            anchor='start'
        )
    )
    
    return best_movies_chart
```

### 2.2 How people rate the 'Star Wars' movie? Recreate the above image using altair (10 Points)

```python
def genRateVis(inpf, eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above
    

    seen_cols = ['seen_EI', 'seen_EII', 'seen_EIII', 'seen_EIV', 'seen_EV', 'seen_EVI']
    df = inpf.dropna(subset=seen_cols).copy()


    rank_cols = ['rank_EI', 'rank_EII', 'rank_EIII', 'rank_EIV', 'rank_EV', 'rank_EVI']
    df[rank_cols] = df[rank_cols].apply(pd.to_numeric, errors='coerce')

  
    df = df.rename(columns={
        'rank_EI': 'EI', 'rank_EII': 'EII', 'rank_EIII': 'EIII',
        'rank_EIV': 'EIV', 'rank_EV': 'EV', 'rank_EVI': 'EVI'
    })


    long_df = df.melt(value_vars=eps, var_name='Episode', value_name='Rank')
    long_df['Movie'] = long_df['Episode'].map(names)

    def to_third(rank):
        if rank in [1, 2]:
            return 'Top third'
        elif rank in [3, 4]:
            return 'Middle third'
        elif rank in [5, 6]:
            return 'Bottom third'
        return None

    long_df['Category'] = long_df['Rank'].apply(to_third)


    counts = long_df.groupby(['Movie', 'Category']).size().reset_index(name='Count')
    total = df.shape[0]
    counts['Percentage'] = (counts['Count'] / total)

    idx = pd.MultiIndex.from_product(
        [names_l, ['Top third', 'Middle third', 'Bottom third']],
        names=['Movie', 'Category']
    )
    counts = counts.set_index(['Movie', 'Category']).reindex(idx, fill_value=0).reset_index()

    color_scale = alt.Scale(
        domain=['Top third', 'Middle third', 'Bottom third'],
        range=['#74BA3C', '#3C99DE', '#F23030']
    )

    bars = alt.Chart(counts).mark_bar(size=14).encode(
        y=alt.Y('Movie:N', sort=names_l, title=None, axis=alt.Axis(labelPadding=10)),
        x=alt.X('Percentage:Q', title=None, axis=None),
        color=alt.Color('Category:N', scale=color_scale, legend=None)
    ).properties(
        width=120, 
        height=180
    )


    y_axis = alt.Y(
        'Movie:N',
        sort=names_l,
        title=None,
        axis=alt.Axis(
            labelFontSize=12,
            labelColor='black',
            labelPadding=10
        )
    )
    text = alt.Chart(counts).mark_text(
        align='left',
        baseline='middle',
        dx=4,
        fontSize=12,
        color='black'
    ).encode(
        y=y_axis,
        x=alt.X('Percentage:Q'),
        text=alt.Text('Percentage:Q', format='.0%')
    ).transform_calculate(
        text="datum.Percentage + '%'"
    ).properties(
        width=120,   
        height=180
    )

    # Combine and facet
    chart = (bars + text).facet(
        column=alt.Column(
            'Category:N',
            sort=['Top third', 'Middle third', 'Bottom third'],
            header=alt.Header(
                labelOrient='top',
                labelFontSize=14,
                labelFontWeight='bold',
                labelColor= 'black',
                title=None
            )
    )
    ).properties(
        title=alt.TitleParams(
            text="How People Rate the ‘Star Wars’ Movies",
            subtitle=["How often each film was rated in the top, middle and bottom third",
                      "(by 471 respondents who have seen all six films)"],
            anchor='start',
            fontSize=20,
            subtitleFontSize=13,
            subtitleColor='black'
        ),
        padding={"left": 10, "top": 10, "right": 10, "bottom": 10}
    ).configure_axis(
        grid=False,
        labelFontSize=13,
        labelColor='black'
    ).configure_view(
        strokeWidth=0
    ).configure_facet(
        spacing=10
    ).configure(background='#F2F2F2'
    ).configure_title(
        anchor='start',
        color='black'
    )
    return chart
```

```python
# Recreate this image using altair here (10 POINTS)

def genFavorabilityVis(inpf):
    # input: inpf, the star wars dataset (as defined at top of file)
    # output: the Altair visualization as described above
    chara_cols = ['Luke Skywalker', 'Han Solo', 'Princess Leia Organa', 'Obi Wan Kenobi',
                  'Yoda', 'R2 D2', 'C-3P0', 'Anakin Skywalker', 'Darth Vader', 'Lando Calrissian',
                  'Padme Amidala', 'Boba Fett', 'Emperor Palpatine', 'Jar Jar Binks']
    df = inpf.dropna(subset= chara_cols).copy()

    df_long = df[chara_cols].melt(var_name='Character', value_name='Favorability')

    def simplify_favorability(val):
        if val in ['Very favorably', 'Somewhat favorably']:
            return 'Favorable'
        elif val in ['Very unfavorably', 'Somewhat unfavorably']:
            return 'Unfavorable'
        elif val == 'Neither favorably nor unfavorably (neutral)':
            return 'Neutral'
        else:

            return 'Unfamiliar'
    
    df_long['FavorabilitySimple'] = df_long['Favorability'].apply(simplify_favorability)
    
    
    counts = df_long.groupby(['Character', 'FavorabilitySimple']).size().reset_index(name='Count')
    
    
    total_per_char = counts.groupby('Character')['Count'].transform('sum')
    counts['Proportion'] = counts['Count'] / total_per_char
    
    
    favorability_order = ['Favorable', 'Neutral', 'Unfavorable', 'Unfamiliar']

    color_scale = alt.Scale(
        domain=['Favorable', 'Neutral', 'Unfavorable', 'Unfamiliar'],
        range=['#74BA3C', '#3C99DE', '#F23030', '#ABABAB']
    )

    base = alt.Chart(counts).encode(
        y=alt.Y('Character:N', sort=chara_cols, title=None, axis=alt.Axis(ticks=False, grid=False, domain=False)),
        x=alt.X('Proportion:Q', axis=alt.Axis(format='%', ticks=False, grid=False, domain=False, title=None, labels = False)),
        color=alt.Color('FavorabilitySimple:N', sort=favorability_order,
                        scale=color_scale, legend = None),
        tooltip=[alt.Tooltip('Character:N'), alt.Tooltip('FavorabilitySimple:N'), alt.Tooltip('Proportion:Q', format='.0%')]
    )

    bars = base.mark_bar()

    labels = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  
    ).encode(
        text=alt.Text('Proportion:Q', format='.0%')
    )

    layered = (bars + labels).properties(
    width=130,
    height=400
    )

    chart = layered.facet(
        column=alt.Column('FavorabilitySimple:N', sort=favorability_order, header=alt.Header(labelAngle=0, title=''))
    ).configure(background='#F2F2F2'
    ).properties(
        title=alt.TitleParams(
            text="'Star Wars' Characters Favorability Ratings",
            subtitle=["By 834 Respondents"],
            anchor='start',
            fontSize=20,
            subtitleFontSize=13,
            subtitleColor='black'
        )
    )
    return chart
```

### 2.4 Who shot first? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genFirstShotVis(inpf):
    # input: inpf, the star wars dataset
    # output: the Altair visualization as described above

    col = 'Which character shot first?'

    who_shot = (
        inpf[col]
        .dropna()
        .value_counts()
        .reset_index()
    )
    who_shot.columns = ['response', 'count']
    who_shot['percent'] = who_shot['count'] / who_shot['count'].sum() * 100

    order = ['Han', 'Greedo', "I don't understand this question"]
    who_shot['response'] = pd.Categorical(who_shot['response'], categories=order, ordered=True)

    bars = (
        alt.Chart(who_shot)
        .mark_bar(color='#3C99DE', size = 20)
        .encode(
            x=alt.X('percent:Q',
                    title = None,
                    axis=alt.Axis(format='.0f', grid=False, labels = False, ticks = False)),
            y=alt.Y('response:N',
                    sort=order,
                    title=None,
                    axis=alt.Axis(
                        labelFontSize= 12,
                        labelColor='black',
                        labelPadding=10
                    )),
            tooltip=[
                alt.Tooltip('response:N', title='Response'),
                alt.Tooltip('percent:Q', format='.1f', title='Percent')
            ]
        )
    )

    text = (
            alt.Chart(who_shot)
            .transform_calculate(
                percent_label="round(datum.percent) + '%'"
            )
            .mark_text(
                align='left',
                baseline='middle',
                dx=3,
                fontSize=12,
                color='black'
            )
            .encode(
                x='percent:Q',
                y=alt.Y('response:N', sort=order),
                text='percent_label:N'
            )
        )

    chart = (
        (bars + text)
        .properties(
            title=alt.TitleParams(
                text="Who shot first?",
                subtitle="According to 834 participants",
                color='black',
                subtitleColor='black',
                fontSize=20,
                subtitleFontSize=15,
                anchor='start'
            ),
            width=400,
            height=180
        )
        .configure_axis(
            grid=False,
            domain=False
        )
        .configure_view(strokeWidth=0)
        .configure(background='#F2F2F2')
    )
          
    return chart
    
```
