```python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    # create the various mapping and order lists/dictionaries
    # return: list of episodes (in order), episode->name map, list of episode names (in order)
    episodes = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    names = {
        'EI' : 'The Phantom Menace', 'EII' : 'Attack of the Clones', 'EIII' : 'Revenge of the Sith', 
        'EIV': 'A New Hope', 'EV': 'The Empire Strikes Back', 'EVI' : 'The Return of the Jedi'
    }

    # we're also going to use this order to sort, so names_l will now have our sort order
    return episodes, names, [names[ep] for ep in episodes]

episodes, namesDict, namesList = genEpisodeNamesDF()
```

```python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
```

```python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN
    salo = indf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    return(salo)
```

```python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
```

```python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total = len(inpf)
    percs = []

    # loop over each column and calculate the number of people who have seen the movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count them
    # and divide by the percent
    for seen_ep in ['seen_' + ep for ep in episodes]:
        perc = len(inpf[~ pd.isna(inpf[seen_ep])]) / total
        percs.append(perc)

    # at this point percs is holding our percentages

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes],percs))
    seen_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    return(seen_per_df)
```

```python
seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
```

```python
# let's see what's inside
seenPerMovieDF
```

```python
def genPercentVis(indf, names_l=namesList):
    # input: indf, the dataframe as seen_per_df
    # input: names_l - a list of all the movies in series order (default namesList)
    # output: simple altair bar chart
    
    # ok, time to make the chart... let's make a bar chart (use mark_bar)
    bars = alt.Chart(indf).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        )
    )

    # at this point we don't really have a great plot (it's missing the annotations, titles, etc.)
    return(bars)

seenPerMovieVis = genPercentVis(seenPerMovieDF)
```

```python
def augmentPercentVis(base):
    # input: base (the base vis, i.e., bars as above)
    # we're going to overlay the text with the percentages, so let's make another visualization
    # that's just text labels


    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    seen_movies = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=180
    ).properties(
        # add a title
        title="Which 'Star Wars' Movies Have you Seen?"
    )

    return(seen_movies)

# note that we are NOT formatting this in the Five Thirty Eight Style yet... we'll leave that to you to figure out
```

### 2.2 How people rate the 'Star Wars' movie? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genRateVis(inpf,eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above

    seen_all_df = inpf.dropna(subset=['seen_' + ep for ep in eps],how='any') #Remove any rows who haven't seen all 6 films
    
    total = len(seen_all_df)

    
    tuples = []

    for ep in eps:
        rank_ep = "rank_" + ep
        top_perc = len(seen_all_df[(seen_all_df[rank_ep] == "1") | (seen_all_df[rank_ep] == "2")]) / total
        middle_perc = len(seen_all_df[(seen_all_df[rank_ep] == "3") | (seen_all_df[rank_ep] == "4")]) / total
        bottom_perc = len(seen_all_df[(seen_all_df[rank_ep] == "5") | (seen_all_df[rank_ep] == "6")]) / total

        tuples.append((names[ep], "Top third", top_perc))
        tuples.append((names[ep], "Middle third", middle_perc))
        tuples.append((names[ep], "Bottom third", bottom_perc))

    percs_df = pd.DataFrame(tuples, columns = ['Name', "Third_Rank", "Percentage"])

    base = alt.Chart(percs_df).mark_bar(size=20).encode(
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             sort=names_l
        ),
        color=alt.Color(
            'Third_Rank:O',
            scale=alt.Scale(domain=["Top third", "Middle third", "Bottom third"],
                            range=['#77ab43', '#008fd5', '#ff2700']),
            legend=None
        )
    ).properties(
        width=100,
        height=200
    )

    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3 
    ).encode(
        text=alt.Text('Percentage:Q',format='.0%')
    )

    vis = alt.layer(base, text).facet(
        column=alt.Column(
            'Third_Rank:O', 
            sort=["Top third", "Middle third", "Bottom third"],
            title=None
        )
    ).resolve_scale(
        y='shared'
    )

    vis = vis.configure_view(
        strokeWidth=0
    ).configure_scale(
        bandPaddingInner=0.2
    ).properties(
        title={
            "text": ["How people rate the 'Star Wars' Movies", "How often each film was rated in the top, middle, and bottom third", f"(by {total} respondents who have seen all six films)"],
            "anchor": "start"
        }
    )

    
    return vis
    
```

### 2.3 Star Wars' Characters Favorability Ratings. Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genFavorabilityVis(inpf):
    characters = np.array(inpf.columns[15:29])
    favorabilities = np.array(inpf["Luke Skywalker"].unique())

    tuples = []

    for chosen in characters:
        df = inpf.dropna(subset=[chosen])
        
        favorable = len(df[(df[chosen] == "Very favorably") | (df[chosen] == "Somewhat favorably")]) / len(df)
        neutral = len(df[df[chosen] == "Neither favorably nor unfavorably (neutral)"]) / len(df)
        unfavorable = len(df[(df[chosen] == "Very unfavorably") | (df[chosen] == "Somewhat unfavorably")]) / len(df)
        unfamiliar = len(df[df[chosen] == "Unfamiliar (N/A)"]) / len(df)

        tuples.append((chosen, "Favorable", favorable))
        tuples.append((chosen, "Neutral", neutral))
        tuples.append((chosen, "Unfavorable", unfavorable))
        tuples.append((chosen, "Unfamiliar", unfamiliar))
    
    char_df = pd.DataFrame(tuples, columns = ['Name', "Favorability", "Percentage"])

    favorable_df = char_df[char_df['Favorability'] == 'Favorable']
    sort_names = favorable_df.sort_values('Percentage', ascending=False)['Name'].tolist()

    

    base = alt.Chart(char_df).mark_bar(size=20).encode(
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
            sort=sort_names
        ),
        color=alt.Color(
            'Favorability:O',
            scale=alt.Scale(domain=["Favorable", "Neutral", "Unfavorable", "Unfamiliar"],
                            range=['#77ab43', '#008fd5', '#ff2700', '#808080']),
            legend=None
        )
    ).properties(
        width=120,
        height=300
    )



    

    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3 
    ).encode(
        text=alt.Text('Percentage:Q',format='.0%')
    )




    

    vis = alt.layer(base, text).facet(
        column=alt.Column(
            'Favorability:O', 
            sort=["Favorable", "Neutral", "Unfavorable", "Unfamiliar"],
            title=None
        )
    ).resolve_scale(
        y='shared'
    )

    vis = vis.configure_view(
        strokeWidth=0
    ).configure_scale(
        bandPaddingInner=0.2
    ).properties(
        title={
            "text": ["'Star Wars' character favorability ratings", "By 834 respondents"],
            "anchor": "start"
        }
    )

    
    return vis



    
```

### 2.4 Who shot first? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genFirstShotVis(inpf):
    indf = inpf.dropna(subset=["Which character shot first?"])
    ans = ["Han", "Greedo", "I don't understand this question"]
    percs = []

    for character in ans:
        perc = len(indf[indf["Which character shot first?"] == character]) / len(indf)
        percs.append(perc)

    tuples = list(zip(ans, percs))
    percs_df = pd.DataFrame(tuples, columns = ['Name', "Percentage"])
    base = alt.Chart(percs_df).mark_bar(size=20).encode(
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
            sort=ans
        )
    )

    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    shot_first = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=180
    ).properties(
        # add a title
        title={
            "text": ["Who Shot First?", f"According to 834 respondents"],
            "anchor": "start"
        }
    )

    return shot_first
    
    
```

```python
# YOUR CODE HERE
def genSeenAllSixAgePie(inpf, eps):
    seen_all_df = inpf.copy().dropna(subset=['seen_' + ep for ep in eps],how='any') #Remove any rows who haven't seen all 6 films
    seen_all_df.dropna(subset=["Age"], inplace=True)
    ages = ["18-29", "30-44", "45-60", "> 60"]
    percs = []
    for age in ages:
        age_perc = len(seen_all_df[seen_all_df["Age"] == age]) / len(seen_all_df)
        percs.append(age_perc)
    tuples = list(zip(ages, percs))

    age_df = pd.DataFrame(tuples, columns=["Age", "Percentage"])
    age_df["Angle"] = age_df["Percentage"].cumsum() - age_df["Percentage"] / 2

    pie = alt.Chart(age_df).mark_arc().encode(
        theta=alt.Theta(field="Percentage", type="quantitative"),
        color=alt.Color("Age:N", sort=ages, scale=alt.Scale(scheme="category10")),
        tooltip=[
            alt.Tooltip("Age:N"),
            alt.Tooltip("Percentage:Q", format=".1%")
        ]
    )

    text = alt.Chart(age_df).mark_text(size=14).encode(
        theta=alt.Theta("Angle:Q", stack=None),
        radius=alt.value(100),
        text=alt.Text('Percentage:Q', format='.0%'),
        color=alt.value("black"),
    )
    chart = (pie + text).properties(
        title={
            "text": ["Which ages have seen all 6 films?", f"According to {len(seen_all_df)} respondents"],
            "anchor": "start"
        }
    )

    return chart


genSeenAllSixAgePie(sw, episodes)
```
