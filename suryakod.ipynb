#### Answer 
By definition, a visualization is more effective than another if the information conveyed by it is more readily perceived than the information in the other.

A pie chart is usually effective for showing a 'part-to-whole' relationship. Using a single pie chart to display the exact data from the bar chart would be ineffective, misleading, and incorrect. Viewing each of the six movies is a mutually exclusive event and does not form a 'whole' (does not add up to 100%, it’s more). The bar chart is effective and the right visualization to compare the percentage of viewers for each movie (compared to a pie chart). One way to accurately represent the data from the bar graph with a pie chart would be to create six pie charts—one for each movie—that show the percentage of viewers and non-viewers. However, this would mostly be ineffective since six pie charts can be overwhelming and cognitively overload the viewer compared to a single, simple bar chart displaying the same data. It would also make comparing data across movies more difficult (especially if the pie chart doesn’t include percentages in text or similar details). Additionally, length is considered a better encoding than area for representing quantitative variables.

In a situation where the question was targeted towards a single movie, for example, 'What percentage of respondents watched the most popular movie - The Empire Strikes Back?', or just to judge at a high level of which movie has more viewers, then a pie chart would be a good enough visual to use, as this question asks about a part of the whole.

```python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    # create the various mapping and order lists/dictionaries
    # return: list of episodes (in order), episode->name map, list of episode names (in order)
    episodes = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    names = {
        'EI' : 'The Phantom Menace', 'EII' : 'Attack of the Clones', 'EIII' : 'Revenge of the Sith', 
        'EIV': 'A New Hope', 'EV': 'The Empire Strikes Back', 'EVI' : 'The Return of the Jedi'
    }

    # we're also going to use this order to sort, so names_l will now have our sort order
    return episodes, names, [names[ep] for ep in episodes]

episodes, namesDict, namesList = genEpisodeNamesDF()
```

```python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
```

```python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN
    salo = indf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    return(salo)
```

```python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
```

```python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total = len(inpf)
    percs = []

    # loop over each column and calculate the number of people who have seen the movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count them
    # and divide by the percent
    for seen_ep in ['seen_' + ep for ep in episodes]:
        perc = len(inpf[~ pd.isna(inpf[seen_ep])]) / total
        percs.append(perc)

    # at this point percs is holding our percentages

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes],percs))
    seen_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    return(seen_per_df)
```

```python
seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
```

```python
# let's see what's inside
seenPerMovieDF
```

```python
def genPercentVis(indf, names_l=namesList):
    # input: indf, the dataframe as seen_per_df
    # input: names_l - a list of all the movies in series order (default namesList)
    # output: simple altair bar chart

    # ok, time to make the chart... let's make a bar chart (use mark_bar)
    bars = alt.Chart(indf)\
        .mark_bar(size=23, color="#008fd5")\
        .encode(
            x=alt.X("Percentage", axis=None), # encode x as the percent, and hide the axis
            y=alt.Y(
                "Name:N",  # encode y using the name, use the movie name to label the axis
                axis=alt.Axis(  # remove the y-axis line, tick lines and title 
                    tickCount=5,
                    title="",
                    ticks=False,
                    domain=False,
                    labelFontSize=16,  # format the labels (movie titles) to match the visual in the article
                    labelColor="#838383",
                    labelPadding=10
                ),
                sort=names_l,  # we give the sorting order to avoid alphabetical order
            )
        )
    
    # at this point we don't really have a great plot (it's missing the annotations, titles, etc.)
    return bars

seenPerMovieVis = genPercentVis(seenPerMovieDF)
```

```python
def augmentPercentVis(base):
    # input: base (the base vis, i.e., bars as above)
    # we're going to overlay the text with the percentages, so let's make another visualization
    # that's just text labels

    text = base.transform_window(
            RowNumber="row_number()"  # create a row number to use for the % sign logic
        ).transform_calculate(
            label="datum.RowNumber == 1 ? \
                    format(datum.Percentage, '.0%') : \
                    format(datum.Percentage * 100, '.0f')" # conditional formatting so only the first row will have the '%' sign
        ).mark_text(
            align="left",
            baseline="middle",
            dx=3,  # Nudges text to right so it doesn't appear on top of the bar
            color="#535353",
            size=15
        ).encode(text="label:N")  # we'll use the percentage as the text

    # finally, we're going to combine the bars and the text and do some styling
    seen_movies = (text + base)\
        .configure_scale(
            bandPaddingInner=0.02,
            bandPaddingOuter=0.1  # add some padding to fix the spacing between the bars
        ).configure(background="#f0f0f0")\
        .configure_view(
            strokeWidth=0,  # we don't want a stroke around the bars
            fill="transparent"  # ensures no gray background
        ).properties(
            width=500,  # set the dimensions of the visualization
            height=180,
            title=alt.TitleParams(
                "Which 'Star Wars' Movies Have You Seen?",
                color="#535353",
                fontSize=23,
                anchor="start",  # Align title to the start (left)
                orient="top",  # Place title at the top
                subtitle=f"Of {len(seenAtLeastOneDF)} respondents who have seen any film",
                subtitleColor="#535353",
                subtitleFontSize=18
            ),
        )

    return seen_movies
# note that we are NOT formatting this in the Five Thirty Eight Style yet... we'll leave that to you to figure out
```

### 2.2 How people rate the 'Star Wars' movie? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)
def genRateVis(inpf, eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above

    # YOUR CODE HERE
    # We need all the respondents who have seen all the movies
    seen_all_df = getSeenAllDF(inpf, eps)
    total = len(seen_all_df)

    # Split the ranks into top, middle and bottom third
    rank_groups = [["1", "2"], 
                   ["3", "4"], 
                   ["5", "6"]]
    rank_group_titles = ["Top third", "Middle third", "Bottom third"]
    percs = []

    # Loop over each column and calculate the percent of people who ranked each movie in the top, middle and bottom third
    for rank_ep in ["rank_" + ep for ep in episodes]:
        perc_group = []
        for rank_group in rank_groups:
            perc = len(seen_all_df[seen_all_df[rank_ep].isin(rank_group)]) / total
            perc_group.append(perc)
        percs.append(perc_group)

    # pair names with percents and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes], percs))
    rank_group_per_df = pd.DataFrame(
        [(movie, percs[0], percs[1], percs[2]) for movie, percs in tuples],
        columns=["Name"] + rank_group_titles
    )

    # Reshape data for the bar chart
    data_long = pd.melt(rank_group_per_df, id_vars = ["Name"], value_vars = rank_group_titles, 
                        var_name = "Rank Group", value_name = "Percentage")

    # Add row number for the '%' formatting in the chart
    data_long["RowNumber"] = data_long["Name"].apply(lambda x: (namesList.index(x)+1))

    # Create the base bar chart
    base = alt.Chart(data_long)\
        .mark_bar(size=23)\
        .encode(
            x=alt.X("Percentage:Q", scale=alt.Scale(domain=[0, 2]), axis=None),
            y=alt.Y(
                "Name:N",  # encode y using the name, use the movie name to label the axis
                axis=alt.Axis(
                    tickCount=5,
                    title="",
                    ticks=False,
                    domain=False,
                    labelFontSize=16,
                    labelColor="#737373",
                    labelPadding=10
                ),
                sort=names_l,
            ),
            color=alt.Color(
                "Rank Group:N",
                sort=rank_group_titles,
                scale=alt.Scale(
                    domain=rank_group_titles,
                    range=["#7cb342", "#42a5f5", "#ef5350"]
                ),
                legend=None
            )
        ).properties(width=140, height=200)

    # Text chart with conditional formatting
    text = alt.Chart(data_long)\
        .transform_calculate(
            label="datum.RowNumber == 1 ? \
                    format(datum.Percentage, '.0%') : \
                    format(datum.Percentage * 100, '.0f')" # only first row will have the '%' sign
        ).mark_text(
            align="left", 
            baseline="middle", 
            dx=3, 
            color="#4a4a4a", 
            fontSize=14
        ).encode( 
            x=alt.X("Percentage:Q"),
            y=alt.Y("Name:N", sort=names_l),
            text="label:N"
        )

    # Use 'facet' to create 'columns' of charts based on the 'Rank Group'
    # Not using the pre-buil functions here as altair doesn't allow charts with the config before applying facet
    chart = (base + text).facet(
            column = alt.Column(
                "Rank Group:N",
                sort = rank_group_titles,
                title = None,
                header = alt.Header(
                    labelFontSize = 18, 
                    labelColor = "#636363", 
                    labelFontWeight = "bold"
                )
            ),
            spacing = 1,
            center = True
        ).properties(
            title=alt.TitleParams(
                "How People Rate the 'Star Wars' Movies",
                color="#535353",
                fontSize=22,
                anchor="start",
                orient="top",
                subtitle=[
                    "How often each film was rated in the top, middle and bottom third",
                    f"(by {total} respondents who have seen all six films)",
                    ""
                ],
                subtitleColor="#636363",
                subtitleFontSize=17
            )
        ).configure(background="#f0f0f0")\
        .configure_view(
            strokeWidth=0,  # we don't want a stroke around the bars
            fill="transparent"  # ensures no gray background
        )
    
    return chart
```

### 2.3 Star Wars' Characters Favorability Ratings. Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)
def genFavorabilityVis(inpf):
    # input: inpf, the star wars dataset (as defined at top of file)
    # output: the Altair visualization as described above

    # YOUR CODE HERE
    all_characters = ["Han Solo", "Luke Skywalker", "Princess Leia Organa", "Anakin Skywalker", "Obi Wan Kenobi",
        "Emperor Palpatine", "Darth Vader", "Lando Calrissian", "Boba Fett", "C-3P0",
        "R2 D2", "Jar Jar Binks", "Padme Amidala", "Yoda"]
    #  Get the rows where atleast one of the characters are given a rating 
    characters = [col for col in inpf.columns if col in all_characters]
    char_fav_df = inpf[characters].dropna(subset=characters, how="all")
    
    total = len(char_fav_df)
    # Update the values in the columns to the ones needed for the visual in the article
    # all_unique_values = pd.unique(char_fav_df[characters].values.flatten())
    updated_values_mapping = {
        "Very favorably": "Favorable",
        "Unfamiliar (N/A)": "Unfamiliar",
        "Somewhat favorably": "Favorable",
        "Somewhat unfavorably": "Unfavorable",
        "Very unfavorably": "Unfavorable",
        "Neither favorably nor unfavorably (neutral)": "Neutral"
    }
    char_fav_df = char_fav_df.replace(updated_values_mapping)
    rating_order = ["Favorable", "Neutral", "Unfavorable", "Unfamiliar"]

    # Melt the DataFrame to long format, group by to get the rating counts and calculate the percentages
    char_fav_df = char_fav_df.melt(var_name="Character", value_name="Rating")
    char_fav_df = char_fav_df.groupby(["Character", "Rating"]).size().reset_index(name="Count")
    char_fav_df["Percentage"] = char_fav_df["Count"] / total

    # Get order of characters based on the favorability percentage to avoid alphabetical ordering
    char_order = list(char_fav_df[char_fav_df["Rating"] == "Favorable"].sort_values(["Percentage"], ascending=False)["Character"])
    
    # Add row number for the '%' formatting in the chart
    char_fav_df["RowNumber"] = char_fav_df["Character"].apply(lambda x: (char_order.index(x)+1))  

    # Create the base bar chart
    base = alt.Chart(char_fav_df)\
        .mark_bar(size=23)\
        .encode(
            x=alt.X("Percentage:Q", axis=None),
            y=alt.Y("Character:N",
                axis=alt.Axis(
                    tickCount=5,
                    title="",
                    ticks=False,
                    domain=False,
                    labelFontSize=16,
                    labelColor="#737373",
                    labelPadding=10
                ),
                sort=char_order,
            ),
            color=alt.Color(
                "Rating:N",
                sort=rating_order,
                scale=alt.Scale(
                    domain=rating_order,
                    range=["#7cb342", "#42a5f5", "#ef5350", "#999999"]
                ),
                legend=None
            )
        ).properties(width=65, height=350)
    
    # Text chart with conditional formatting
    text = alt.Chart(char_fav_df)\
        .transform_calculate(
            label="datum.RowNumber == 1 ? \
                    format(datum.Percentage, '.0%') : \
                    format(datum.Percentage * 100, '.0f')" # only first row will have the '%' sign
        ).mark_text(
            align="left", 
            baseline="middle", 
            dx=3, 
            color="#4a4a4a", 
            fontSize=15
        ).encode(
            y=alt.Y("Character:N", sort=char_order),
            x=alt.X("Percentage:Q"),
            text="label:N"
        )
    
    # Layer bars and text
    layered = alt.layer(base, text)

    # Use 'facet' to create 'columns' of charts based on the 'Rating'
    # Not using the pre-buil functions here as altair doesn't allow charts with the config before applying facet
    chart = (base + text).facet(
            column=alt.Column(
                "Rating:N",
                sort=rating_order, # specify the order to avoid alphabetical ordering
                title=None,
                header=alt.Header(
                    labelFontSize=16,
                    labelColor="#636363",
                    labelFontWeight="bold"
                )
            )
        ).properties(
            title=alt.TitleParams(
                "'Star Wars' Characters Favorability Ratings",
                color="#535353",
                fontSize=22,
                anchor="start",
                orient="top",
                subtitle=[f"By {total} respondents", ""],
                subtitleColor="#636363",
                subtitleFontSize=17
            )
        ).configure(background="#f0f0f0")\
        .configure_view(
            strokeWidth=0,  # we don't want a stroke around the bars
            fill="transparent"  # ensures no gray background
        )
    
    return chart
```

#### Justification
*Provide your justification here* to argue why your "better" version is actually better. Make sure to defend this based on expressiveness and effectiveness as well as perceptual arguments.

While both the bar chart and pie charts depict the same data - of the respondents, based on gender, how many are viewers and how many are fans?, based on expressiveness, effectiveness and the principles of perception and cognition, the bar chart is the "better" visual.


#### Expressiveness 
For a visual to be expressive, it should encode all the information in the data and only that information.
- The bar chart shows the precise numbers and the split both within and across categories (Viewer x Fan) and genders. 
- The pie charts show the split within a category but eliminate the cross-category comparison, which is available in the raw data. Also, there is no way to get a precise number of respondents for a category/gender, only estimated percentages. 
- The bar chart expresses quantitative information more accurately than the pie charts. 

#### Effectiveness
A visualization is more effective if the information conveyed by one visualization is more readily perceived than the information in another visualization. 
- The bar chart leverages position and length - the most effective visual encodings for quantitative data.
- The bar chart makes it easy to draw conclusions about the data without mental estimation. For example, based on the bar chart, we can say with 100% certainty that the gender gap in the viewership is smaller than the gap in the fandom. The pie charts do not allow for this comparison. 
- Pie charts rely on angle and area, which are cognitively weaker encodings and cause more cognitive load and misinterpretation. 
- The pie charts make between-category comparisons (Viewer vs. Fan) impossible in just numbers because each pie adds up to a different total. You cannot compare a slice from the ‘Viewer’ pie to a slice from the ‘Fan’ pie because they're proportions of different wholes. The bar chart avoids this entirely by using absolute counts on a common scale.
- The bar chart is clearly more effective

#### Perception and Cognition 

When it comes to perception and cognition, the simplest and most stable interpretations are favored. 
- According to Cleveland & McGill's perceptual hierarchy, Length is more accurate than angle and area, making the bar chart easier on Perception and Cognition. 
- Two important questions in Perception and Cognition are: ‘Which is bigger?’ and ‘How much bigger?’ While you can answer the first question with the pie chart, you can answer both easily with the bar chart.
- The bar chart offers continuity and proximity - important Gestalt principles. The pie charts break the continuity and make comparing across categories difficult.
- In the bar chart, all bars originate from a common baseline, are compared on the same scale, and use the strongest encodings—position and length. These factors make it much more perceptive and lighter on cognitive load. 
- In the pie chart, there is no common baseline for the slices, and the encodings are angle and arc length. Perceiving these encodings requires significantly higher cognitive load, leading to higher errors.

Based on all three categories—expressiveness, effectiveness, perception and cognition—the bar chart visual is the "better" version.
