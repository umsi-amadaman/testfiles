Your colleague is arguing that the chart above could be made more *effective* by using one or more pie charts. In what situations are they right? In what situations would they be wrong? If it helps makes your case, feel free to create a picture (hand drawn is fine) and load it into the notebook.

They are right in situations when:

-There are only 2–3 categories that add up to a whole (100%), for example, male vs. female viewers.
-The main goal is to show the proportion of one part to the total, not to compare many items.
- People only need to see a rough share, not precise values.

They are wrong in situations when:

- There are 6 films, not 1 whole. Each movie’s percentage is independent, not parts of one circle.
- It is hard to compare slices across multiple pies — humans are much worse at comparing angles than lengths.
-Small differences (for example- 80% vs 88%) are easy to see in bar chart, but hard to assess in a pie chart.

```python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    """Return (episodes_in_order, mapping to full names, full names in order)."""
    eps = ['EI','EII','EIII','EIV','EV','EVI']
    names = {
        'EI'  : 'The Phantom Menace',
        'EII' : 'Attack of the Clones',
        'EIII': 'Revenge of the Sith',
        'EIV' : 'A New Hope',
        'EV'  : 'The Empire Strikes Back',
        'EVI' : 'The Return of the Jedi'
    }
    return eps, names, [names[e] for e in eps]

episodes, namesDict, namesList = genEpisodeNamesDF()
print("\nEpisode codes\n", episodes)
print("\nFull names in order\n", namesList)
```

```python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
```

```python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN

    return indf.dropna(subset=[f'seen_{ep}' for ep in eps], how='all')

seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print(f"\nTotal who have seen at least 1\n\n {len(seenAtLeastOneDF)}")
display(seenAtLeastOneDF.sample(2))
```

```python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
```

```python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total=len(inpf)
    percs = []


    for ep in episodes:
        cnt = len(inpf[~pd.isna(inpf[f'seen_{ep}'])])
        percs.append(cnt / total)

    # build the result table in the series order
    out = pd.DataFrame(list(zip([names[e] for e in episodes], percs)),
                       columns=['Name', 'Percentage'])
    return out


seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
print("\nPerc-seen table (head)\n")
display(seenPerMovieDF.head())
```

```python


def getBestMovieShareDf(indf, names=namesDict, names_l=namesList):

    rank_cols = [f'rank_{ep}' for ep in episodes]
    sub = indf.dropna(subset=rank_cols, how='any')
    total = len(sub)


    best_ep = []
    for _, row in sub[rank_cols].iterrows():
        i = np.argmin(row.values.astype(float))
        best_ep.append(episodes[i])
    best_ep = pd.Series(best_ep, name='best')

    share = (best_ep.value_counts(normalize=True)
             .reindex(episodes)
             .fillna(0.0)
             .rename(index=names)
             .reset_index())
    share.columns = ['Name','Percentage']
    
    return share

bestShareDF = getBestMovieShareDf(sw)
print(f"\nN who ranked all 6\n {len(bestShareDF)} (below is the table of shares)")
display(bestShareDF)


def bestMovieVis(indf, names_l=namesList):
    base= alt.Chart(indf).mark_bar(size=19).encode(
        x=alt.X('Percentage:Q', axis=None),
        y=alt.Y('Name:N', sort=names_l, axis=alt.Axis(title='')),
        color=alt.value('#008fd5')
    )
    labels = base.mark_text(align='left', baseline='middle', dx=3).encode(
        text=alt.Text('Percentage:Q', format='.0%')
    )
    
    return (labels + base).properties(width=550, height=230,
                                      title="What's the Best 'Star Wars' Movie?")

bestBar=bestMovieVis(bestShareDF)
bestBar
```

```python
# Recreate this image using altair here (10 POINTS)

def keep_digits_or_nan(x):
    if pd.isna(x):
        return np.nan
    if isinstance(x, (int, np.integer)):
        return int(x)
    if isinstance(x, (float, np.floating)):
        return int(x) if np.isfinite(x) else np.nan
    s = str(x).strip()
    try:
        return int(float(s))
    except Exception:
        pass
    dig =''.join(ch for ch in s if ch.isdigit())
    return int(dig) if dig else np.nan

def clean_rank_columns(inpf, episodes):
    df = inpf.copy()
    for e in episodes:
        c = f'rank_{e}'
        if c in df.columns:
            df[c] = df[c].apply(keep_digits_or_nan)
            df[c] = pd.to_numeric(df[c], errors='coerce')
    return df

def genRateVis(inpf, eps, names=namesDict, names_l=namesList):
    print("\nRunning genRateVis() ...\n")
    print("Column check (first 8):", list(inpf.columns[:8]))
    print("Rank columns:", [c for c in inpf.columns if c.startswith('rank_')])


    df =clean_rank_columns(inpf, eps)
    rank_cols = [f'rank_{e}' for e in eps]

    have_all = df[rank_cols].notna().all(axis=1)
    sub = df.loc[have_all, rank_cols].copy()
    total =len(sub)
    print(f"\nRows with all 6 ranks present: {total}\n")

    rows =[]
    for e in eps:
        c = f'rank_{e}'
        top    =  (sub[c] <= 2).sum() / total if total else 0.0
        middle = ((sub[c] >= 3) & (sub[c] <= 4)).sum() / total if total else 0.0
        bottom =  (sub[c] >= 5).sum() / total if total else 0.0
        rows.append((names[e], "Top third",    top))
        rows.append((names[e], "Middle third", middle))
        rows.append((names[e], "Bottom third", bottom))

    tidy = pd.DataFrame(rows, columns=["Name", "Group", "Perc"])
    tidy["Name"]  = pd.Categorical(tidy["Name"], categories=names_l, ordered=True)
    tidy["Group"] = pd.Categorical(tidy["Group"],
                                   categories=["Top third","Middle third","Bottom third"],
                                   ordered=True)

    print("\nHead of tidy rate table:\n")
    display(tidy.head(9))

    color_scale = alt.Scale(
        domain=["Top third","Middle third","Bottom third"],
        range=["#3BA44D", "#1f77b4", "#d62728"]
    )

    base   = alt.Chart(tidy)
    bars   = base.mark_bar(size=20).encode(
        x=alt.X("Perc:Q", axis=None),
        y=alt.Y("Name:N", sort=names_l, axis=alt.Axis(title="")),
        color=alt.Color("Group:N", scale=color_scale, legend=None)
    )
    labels = base.mark_text(align="left", baseline="middle",
                            dx=6, fontSize=12).encode(
        x="Perc:Q",
        y=alt.Y("Name:N", sort=names_l),
        text=alt.Text("Perc:Q", format=".0%")
    )

    spec=(bars + labels).properties(width=240, height=220)

    faceted = spec.facet(
        column=alt.Column(
            "Group:N",
            sort=["Top third","Middle third","Bottom third"],
            title=None
        )
    ).resolve_scale(x='independent')


    title_text = "How People Rate the ‘Star Wars’ Movies"
    subtitle_text = ("How often each film was rated in the top, middle and bottom third "
                     "(by 471 respondents who have seen all six films)")

    final = faceted.properties(
        title=alt.TitleParams(
            text=title_text,
            subtitle=subtitle_text,
            anchor='start',
            font='Source Sans Pro',
            fontSize=22,
            subtitleFont='Source Sans Pro',
            subtitleFontSize=13
        )
    ).configure_axis(
        labelFont="Source Sans Pro",
        titleFont="Source Sans Pro"
    ).configure_view(
        strokeWidth=0
    )

    print("\nFinished genRateVis().\n")
    return final


genRateVis(sw, episodes)


```

### 2.3 Star Wars' Characters Favorability Ratings. Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genFavorabilityVis(inpf):
    # input: inpf, the star wars dataset (as defined at top of file)
    # output: the Altair visualization as described above
    

    print("\nRunning genFavorabilityVis\n")

    candidate_cols = []
    for c in inpf.columns:
        if inpf[c].dtype == 'O':
            # sample a few values to see if they look like survey answers
            vals = inpf[c].dropna().astype(str).str.lower().head(30)
            if any(v.startswith(('very', 'somewhat', 'neutral', 'unfamiliar')) for v in vals):
                candidate_cols.append(c)

    print(f"Detected character columns\n {candidate_cols}\n")

    def map_bucket(x):
        if pd.isna(x):
            return np.nan
        s = str(x).strip().lower()
        if s.startswith('very favor') or s.startswith('somewhat favor'):
            return 'Favorable'
        if s.startswith('very unfavor') or s.startswith('somewhat unfavor'):
            return 'Unfavorable'
        if s.startswith('neutral'):
            return 'Neutral'
        if s.startswith('unfamiliar'):
            return 'Unfamiliar'

        return 'Neutral'

    rows = []
    ordered_groups = ['Favorable', 'Neutral', 'Unfavorable', 'Unfamiliar']
    for ch in candidate_cols:
        resp = inpf[ch].dropna().apply(map_bucket)
        denom = len(resp)
        if denom == 0:
            continue
        for g in ordered_groups:
            num = (resp == g).sum()
            rows.append((ch, g, num / denom))

    tidy = pd.DataFrame(rows, columns=['Character', 'Group', 'Perc'])
    tidy['Character'] = pd.Categorical(tidy['Character'],
                                       categories=candidate_cols,
                                       ordered=True)
    tidy['Group'] = pd.Categorical(tidy['Group'],
                                   categories=ordered_groups,
                                   ordered=True)

    print("\nHead of tidy favorability table\n")
    display(tidy.head(12))

    print("\nChecks —each character should sum ~1 across groups\n")
    checks=(tidy.groupby('Character')['Perc'].sum()
                   .reset_index(name='Sum').head(8))
    display(checks)


    color_scale =alt.Scale(
        domain=ordered_groups,
        range=['#3BA44D',
               '#1f77b4',
               '#d62728',
               '#9e9e9e']
    )

    base = alt.Chart(tidy)

    bars = base.mark_bar().encode(
        y=alt.Y('Character:N', sort=candidate_cols, axis=alt.Axis(title='')),
        x=alt.X('Perc:Q', stack='normalize',
                axis=alt.Axis(title=None, format='.0%')),
        color=alt.Color('Group:N', scale=color_scale, legend=alt.Legend(title=None))
    )

    labels = base.transform_aggregate(
        total='sum(Perc)', groupby=['Character']
    ).transform_calculate(
        label="datum.total"  # not used here since we show legend & x-axis
    )

    title_text = "Star Wars’ Character Favorability Ratings"
    subtitle_text = f"By {len(inpf)} respondents (percent, among those answering per character)"

    chart = (bars.properties(width=780, height=420)
                 .resolve_scale(x='independent'))


    
    final = chart.properties(
        title=alt.TitleParams(
            text=title_text,
            subtitle=subtitle_text,
            anchor='start',
            font='Source Sans Pro',
            fontSize=22,
            subtitleFont='Source Sans Pro',
            subtitleFontSize=13
        )
    ).configure_axis(
        labelFont='Source Sans Pro',
        titleFont='Source Sans Pro'
    ).configure_view(
        strokeWidth=0
    )

    print("\nFinished genFavorabilityVis\n")
    return final

genFavorabilityVis(sw)
```

### 2.4 Who shot first? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genFirstShotVis(inpf):
    # input: inpf, the star wars dataset
    # output: the Altair visualization as described above
    
    print("\nRunning genFirstShotVis\n")

    shot_col=None
    
    for c in inpf.columns:
        if "Which character shot first" in c:
            shot_col = c
            break
    if shot_col is None:
        raise ValueError("Could not find the 'Which character shot first?' column.")
    print("Using column:", shot_col)

    cat_map = {
        "Han": "Han",
        "Greedo": "Greedo",
        "I don't understand this question": "I don't understand this question"
    }

    def norm_shot(x):
        if pd.isna(x):
            return np.nan
        s=str(x).strip()
        
        if s.lower().startswith("han"):
            return "Han"
            
        if s.lower().startswith("greedo"):
            return "Greedo"

        if "don't understand" in s.lower():
            return "I don't understand this question"

        return s

    s = inpf[shot_col].apply(norm_shot).dropna()
    counts = s.value_counts()
    total  = counts.sum()

    print("\nRaw count\n")
    print(counts)

    order = ["Han", "Greedo", "I don't understand this question"]
    rows = []
    for k in order:
        n = counts.get(k, 0)
        rows.append((k, n / total if total else 0.0))

    shot_df = pd.DataFrame(rows, columns=["Answer", "Perc"])

    print("\nPercent table\n")
    display(shot_df)

    base = alt.Chart(shot_df)

    bars = base.mark_bar(size=24, color="#1f77b4").encode(
        x=alt.X("Perc:Q", axis=None, scale=alt.Scale(domain=[0, shot_df["Perc"].max()*1.1])),
        y=alt.Y("Answer:N", sort=order, axis=alt.Axis(title=""))
    )

    labels = base.mark_text(align="left", baseline="middle", dx=6, fontSize=12).encode(
        x="Perc:Q",
        y=alt.Y("Answer:N", sort=order),
        text=alt.Text("Perc:Q", format=".0%")
    )

    title_line = "Who shot first?"
    subtitle   = f"According to {total} respondents"

    subtext = alt.Chart(pd.DataFrame({'t':[subtitle]})).mark_text(
        align='left', font='Source Sans Pro', fontSize=13, dy=-10
    ).encode(text='t:N').properties(width=520)

    chart = (bars + labels).properties(width=520, height=160)

    vis = alt.vconcat(subtext, chart)\
             .resolve_scale(x='shared')\
             .properties(title=title_line)\
             .configure_title(font='Source Sans Pro', fontSize=18, anchor='start')\
             .configure_axis(labelFont='Source Sans Pro',
                             titleFont='Source Sans Pro')\
             .configure_view(strokeWidth=0)

    print("\nFinished genFirstShotVis\n")
    return vis

genFirstShotVis(sw)
```

```python


def _prep_age_seen(inpf, eps):

    print("\nRunning _prep_age_seen\n")

    seen_cols =[f"seen_{e}" for e in eps if f"seen_{e}" in inpf.columns]
    print("\nSeen columns detected\n", seen_cols)

    df =inpf.copy()
    df["num_seen"] = df[seen_cols].notna().sum(axis=1)

    out = df[["Age", "num_seen"]].dropna(subset=["Age"]).copy()

    def _age_key(a):
        a = str(a).strip()
        token = a.split('-')[0].strip(' +>')
        return (int(token) if token.isdigit() else 999, a)

    age_order = sorted(out["Age"].dropna().astype(str).unique(), key=_age_key)
    out["Age"] = pd.Categorical(out["Age"], categories=age_order, ordered=True)

    print("\nUnique Age categories (ordered)\n", list(out["Age"].cat.categories))
    print("\nSample of cleaned data\n")
    display(out.head(10))

    return out



'''chart_version_1'''


def genAgeSeenBar(inpf, eps):

    print("\nRunning genAgeSeenBar\n")

    df =_prep_age_seen(inpf, eps)

    prop_df = df.groupby(["Age", "num_seen"]).size().reset_index(name="count")
    total_df = df.groupby("Age").size().reset_index(name="total")
    merged   = prop_df.merge(total_df, on="Age")
    merged["Perc"] = merged["count"] / merged["total"]

    print("\nPreview of proportional table\n")
    
    display(merged.head(12))

    chartA =(
        alt.Chart(merged)
        .mark_bar()
        .encode(
            y=alt.Y("Age:N",
                    sort=list(df["Age"].cat.categories),
                    axis=alt.Axis(title="Age Group")),
            x=alt.X("Perc:Q",
                    stack="normalize",
                    axis=alt.Axis(format=".0%", title="Proportion within age group")),
            color=alt.Color("num_seen:O",
                            scale=alt.Scale(scheme="blues"),
                            legend=alt.Legend(title="# of Films Seen")),
            tooltip=["Age", "num_seen", alt.Tooltip("Perc:Q", format=".1%")]
        )
        .properties(width=520, height=260,
                    title="How many Star Wars films has each age group seen?")
        .configure_title(font='Source Sans Pro', fontSize=18, anchor='start')
        .configure_axis(labelFont='Source Sans Pro', titleFont='Source Sans Pro')
        .configure_view(strokeWidth=0)
    )

    print("\nCompleted genAgeSeenBar\n")
    return chartA



'''chart_version_2'''


def genAgeSeenStrip(inpf, eps):

    print("\nRunning genAgeSeenStrip\n")

    df =_prep_age_seen(inpf, eps)

    base = alt.Chart(df).transform_calculate(
        jitter="sqrt(-2*log(random()))*cos(2*PI*random())/10"
    )

    points = (
        base.mark_circle(size=40, opacity=0.45, color="#1f77b4")
        .encode(
            y=alt.Y("Age:N",
                    sort=list(df["Age"].cat.categories),
                    axis=alt.Axis(title="Age Group")),
            x=alt.X("num_seen:Q",
                    axis=alt.Axis(title="# of Films Seen"),
                    scale=alt.Scale(domain=[-0.5, 6.5])),
            xOffset="jitter:Q",
            tooltip=["Age", "num_seen"]
        )
    )

    medians =(
        alt.Chart(df)
        .mark_tick(color="black", thickness=2, size=28)
        .encode(
            y=alt.Y("Age:N", sort=list(df["Age"].cat.categories)),
            x=alt.X("median(num_seen):Q")
        )
    )

    chartB = (
        (points + medians)
        .properties(width=520, height=260,
                    title="Distribution of films seen by age (jittered points + median)")
        .configure_title(font='Source Sans Pro', fontSize=18, anchor='start')
        .configure_axis(labelFont='Source Sans Pro', titleFont='Source Sans Pro')
        .configure_view(strokeWidth=0)
    )

    print("\nCompleted genAgeSeenStrip\n")
    return chartB


chartA = genAgeSeenBar(sw, episodes)
display(chartA)

chartB = genAgeSeenStrip(sw, episodes)
display(chartB)
```
