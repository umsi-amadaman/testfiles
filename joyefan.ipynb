```python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    # create the various mapping and order lists/dictionaries
    # return: list of episodes (in order), episode->name map, list of episode names (in order)
    episodes = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    names = {
        'EI' : 'The Phantom Menace', 'EII' : 'Attack of the Clones', 'EIII' : 'Revenge of the Sith', 
        'EIV': 'A New Hope', 'EV': 'The Empire Strikes Back', 'EVI' : 'The Return of the Jedi'
    }

    # we're also going to use this order to sort, so names_l will now have our sort order
    return episodes, names, [names[ep] for ep in episodes]

episodes, namesDict, namesList = genEpisodeNamesDF()
```

```python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
```

```python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN
    salo = indf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    return(salo)
```

```python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
```

```python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total = len(inpf)
    percs = []

    # loop over each column and calculate the number of people who have seen the movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count them
    # and divide by the percent
    for seen_ep in ['seen_' + ep for ep in episodes]:
        perc = len(inpf[~ pd.isna(inpf[seen_ep])]) / total
        percs.append(perc)

    # at this point percs is holding our percentages

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes],percs))
    seen_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    return(seen_per_df)
```

```python
seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
```

```python
# let's see what's inside
seenPerMovieDF
```

```python
def genPercentVis(indf, names_l=namesList):
    # input: indf, the dataframe as seen_per_df
    # input: names_l - a list of all the movies in series order (default namesList)
    # output: simple altair bar chart
    
    # ok, time to make the chart... let's make a bar chart (use mark_bar)
    bars = alt.Chart(indf).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        )
    )

    # at this point we don't really have a great plot (it's missing the annotations, titles, etc.)
    return(bars)

seenPerMovieVis = genPercentVis(seenPerMovieDF)
```

```python
def augmentPercentVis(base):
    # input: base (the base vis, i.e., bars as above)
    # we're going to overlay the text with the percentages, so let's make another visualization
    # that's just text labels


    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    seen_movies = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=180
    ).properties(
        # add a title
        title="Which 'Star Wars' Movies Have you Seen?"
    )

    return(seen_movies)

# note that we are NOT formatting this in the Five Thirty Eight Style yet... we'll leave that to you to figure out
```

### 2.2 How people rate the 'Star Wars' movie? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genRateVis(inpf,eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above

    def get_ranking_percentage(inpf, names):
        """Calculating each movie ranking parcentage in Top Third / Middle Third / Bottom Third"""
        total = len(inpf)
        percs = []
        
        for k, v in names.items():
            used_col = f"rank_{k}"
            rank_12 = len(inpf[inpf[used_col].astype(int).isin([1, 2])])
            rank_34 = len(inpf[inpf[used_col].astype(int).isin([3, 4])])
            rank_56 = len(inpf[inpf[used_col].astype(int).isin([5, 6])])
            percs.append((v, 'Top Third', rank_12 / total))
            percs.append((v, 'Middle Third', rank_34 / total))
            percs.append((v, 'Bottom Third', rank_56 / total))
        
        movie_rankings = pd.DataFrame(percs, columns = ['Name', 'Rank', 'Percentage'])
        
        return movie_rankings

    def fill_missing_rank(row, columns):
        """Fill NAs in a row based on ranking rules"""
        values = row[columns]
        na_count = values.isna().sum()

        # IF ONLY ONE MISSING, FILL NA WITH THE MISSING RANK
        if na_count == 1:
            non_na_values = set(values.dropna())
            missing_ranks = {'1', '2', '3', '4', '5', '6'} - non_na_values
            fill_value = missing_ranks.pop() if len(missing_ranks) == 1 else '99'
        # IF MORE THAN ONE MISSING, WE DON'T KNOW WHICH RANK TO FILL, FILL 99 FOR FUTURE EXCLUDING
        else:
            fill_value = '99'

        # FILL NA WITH THE FILL_VALUE
        for col in columns:
            if pd.isna(row[col]):
                row[col] = fill_value
        
        return row
        

    # RANKING COLUMNS
    rank_cols = ['rank_' + ep for ep in eps]
    
    # GET PEOPLE WHO HAVE SEEN ALL MOVIES
    df_all = inpf.dropna(subset=['seen_' + ep for ep in eps],how='any').copy()

    # CHOOSE NEEDED COLUMNS
    df_rank = df_all[rank_cols].copy()
    
    # HANDLE ROWS WITH MISSING RANK
    mask = df_rank[rank_cols].isna().any(axis=1)
    df_rank.loc[mask] = df_rank.loc[mask].apply(
        lambda row: fill_missing_rank(row, rank_cols), 
        axis=1
    )

    # CALCULATE EACH MOVIE RANKING 1 PERCENTAGE
    movie_rankings = get_ranking_percentage(df_rank, names)

    # CREATING THE CHART
    base = alt.Chart(movie_rankings).encode(
        x = alt.X(
            'Percentage:Q',
            title = None,
            axis = None,
            scale = alt.Scale(domain = [0, 0.7])
        ),
        y = alt.Y(
            'Name:N',
            title = None,
            axis = alt.Axis(labelFontSize = 11, ticks = False, domain = False),
            sort = names_l
        ),
        color = alt.Color(
            'Rank:N',
           scale = alt.Scale(
               domain = ['Top Third', 'Middle Third', 'Bottom Third'],
               range = ['#7cb342', '#42a5f5', '#ef5350']
           ),
           legend = None
        )
    ).properties(
        width = 150, 
        height = 220
    )
    
    bars = base.mark_bar(size = 20)
    
    text = base.mark_text(
        align = 'left',
        baseline = 'middle',
        dx = 3,
        fontSize = 10,
    ).encode(text = alt.Text('Percentage:Q', format = '.0%'))
    
    final_chart = (bars + text).facet(
        column = alt.Column(
            'Rank:N',
             title = None,
             header = alt.Header(labelFontSize = 13, labelFontWeight = 'bold', labelAlign = 'center'),
             sort=['Top Third', 'Middle Third', 'Bottom Third']
        )
    ).properties(
        title={
            "text": "How People Rate the 'Star Wars' Movies",
            "subtitle": "How often each film was rated in the top, middle and bottom third (by 471 respondents who have seen all six films)",
            "fontSize": 18,
            "fontWeight": "bold",
            "subtitleFontSize": 12,
            "subtitleColor": "gray",
            "anchor": "start"
        }
    ).configure_view(strokeWidth = 0)
    
    return final_chart
    
    # YOUR CODE HERE
    # raise NotImplementedError()
```

```python
# Recreate this image using altair here (10 POINTS)

def genFavorabilityVis(inpf):
    # input: inpf, the star wars dataset (as defined at top of file)
    # output: the Altair visualization as described above

    def categorize_favorability(text):
        """Categoize each input text to favorable level"""
        try:
            text = text.lower()
            if 'neutral' in text:
                return 'Neutral'
            elif 'unfavorably' in text:
                return 'Unfavorable'
            elif 'favorably' in text:
                return 'Favorable'
            elif 'unfamiliar' in text:
                return 'Unfamiliar'
            else:
                return text
        # NULL VALUE RETURN NULL BACK
        except:
            return text

    def get_favorability_percentage(inpf, characters, favorable_level):
        """Calculating each character's favorability parcentage in each level"""
        percs = []
        
        for character in characters:
            # COUNT TOTAL EXCLUDING NULL VALUES
            total = len(inpf[~inpf[character].isna()])
            for level in favorable_level:
                count = len(inpf[inpf[character] == level])
                percs.append((character, level, count/total))
        
        favorability_percentage = pd.DataFrame(percs, columns = ['Name', 'Level', 'Percentage'])
        
        return favorability_percentage

    # FIND PEOPLE WHO HAVE SEEN AT LEAST ONE MOVIE
    eps = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    df_one = inpf.dropna(subset=['seen_' + ep for ep in eps], how = 'all').copy()
    
    # CHOOSE NEEDED COLUMNS
    character_lst = [
        'Han Solo', 'Luke Skywalker', 'Princess Leia Organa', 'Anakin Skywalker',
        'Obi Wan Kenobi', 'Emperor Palpatine', 'Darth Vader', 'Lando Calrissian',
        'Boba Fett', 'C-3P0', 'R2 D2', 'Jar Jar Binks', 'Padme Amidala', 'Yoda'
    ]
    
    df_one = df_one[character_lst]

    # RE-CATEGORIZE ALL FAVORABILITY
    for character in character_lst:
        df_one[character] = df_one[character].apply(lambda x: categorize_favorability(x))

    # CALCULATE EACH CHARACTOR'S FAVORABILITY PERCENTAGE
    favorable_level = ['Favorable', 'Neutral', 'Unfavorable', 'Unfamiliar']
    favorability_percentage = get_favorability_percentage(df_one, character_lst, favorable_level)

    # GENERATE CHARACTOR LIST ORDER BY LEVEL = Favorable
    character_order = list(favorability_percentage[favorability_percentage['Level'] == 'Favorable'].sort_values('Percentage', ascending = False)['Name'])

    # CREATING THE CHART
    base = alt.Chart(favorability_percentage).encode(
        x = alt.X(
            'Percentage:Q',
            title = None,
            axis = None,
            scale = alt.Scale(domain = [0, 1])
        ),
        y = alt.Y(
            'Name:N',
            title = None,
            axis = alt.Axis(labelFontSize = 11, domain = False),
            sort = character_order
        ),
        color = alt.Color(
            'Level:N',
           scale = alt.Scale(
               domain = favorable_level,
               range = ['#7cb342', '#42a5f5', '#ef5350', '#808080']
           ),
           legend = None
        )
    ).properties(
        width = 120, 
        height = 300
    )
    
    bars = base.mark_bar(size = 15)
    
    text = base.mark_text(
        align = 'left',
        baseline = 'middle',
        dx = 3,
        fontSize = 10,
    ).encode(text = alt.Text('Percentage:Q', format = '.0%'))
    
    final_chart = (bars + text).facet(
        column = alt.Column(
            'Level:N',
             title = None,
             header = alt.Header(labelFontSize = 13, labelFontWeight = 'bold', labelAlign = 'center'),
             sort=favorable_level
        )
    ).properties(
        title={
            "text": "'Star Wars' Characters Favorability Ratings",
            "subtitle": f"by {len(df_one)} respondents",
            "fontSize": 18,
            "fontWeight": "bold",
            "subtitleFontSize": 12,
            "subtitleColor": "gray",
            "anchor": "start"
        }
    ).configure_view(strokeWidth = 0)

    return final_chart
    # YOUR CODE HERE
    # raise NotImplementedError()
```
