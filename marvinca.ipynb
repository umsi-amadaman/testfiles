A pie chart could be useful only in limited circumstances, essentially when you want to show parts of a single whole rather than compare multiple items side by side.

A pie chart might be effective if we were showing one movie at a time. For example, to display what percentage of people have seen vs not seen “The Phantom Menace,” a single pie chart could easily depict that binary breakdown (seen = 80%, not seen = 20%). Pie charts are most effective when there are few categories and the goal is to highlight proportions that sum to 100%.

A pie chart would be ineffective when comparing multiple movies across respondents. Multiple pie charts make it difficult to compare relative slice sizes accurately. The current bar chart emphasizes comparative magnitude between films, making it easier to analyze differences across categories. With several pies, those comparisons would become unclear.

In summary, bar charts are better for comparing categories (movies) by a quantitative measure (percent seen), while pie charts are only useful for showing a single categorical breakdown, not for side by side comparison across multiple categories.

```python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    # create the various mapping and order lists/dictionaries
    # return: list of episodes (in order), episode->name map, list of episode names (in order)
    episodes = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    names = {
        'EI' : 'The Phantom Menace', 'EII' : 'Attack of the Clones', 'EIII' : 'Revenge of the Sith', 
        'EIV': 'A New Hope', 'EV': 'The Empire Strikes Back', 'EVI' : 'The Return of the Jedi'
    }

    # we're also going to use this order to sort, so names_l will now have our sort order
    return episodes, names, [names[ep] for ep in episodes]

episodes, namesDict, namesList = genEpisodeNamesDF()
```

```python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
```

```python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN
    salo = indf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    return(salo)
```

```python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
```

```python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total = len(inpf)
    percs = []

    # loop over each column and calculate the number of people who have seen the movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count them
    # and divide by the percent
    for seen_ep in ['seen_' + ep for ep in episodes]:
        perc = len(inpf[~ pd.isna(inpf[seen_ep])]) / total
        percs.append(perc)

    # at this point percs is holding our percentages

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes],percs))
    seen_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    return(seen_per_df)
```

```python
seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
```

```python
# let's see what's inside
seenPerMovieDF
```

```python
def genPercentVis(indf, names_l=namesList):
    # input: indf, the dataframe as seen_per_df
    # input: names_l - a list of all the movies in series order (default namesList)
    # output: simple altair bar chart
    
    # ok, time to make the chart... let's make a bar chart (use mark_bar)
    bars = alt.Chart(indf).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        )
    )

    # at this point we don't really have a great plot (it's missing the annotations, titles, etc.)
    return(bars)

seenPerMovieVis = genPercentVis(seenPerMovieDF)
```

```python
def augmentPercentVis(base):
    # input: base (the base vis, i.e., bars as above)
    # we're going to overlay the text with the percentages, so let's make another visualization
    # that's just text labels


    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    seen_movies = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=180
    ).properties(
        # add a title
        title="Which 'Star Wars' Movies Have you Seen?"
    )

    return(seen_movies)

# note that we are NOT formatting this in the Five Thirty Eight Style yet... we'll leave that to you to figure out
```

```python
# Recreate this image using Altair
# match the "538 style" as best you can (hint: look at the altair lab at the start of the semester)

def genBestVis(inpf,eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above

    seen_cols = [f'seen_{e}' for e in eps]
    rank_cols = [f'rank_{e}' for e in eps]
    seen_all = inpf.dropna(subset=seen_cols, how='any').copy()
    denom = len(seen_all)
    
    df = seen_all.dropna(subset=rank_cols, how='any').copy()
    for c in rank_cols:
        df[c] = pd.to_numeric(df[c], errors='coerce')
    
    long = df.melt(value_vars=rank_cols, var_name='ep', value_name='rank')
    long['ep'] = long['ep'].str.replace('rank_', '', regex=False)
    long['Name'] = long['ep'].map(names)
    
    fav_counts = long[long['rank'] == 1].groupby('Name').size()
    
    data_list = []
    for name in names_l:
        count = fav_counts.get(name, 0)
        pct = count / denom
        data_list.append({'Name': name, 'Percentage': pct})
    
    out = pd.DataFrame(data_list)
    
    FTE_BLUE = '#30a2da'
    FTE_GRAY = '#f0f0f0'
    TEXT_GRAY = '#999999'
    TITLE_GRAY = '#000000'
    
    bars = alt.Chart(out).mark_bar(size=20, color=FTE_BLUE).encode(
        x=alt.X('Percentage:Q', scale=alt.Scale(domain=[0, 0.40]), axis=None),
        y=alt.Y('Name:N', sort=names_l, axis=alt.Axis(
            title=None, 
            labelAngle=0, 
            tickSize=8,
            domain=False,
            labelColor=TEXT_GRAY,
            labelFontSize=11,
            tickColor='#cccccc',
            labelPadding=12
        ))
    )
    
    text = bars.mark_text(
        align='left', 
        baseline='middle', 
        dx=3,
        fontSize=11,
        color='#333333',
        fontWeight='bold'
    ).encode(
        text=alt.Text('Percentage:Q', format='.0%')
    )
    
    chart = (bars + text).properties(
        width=600,
        height=180,
        title={
            "text": "What's the best 'Star Wars' Movie?",
            "subtitle": f"Of {denom} respondents who have seen all six films",
            "anchor": "start",
            "fontSize": 18,
            "fontWeight": 'bold',
            "color": TITLE_GRAY,
            "subtitleFontSize": 12,
            "subtitleColor": '#000000',
            "subtitlePadding": 3
        }
    ).configure_view(
        strokeWidth=0,
        fill=FTE_GRAY
    ).configure_axis(
        grid=False,
        domain=False
    ).configure_scale(
        bandPaddingInner=0.2
    )
    
    return chart
```

```python
# Recreate this image using altair here (10 POINTS)

def genRateVis(inpf,eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above

    seen_cols = [f'seen_{e}' for e in eps]
    rank_cols = [f'rank_{e}' for e in eps]
    
    seen_all = inpf.dropna(subset=seen_cols, how='any').copy()
    denom = len(seen_all)
    
    df = seen_all.dropna(subset=rank_cols, how='any').copy()
    for c in rank_cols:
        df[c] = pd.to_numeric(df[c], errors='coerce')
    
    long = df.melt(value_vars=rank_cols, var_name='ep', value_name='rank')
    long['ep'] = long['ep'].str.replace('rank_', '', regex=False)
    long['Name'] = long['ep'].map(names)
    
    def categorize_rank(rank):
        if pd.isna(rank):
            return None
        elif rank <= 2:
            return 'Top third'
        elif rank <= 4:
            return 'Middle third'
        else:
            return 'Bottom third'
    
    long['Category'] = long['rank'].apply(categorize_rank)
    long = long.dropna(subset=['Category'])
    
    category_counts = long.groupby(['Name', 'Category']).size().reset_index(name='Count')
    
    total_counts = long.groupby('Name').size().reset_index(name='Total')
    category_counts = category_counts.merge(total_counts, on='Name')
    category_counts['Percentage'] = category_counts['Count'] / category_counts['Total']
    
    category_order = ['Top third', 'Middle third', 'Bottom third']
    color_map = {
        'Top third': '#6bab5a',
        'Middle third': '#30a2da', 
        'Bottom third': '#fc4f30'
    }
    
    data1 = category_counts[category_counts['Category'] == 'Top third']
    data2 = category_counts[category_counts['Category'] == 'Middle third']
    data3 = category_counts[category_counts['Category'] == 'Bottom third']
    
    bar1 = alt.Chart(data1, title=alt.TitleParams('Top third', fontSize=14, fontWeight='bold', color='#2e2e2e', anchor='middle')).mark_bar(size=22, color=color_map['Top third']).encode(
        x=alt.X('Percentage:Q', axis=None, scale=alt.Scale(domain=[0, 0.7])),
        y=alt.Y('Name:N', sort=names_l, title='', axis=alt.Axis(labelAngle=0, labelFontSize=13, labelColor='#6d6d6d', tickSize=0, domain=False))
    ).properties(width=160, height=180)
    
    text1 = bar1.mark_text(align='left', baseline='middle', dx=3, fontSize=13, color='#2e2e2e', fontWeight='normal').encode(
        text=alt.Text('Percentage:Q', format='.0%')
    )
    
    bar2 = alt.Chart(data2, title=alt.TitleParams('Middle third', fontSize=14, fontWeight='bold', color='#2e2e2e', anchor='middle')).mark_bar(size=22, color=color_map['Middle third']).encode(
        x=alt.X('Percentage:Q', axis=None, scale=alt.Scale(domain=[0, 0.7])),
        y=alt.Y('Name:N', sort=names_l, axis=None)
    ).properties(width=160, height=180)
    
    text2 = bar2.mark_text(align='left', baseline='middle', dx=3, fontSize=13, color='#2e2e2e', fontWeight='normal').encode(
        text=alt.Text('Percentage:Q', format='.0%')
    )
    
    bar3 = alt.Chart(data3, title=alt.TitleParams('Bottom third', fontSize=14, fontWeight='bold', color='#2e2e2e', anchor='middle')).mark_bar(size=22, color=color_map['Bottom third']).encode(
        x=alt.X('Percentage:Q', axis=None, scale=alt.Scale(domain=[0, 0.7])),
        y=alt.Y('Name:N', sort=names_l, axis=None)
    ).properties(width=160, height=180)
    
    text3 = bar3.mark_text(align='left', baseline='middle', dx=3, fontSize=13, color='#2e2e2e', fontWeight='normal').encode(
        text=alt.Text('Percentage:Q', format='.0%')
    )
    
    chart = alt.hconcat(
        bar1 + text1, bar2 + text2, bar3 + text3,
        spacing=10,
        title=alt.TitleParams(
            text="How People Rate the 'Star Wars' Movies",
            subtitle=["How often each film was rated in the top, middle and bottom third",
                     f"(by {denom} respondents who have seen all six films)"],
            anchor='start',
            fontSize=22,
            fontWeight='bold',
            color='#2e2e2e',
            subtitleFontSize=14,
            subtitleColor='#6d6d6d',
            subtitlePadding=5
        )
    ).configure_axis(
        grid=False,
        domain=False
    ).configure_view(
        stroke='#f0f0f0'
    ).configure(
        background='#f0f0f0'
    ).configure_scale(
        bandPaddingInner=0.2
    )
    
    return chart
```

#### Justification
*Provide your justification here* to argue why your "better" version is actually better. Make sure to defend this based on expressiveness and effectiveness as well as perceptual arguments.

I compare how Star Wars fans and non-fans rate characters by computing “favorable” only among respondents who expressed an opinion. I exclude “Unfamiliar/Not sure” so the percentages aren’t diluted by people who don’t know the character. I split the population using the survey’s fan field (case-insensitive “yes”) and include the 14 named characters from the dataset. For the grouped bar version, I sort characters by their average favorable share across groups. For the slope version, I keep the same data rules but plot Non-Fans on the left and Fans on the right.

The first visualization is a grouped bar chart. For each character, Fans and Non Fans are shown side-by-side, using color to encode the group and a common zero baseline on the x-axis for the favorable percentage. This design makes the central task “for Character X, do fans like them more than non-fans, and by how much?” immediately readable. Judging bar length against a shared baseline is a highly accurate perceptual task, so the viewer can quickly compare magnitudes and also read precise values from the axis or tooltips. I also surface the denominator used (“N among those with an opinion”) for transparency.

The second visualization is a slope graph that connects each character’s Non-Fan percentage to their Fan percentage with a line. It emphasizes direction and change. If most lines tilt upward, you instantly see that fans tend to rate characters more favorably. This is useful for communicating overall sentiment and, when sorted appropriately, can highlight which characters are most polarizing. However, with many characters the colored lines and legend create clutter, and comparing two separated positions (left vs. right axis) is heavier and less precise than comparing bar lengths from a shared baseline.

Given the goal of clear, character by character comparison, the grouped bar chart is the better choice. It aligns with how readers naturally scan for specific characters, supports fast and accurate magnitude judgments, and preserves numeric precision. The slope graph offers a perspective that’s strongest when the story centers on the magnitude and direction of change—e.g., highlighting disagreement or overall upward shifts among fans—but for this task the grouped bars provide the most effective view.
