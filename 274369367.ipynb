YOUR ANSWER HERE

1. **When The Colleague Might Be Right**
    * **Situation**: To show the proportion of people who have seen one single, specific film out of the entire survey group.

    * **Reasoning**: If the goal is to make a standalone graphic for, say, "The Empire Strikes Back" to highlight that 91% of the audience has seen it, a single pie chart could work. It would dramatically visualize the overwhelming majority (91% Seen vs. 9% Not Seen). In this case, the part-to-whole relationship is very clear with only two segments.

2. **When The Colleague Would Be Wrong**

    **i. If the goal is comparison, not part-to-whole for each Item**
    * **Bar Chart (Current)**: Good for comparison. The aligned baselines (either on the left for a horizontal bar chart, or the bottom for a vertical one) make it easy to judge the relative lengths of the bars. You can instantly see that "Empire" is the longest and "Revenge of the Sith" is the shortest, and you can calculate the exact differences between two (or more) films.

    * **Pie Chart (Proposed)**: Not good for comparison. To compare values across multiple pie charts, the viewer must compare the angles or areas of slices in different circles that are not aligned. This is a much more mentally demanding task and is highly inaccurate. Is the "Clones" slice bigger or smaller than the "Sith" slice? It's very difficult to tell.
    
    **ii. There are too many categories for a single pie chart**
    * If the colleague suggests a single pie chart with a slice for each movie, this is a well-known "chart junk" anti-pattern. It's like you would have six slices, many of similar size (e.g., 68%, 66%, 73%). So it becomes nearly impossible to rank the movies or perceive precise differences between them. The chart would be a visual mess and effectively useless.

   **iii. Pie charts waste space and lack a clear order**
    * **Bar Chart:** It's compact and efficient. The movies can be ordered by percentage (which they are, roughly, from highest to lowest), making the pattern even clearer.
    
    * **Pie Charts:** Multiple pie charts take up more space to convey the same information. There is no inherent ordering in a pie chart, so the logical ranking of the data is lost.

````python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    # create the various mapping and order lists/dictionaries
    # return: list of episodes (in order), episode->name map, list of episode names (in order)
    episodes = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    names = {
        'EI' : 'The Phantom Menace', 'EII' : 'Attack of the Clones', 'EIII' : 'Revenge of the Sith', 
        'EIV': 'A New Hope', 'EV': 'The Empire Strikes Back', 'EVI' : 'The Return of the Jedi'
    }

    # we're also going to use this order to sort, so names_l will now have our sort order
    return episodes, names, [names[ep] for ep in episodes]

episodes, namesDict, namesList = genEpisodeNamesDF()
````

````python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
````

````python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN
    salo = indf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    return(salo)
````

````python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
````

````python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total = len(inpf)
    percs = []

    # loop over each column and calculate the number of people who have seen the movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count them
    # and divide by the percent
    for seen_ep in ['seen_' + ep for ep in episodes]:
        perc = len(inpf[~ pd.isna(inpf[seen_ep])]) / total
        percs.append(perc)

    # at this point percs is holding our percentages

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes],percs))
    seen_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    return(seen_per_df)
````

````python
seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
````

````python
# let's see what's inside
seenPerMovieDF
````

````python
def genPercentVis(indf, names_l=namesList):
    # input: indf, the dataframe as seen_per_df
    # input: names_l - a list of all the movies in series order (default namesList)
    # output: simple altair bar chart
    
    # ok, time to make the chart... let's make a bar chart (use mark_bar)
    bars = alt.Chart(indf).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        )
    )

    # at this point we don't really have a great plot (it's missing the annotations, titles, etc.)
    return(bars)

seenPerMovieVis = genPercentVis(seenPerMovieDF)
````

````python
def augmentPercentVis(base):
    # input: base (the base vis, i.e., bars as above)
    # we're going to overlay the text with the percentages, so let's make another visualization
    # that's just text labels


    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    seen_movies = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=180
    ).properties(
        # add a title
        title="Which 'Star Wars' Movies Have you Seen?"
    )

    return(seen_movies)

# note that we are NOT formatting this in the Five Thirty Eight Style yet... we'll leave that to you to figure out
````

````python
# Recreate this image using altair here (10 POINTS)

def genRateVis(inpf,eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above
    
    # YOUR CODE HERE
    # raise NotImplementedError()
    def _genRankPercentDf():
        nonlocal inpf, eps, names, names_l
        # I'm using this func to transform rank to rank tier
        def get_tier(x):
            try:
                x = int(x)
                if x in [1, 2]:
                    return 'Top third'
                elif x in [3, 4]:
                    return 'Middle third'
                else:
                    return 'Bottom third'
            except:
                return np.nan
        # filter df
        seen_all = inpf.dropna(subset=[f'seen_{ep}' for ep in eps], how='any')
        total = len(seen_all)
        print(f'Number of respondents who have seen all six films: {total}')
        # melt the df in order to label the rank
        melt_vars = [f'rank_{ep}' for ep in eps]
        seen_all_melt = seen_all.melt(value_vars=melt_vars,var_name='rank_ep',value_name='rank')
        seen_all_melt['name'] = seen_all_melt['rank_ep'].apply(lambda x: names[x.split('_')[-1]])
        seen_all_melt['tier'] = seen_all_melt['rank'].apply(get_tier)
        seen_all_melt['tier_num'] = [1]*len(seen_all_melt)
        print(seen_all_melt)
        # then pivot the df and calculate the tier percents
        seen_all_piv = seen_all_melt.pivot_table(index='name',columns='tier',values='tier_num',aggfunc='count')
        seen_all_piv['total'] =seen_all_piv.sum(1)
        print(seen_all_piv)
        tier_result = seen_all_piv[seen_all_piv.columns[:-1]].apply(lambda x: x / seen_all_piv['total'])
        tier_result = tier_result[['Top third','Middle third','Bottom third']].loc[names_l]
        print(tier_result)
        return tier_result, total
    
    def _augmentPercentVis(base, total):
        # mark text
        text = base.mark_text(
            align='left',
            baseline='middle',
            dx=3,
            fontWeight='bold'
        ).encode(
            text=alt.Text('Percentage:Q',format='.0%'),
        )

        # layered charts
        rank_movies = (base + text).facet(
            # split charts horizontally
            column=alt.Column(
                'rank_tier:N', 
                header=alt.Header(
                    title='', # do not present header
                    labelFontWeight='bold' # label font being bold
                ), 
                sort=['Top third', 'Middle third', 'Bottom third'], # sort chart order
            )
        ).configure(
            # background is set at the very first stage
            background='hsl(0, 0%, 95%)'
        ).configure_view(
            # we don't want a stroke around the bars
            stroke=None,
            strokeWidth=0 
        ).configure_scale(
            # add some padding
            bandPaddingInner=0.2
        ).properties(
            # paramerter title is a dict
            title={
                'text': "How People Rate the 'Star Wars' Movies",
                'subtitle': ['How often each film was rated in the top, middle and bottom third',
                             f'(by {total} respondents who have seen all six films)'],
                'font': 'Arial',
                'fontSize': 22,
                'fontWeight': 900,
                'subtitleFontSize': 11, 
                'subtitleFontWeight': "bold",
                'anchor': 'start'
            },
        ).configure_axisY(
            # use hsl to control gray
            domainColor='hsl(0, 0%, 85%)', 
            tickColor='hsl(0, 0%, 85%)',
            labelColor='hsl(0, 0%, 60%)',
            labelFontWeight='bold'
        ).configure_axisX(
            title=None
        ).configure_title(
            dy=-20, # move up titles a bit
        )
    
        return(rank_movies)
        
    tier_result, total = _genRankPercentDf()
    melted_df = tier_result.reset_index().melt(id_vars='name', 
                                value_vars=['Top third', 'Middle third', 'Bottom third'],
                                var_name='rank_tier', 
                                value_name='Percentage')
    # generate base bars
    bars = alt.Chart(melted_df).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage:Q',
            axis=None
        ),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'name:N',
             axis=alt.Axis(tickCount=5, title=''),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        ),
        color=alt.Color(
            'rank_tier:N',
            scale=alt.Scale(
               domain=['Top third', 'Middle third', 'Bottom third'],
               range=['hsl(120, 45%, 45%)', '#008fd5', 'hsl(0, 100%, 50%)']
            ),
            legend=None
        ),
    ).properties(
        # width and height should be defined before facet
        width=80,height=180
    ) 
    # get ranked bars
    tier_pct =  _augmentPercentVis(bars, total)
    return tier_pct
    
    
````

The bar chart generated by ```genAgeBarVis(sw, episodes, label='Age')``` is better than the pie chart generated by ```genAgePieVis(sw, episodes, label='Age')```.

**1. Expressiveness**

    Both charts are expressing the same thing: how do different generations rate each Start War film. Viewers can get the same information from both charts, e.g., people whose ages greater than 60 like The Phantom Memnace most.

**2. Effectiveness**

    Although viewers are possible to get information from both charts, the bar chart is more effective than the pie chart:
    
    a. the bar chart enables viewers quickly identify different films and generations, while the pie chart is making things complicated since viewers have to continously scan each pie and then refer to (or recall) the coresponding legends in order to figure out the meaning of colors;
    
    b. When it comes to making comparison among different generations, the bar chart provides a more straightforward way for viewers. By comparing the length and color among bars, viewers will know how ratings differ among ages in a short time. Besides, consider the situation where values are hidden in both charts, viewers can still learn the same thing from the bar chart but they might fail to make it in pie chart since comparing the angles requires more mental calculations.

**3. Perception**

    From Gestalt psychology we learn that we perceive things in the simplest, most regular, and symmetric way possible. Elements that are close to each other are perceived as a group, and similar elements (like shape, color, size) are perceived as belonging together. Viewers will preattentively distinguish generations by looking at bars with same colors. However, pie chart contains too many segmented elements (e.g. six pies with four colors), which brings challenges for preattentive processing. Therefore, the bar chart outperforms the pie chart in terms of perception.
