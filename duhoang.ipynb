```python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    # create the various mapping and order lists/dictionaries
    # return: list of episodes (in order), episode->name map, list of episode names (in order)
    episodes = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    names = {
        'EI' : 'The Phantom Menace', 'EII' : 'Attack of the Clones', 'EIII' : 'Revenge of the Sith', 
        'EIV': 'A New Hope', 'EV': 'The Empire Strikes Back', 'EVI' : 'The Return of the Jedi'
    }

    # we're also going to use this order to sort, so names_l will now have our sort order
    return episodes, names, [names[ep] for ep in episodes]

episodes, namesDict, namesList = genEpisodeNamesDF()
```

```python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
```

```python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN
    salo = indf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    return(salo)
```

```python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
```

```python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total = len(inpf)
    percs = []

    # loop over each column and calculate the number of people who have seen the movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count them
    # and divide by the percent
    for seen_ep in ['seen_' + ep for ep in episodes]:
        perc = len(inpf[~ pd.isna(inpf[seen_ep])]) / total
        percs.append(perc)

    # at this point percs is holding our percentages

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes],percs))
    seen_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    return(seen_per_df)
```

```python
seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
```

```python
# let's see what's inside
seenPerMovieDF
```

```python
def genPercentVis(indf, names_l=namesList):
    # input: indf, the dataframe as seen_per_df
    # input: names_l - a list of all the movies in series order (default namesList)
    # output: simple altair bar chart
    
    # ok, time to make the chart... let's make a bar chart (use mark_bar)
    bars = alt.Chart(indf).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        )
    )

    # at this point we don't really have a great plot (it's missing the annotations, titles, etc.)
    return(bars)

seenPerMovieVis = genPercentVis(seenPerMovieDF)
```

```python
# display it
seenPerMovieVis
```

```python
def augmentPercentVis(base):
    # input: base (the base vis, i.e., bars as above)
    # we're going to overlay the text with the percentages, so let's make another visualization
    # that's just text labels


    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    seen_movies = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=180
    ).properties(
        # add a title
        title="Which 'Star Wars' Movies Have you Seen?"
    )

    return(seen_movies)

# note that we are NOT formatting this in the Five Thirty Eight Style yet... we'll leave that to you to figure out
```

```python
# Recreate this image using altair here (10 POINTS)

def genRateVis(inpf,eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above
    
    # YOUR CODE HERE
    # print(names_l)
    # print("--- Dropping rows where people have NOT seen at least one movie -> seen_ == NaN ---") # debugging, comment out before submitting
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs and get the total count.
    # find people who have at least on of the columns (seen_*) not NaN.
    inpf = inpf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    # display(inpf) # debugging, comment out before submitting
    
    # print("--- Dropping rows where people have NOT seen at all six movies -> seen_ + ep == NaN ---") # debugging, comment out before submitting
    inpf_all_six = inpf.dropna(subset=['seen_' + ep for ep in eps])
    # display(inpf_all_six) # debugging, comment out before submitting
    
    total = len(inpf_all_six) # Get the total number of respondents for this subset. 
    percs = []
    rank_cols =  ['rank_' + ep for ep in eps]  # get all rank columns.
    
    # print(rank_cols) # debugging, comment out before submitting

    # print("--- Calculate the Top, Middle, and Bottom Third rankings for each film ---") # debugging, comment out before submitting
    for rank_ep in rank_cols:
        # print(rank_ep) # debugging, comment out before submitting

        all_percs = []
        top_third = ((inpf_all_six[rank_ep] == "1") | (inpf_all_six[rank_ep] == "2")).sum() # Top Third count all of the times respondents listed movie (rank_ep) as number 1 or 2.
        mid_third = ((inpf_all_six[rank_ep] == "3") | (inpf_all_six[rank_ep] == "4")).sum() # Top Third count all of the times respondents listed movie (rank_ep) as number 3 or 4.
        bottom_third = ((inpf_all_six[rank_ep] == "5") | (inpf_all_six[rank_ep] == "6")).sum() # Top Third count all of the times respondents listed movie (rank_ep) as number 5 or 6.
        
        # print(top_third) # debugging, comment out before submitting
        # print(mid_third) # debugging, comment out before submitting
        # print(bottom_third) # debugging, comment out before submitting
        
        top_third_perc = (top_third / total) # Calculate the percentage top third.
        mid_third_perc = (mid_third / total) # Calculate the percentage top third
        bottom_third_perc = (bottom_third / total) # Calculate the percentage top third
        all_percs = [top_third_perc, mid_third_perc, bottom_third_perc] # create a list of all percentages.
        percs.append(all_percs)        
        

    # at this point percs is holding our percentages
    # print(percs) # debugging, comment out before submitting

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes], percs))
    # print(tuples)

    # Need to extract the percentages within the list, then transform each element of that list into individual elements of the new tuple.
    # The new_tuple will then be able to used to make the dataframe.
    new_tuples = []
    for i in tuples:
        # print(i) # debugging, comment out before submitting
        i2 = (i[0], *i[1])
        # print(i2) # debugging, comment out before submitting
        new_tuples.append(i2)
    # print(new_tuples) # debugging, comment out before submitting
    
    df_favs = pd.DataFrame(new_tuples, columns = ['Name', 'Top third', 'Middle third', 'Bottom third'])
    # display(df_favs) # debugging, comment out before submitting
    # print(df_favs) # debugging, comment out before submitting

    # Since Altair requires a long format, we need to convert the df_favs into a long table.
    df_favs_melted = df_favs.melt(id_vars='Name', var_name='Category', value_name='Percentage')
    # print(df_favs_melted) # debugging, comment out before submitting
    df_favs_melted['Name'] = pd.Categorical(df_favs_melted['Name'], categories=names_l, ordered=True)

    # print("--- Generate the Chart ---") # debugging, comment out before submitting
    # Set the bar colors - used https://htmlcolorcodes.com/ to try and match colors from example.
    bar_colors = {
        'Top third': '#13911A',
        'Middle third': '#1F77B4',
        'Bottom third': '#CF0C0C'
    }

    # Generate a base chart.
    base = alt.Chart(df_favs_melted).encode(
        y=alt.Y(
            'Name:N',
            # Domain=False -> hides the y-axis vertical line, ticks=False -> hides the ticks marks, labelPadding=10 -> increase the space between variable and bar (in pixels).
             axis=alt.Axis(tickCount=5, title='', labelFontSize=14, domain=False, ticks=False, labelPadding=10),
            sort=names_l)
    ).properties(
        width=180,
        height=500
    )

    bars = base.mark_bar().encode(
        x=alt.X(
            'Percentage:Q',
            axis=None,
        ),
        # Set the colors of each bars using the bar_color dictionary.
        color=alt.Color('Category:N',
                        scale=alt.Scale(domain=list(bar_colors.keys()),
                                        range=list(bar_colors.values())),
                        legend=None
        )        
    )

    # Set text layer option:
    # aligns text to the left of the bar, shows percentage value next to each bar, colors the text to the bar color.
    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=4,
        color='black'
    ).encode(
        x='Percentage:Q',
        y=alt.Y('Name:N', sort=names_l, title=None),
        text=alt.Text('Percentage:Q', format='.0%'),
        # Set the colors of each text using the bar_color dictionary.
        color=alt.Color('Category:N',
                        scale=alt.Scale(domain=list(bar_colors.keys()),
                                        range=list(bar_colors.values())),
                        legend=None)
    )

    # Combines the bar chart and text layers.
    layer = alt.layer(bars, text).facet(
        column=alt.Column('Category:N', 
                          sort=['Top third', 'Middle third', 'Bottom third'], 
                          title=None)
    ).properties(
        # Set the title and subtitle. 
        title={
            "text": "How People Rate the 'Star Wars' Movies",     
            "subtitle": [
                "How often each film was rated in the top, middle and bottom third", 
                "(by 471 respondents who have seen all six films)"
            ], 
            "anchor": "start",  # align title and subtitle to the left.                                  
            "offset": 30        # add 30 pixels below the subtitle before the bar graphic starts.
        }
    )

    # Set: the background to grey, title/subtitle font size, bar column headers font size and made it bold, removed box round bar chart graphic.
    final = (
        layer.configure_axis(grid=False).configure(background='#E5E5E5').configure_title(fontSize=24, subtitleFontSize=14).configure_header(labelFontSize=14, labelFontWeight='bold').configure_view(stroke=None)
    )

    return(final)
    
    # raise NotImplementedError()
```

In short, while both visualizations are equally expressive, the pie chart is more effective for this task because it aligns better with the viewer's perceptual strengths and provides a more intuitive understanding of how each fan group contributes to the whole.
