```python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    # create the various mapping and order lists/dictionaries
    # return: list of episodes (in order), episode->name map, list of episode names (in order)
    episodes = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    names = {
        'EI' : 'The Phantom Menace', 'EII' : 'Attack of the Clones', 'EIII' : 'Revenge of the Sith', 
        'EIV': 'A New Hope', 'EV': 'The Empire Strikes Back', 'EVI' : 'The Return of the Jedi'
    }

    # we're also going to use this order to sort, so names_l will now have our sort order
    return episodes, names, [names[ep] for ep in episodes]

episodes, namesDict, namesList = genEpisodeNamesDF()
```

```python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
```

```python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN
    salo = indf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    return(salo)
```

```python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
```

```python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total = len(inpf)
    percs = []

    # loop over each column and calculate the number of people who have seen the movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count them
    # and divide by the percent
    for seen_ep in ['seen_' + ep for ep in episodes]:
        perc = len(inpf[~ pd.isna(inpf[seen_ep])]) / total
        percs.append(perc)

    # at this point percs is holding our percentages

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes],percs))
    seen_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    return(seen_per_df)
```

```python
seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
```

```python
def genPercentVis(indf, names_l=namesList):
    # input: indf, the dataframe as seen_per_df
    # input: names_l - a list of all the movies in series order (default namesList)
    # output: simple altair bar chart
    
    # ok, time to make the chart... let's make a bar chart (use mark_bar)
    bars = alt.Chart(indf).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        )
    )

    # at this point we don't really have a great plot (it's missing the annotations, titles, etc.)
    return(bars)

seenPerMovieVis = genPercentVis(seenPerMovieDF)
```

```python
def augmentPercentVis(base):
    # input: base (the base vis, i.e., bars as above)
    # we're going to overlay the text with the percentages, so let's make another visualization
    # that's just text labels


    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    seen_movies = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=180
    ).properties(
        # add a title
        title="Which 'Star Wars' Movies Have you Seen?"
    )

    return(seen_movies)

# note that we are NOT formatting this in the Five Thirty Eight Style yet... we'll leave that to you to figure out
```

```python
def prepareData():

    #drop NaN values
    filtered_data = sw.dropna(subset=['seen_EI', 'seen_EII', 'seen_EIII', 'seen_EIV', 'seen_EV', 'seen_EVI'], how='any')

    #extract the rankings columns
    movie_rankings = filtered_data[['rank_EI', 'rank_EII', 'rank_EIII', 'rank_EIV', 'rank_EV', 'rank_EVI']] 
    movie_rankings.columns = seenPerMovieDF['Name']

    #melt rows into columns for easier readability
    long_format_rankings = movie_rankings.melt()
    
    
    grouped_rankings = long_format_rankings.groupby(['Name', 'value']).size().reset_index(name='Count')

    #pivot the dataframe for analysis
    rankings_pivot = grouped_rankings.pivot(index='Name', columns='value', values='Count').reset_index()

    #calculate percentages for top, middle and bottom thirds
    total_respondents = len(filtered_data)
    rankings_pivot['Top Third'] = (rankings_pivot['1'] + rankings_pivot['2']) / total_respondents
    rankings_pivot['Middle Third'] = (rankings_pivot['3'] + rankings_pivot['4']) / total_respondents
    rankings_pivot['Bottom Third'] = (rankings_pivot['5'] + rankings_pivot['6']) / total_respondents

    #create df to show summary of rankings
    rankings_summary = rankings_pivot[['Name', 'Top Third', 'Middle Third', 'Bottom Third']]

    return rankings_summary
```

```python
def create_ranking_chart(data, column_name, color, title):
    chart = alt.Chart(data).mark_bar(size=20, color=color).encode(
        x=alt.X(column_name, axis=None),
        y=alt.Y('Name:N', sort=namesList, axis=alt.Axis(tickCount=5, labelColor='#8a9499', tickSize=10, tickColor='#bbc1c4', labelPadding=5, domainColor='#bbc1c4'))
    ).properties(title={'text':title, "fontSize": 11, 'dy': -5, 'fontWeight': 'normal'})
    text = chart.mark_text(align='left', baseline='middle', fontWeight='normal', color=color, dx=3).encode(
        text=alt.Text(f'{column_name}:Q', format='.0%')
    )

    return (text + chart).properties(width=70, height=180)
```

```python
# Recreate this image using altair here (10 POINTS)

def genRateVis(inpf,eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above

    #create charts for each third
    top_third_chart = create_ranking_chart(rankings_summary, 'Top Third', '#47a244', 'Top Third')
    middle_third_chart = create_ranking_chart(rankings_summary, 'Middle Third', '#1798d9', 'Middle Third').encode(
        y=alt.Y('Name:N', sort=namesList, axis=None))
    bottom_third_chart = create_ranking_chart(rankings_summary, 'Bottom Third', '#ff1a1a', 'Bottom Third').encode(
        y=alt.Y('Name:N', sort=namesList, axis=None))

    final_chart = alt.hconcat(top_third_chart, middle_third_chart, bottom_third_chart, spacing=40).properties(
        background='#f5f5f5',
        title={
            "text":"How People Rate the 'Star Wars' Movies",
            "subtitle": "by 471 respondents who have seen all six films",
            "fontSize": 24,
            "subtitleFontSize": 14,
            "anchor": "start"
        }
    ).configure_title(fontSize=24, dy=-15).configure_view(stroke=None) 

    return final_chart
    
    return rankings_summary
```

### 2.3 Star Wars' Characters Favorability Ratings. Recreate the above image using altair (10 Points)

```python
def favorite_character_data():

    favorite_characters = sw[sw.columns[15:29]]

    #clean and reshape data for analysis
    favorite_characters_melted = favorite_characters.melt(var_name='Character', value_name='Vote')
    favorite_characters_melted.dropna(inplace=True)

    #aggregate votes for each character
    character_votes = favorite_characters_melted.groupby(['Character', 'Vote']).size().reset_index(name='Count')

    #pivot data to get counts for each vote type per character
    character_summary = character_votes.pivot(index='Character', columns='Vote', values='Count').reset_index()

    character_summary.columns = ['Name', 'Very unfavorably', 'Somewhat unfavorably', 'Neutral', 'Somewhat favorably', 'Very favorably', 'Unfamiliar (N/A)']

    #calculate proportions
    character_summary['Total'] = character_summary.sum(axis=1)
    character_summary['Favorable'] = (character_summary['Very favorably'] + character_summary['Somewhat favorably']) / character_summary['Total']
    character_summary['Unfavorable'] = (character_summary['Very unfavorably'] + character_summary['Somewhat unfavorably']) / character_summary['Total']
    character_summary['Neutral'] = (character_summary['Neutral']) / character_summary['Total']
    character_summary['Unfamiliar'] = (character_summary['Unfamiliar (N/A)']) / character_summary['Total']

    #Keep relevant columns
    character_summary = character_summary[['Name', 'Favorable', 'Neutral', 'Unfavorable', 'Unfamiliar']]
    character_sort_order = list(character_summary['Name'])
    
    return character_summary
```

REDUCING VISUAL CLUTTER:
The thing we are trying to compare is the number of people of each gender who have seen the films AND consider themselves fans. Since 100% of the people in this visualization have seen the films, that information would be better included in the legend or perhaps a subtitle.

The second visualization moves the "Seen at least one film" category to its own area of the chart so the two genders can be compared directly.


MORE EFFECTIVE COLOR USAGE:
The first visualization separates all the data by gender, using one color to code both the percentage that have seen the movies *and* the percent that consider themselves fans. That color grouping is distracting because our pre-cognitive brains are comparing the length of the bars *within* the genders instead of *between* the genders.

The second visualization color codes female as blue and male as maize, but puts the genders next to each other, grouped by the questions that are being asked about each gender. The viewer can then more efficiently and effectively, pre-cognitively compare the lengths of the gender-colored bars to see the comparative answer to the each question individually.
