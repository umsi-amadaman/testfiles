```python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    # create the various mapping and order lists/dictionaries
    # return: list of episodes (in order), episode->name map, list of episode names (in order)
    episodes = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    names = {
        'EI' : 'The Phantom Menace', 'EII' : 'Attack of the Clones', 'EIII' : 'Revenge of the Sith', 
        'EIV': 'A New Hope', 'EV': 'The Empire Strikes Back', 'EVI' : 'The Return of the Jedi'
    }

    # we're also going to use this order to sort, so names_l will now have our sort order
    return episodes, names, [names[ep] for ep in episodes]

episodes, namesDict, namesList = genEpisodeNamesDF()
```

```python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
```

```python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN
    salo = indf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    return(salo)
```

```python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
```

```python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total = len(inpf)
    percs = []

    # loop over each column and calculate the number of people who have seen the movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count them
    # and divide by the percent
    for seen_ep in ['seen_' + ep for ep in episodes]:
        perc = len(inpf[~ pd.isna(inpf[seen_ep])]) / total
        percs.append(perc)

    # at this point percs is holding our percentages

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes],percs))
    seen_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    return(seen_per_df)
```

```python
seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
```

```python
# let's see what's inside
seenPerMovieDF
```

```python
def genPercentVis(indf, names_l=namesList):
    # input: indf, the dataframe as seen_per_df
    # input: names_l - a list of all the movies in series order (default namesList)
    # output: simple altair bar chart
    
    # ok, time to make the chart... let's make a bar chart (use mark_bar)
    bars = alt.Chart(indf).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        )
    )

    # at this point we don't really have a great plot (it's missing the annotations, titles, etc.)
    return(bars)

seenPerMovieVis = genPercentVis(seenPerMovieDF)
```

```python
def augmentPercentVis(base):
    # input: base (the base vis, i.e., bars as above)
    # we're going to overlay the text with the percentages, so let's make another visualization
    # that's just text labels


    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    seen_movies = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=180
    ).properties(
        # add a title
        title="Which 'Star Wars' Movies Have you Seen?"
    )

    return(seen_movies)

# note that we are NOT formatting this in the Five Thirty Eight Style yet... we'll leave that to you to figure out
```

```python
# correct subset: people who have SEEN all six films
def getSeenAllSixDF(indf, eps):
    df = indf.replace(r'^\s*$', np.nan, regex=True)   # clean empty strings
    seen_cols = [f'seen_{ep}' for ep in eps]
    return df.dropna(subset=seen_cols, how='any')

seenAllSixDF = getSeenAllSixDF(sw, episodes)
print("n who ranked all six:", len(seenAllSixDF))  # should be ~471

# % who ranked each movie #1 among that group
def genBestMoviePercentDf(indf, namesDict, episodes):
    total = len(indf)
    rows = []
    for ep in episodes:
        col = f'rank_{ep}'
        top1 = (pd.to_numeric(indf[col], errors='coerce') == 1).sum()
        rows.append((namesDict[ep], top1 / total * 100.0))
    return pd.DataFrame(rows, columns=['Name','Percentage'])

bestPerMovieDF = genBestMoviePercentDf(seenAllSixDF, namesDict, episodes)
bestPerMovieDF.sort_values('Percentage', ascending=False)
```

```python

def bestMovieBar(df):
    base = alt.Chart(df).encode(
        y=alt.Y('Name:N', sort='-x', title=''),
        x=alt.X('Percentage:Q',
                title='Percent of respondents (n = 471)',
                scale=alt.Scale(domain=[0,100]),
                axis=alt.Axis(format='.0f'))
    )

    bars = base.mark_bar(size=20)
    text = base.mark_text(align='left', baseline='middle', dx=3).encode(
        text=alt.Text('Percentage:Q', format='.0f')
    )

    chart = (bars + text).properties(
        width=520, height=180,
        title="What’s the Best ‘Star Wars’ Movie?\nOf 471 respondents who have seen all six films"
    ).configure_view(strokeWidth=0)

    return chart

bestMovieBar(bestPerMovieDF)
```

## Why this is the “good” version 
    Uses aligned lengths/positions (bar length on a common baseline), which humans judge more accurately than angles/areas.
    Sorted by value to support quick comparisons.
    Single hue avoids implying extra categorical meaning.
    Direct labels reduce eye movement and reading load.
   Axis starts at 0 to avoid exaggeration

```python
def genBestVis(inpf, eps, names=namesDict, names_l=namesList):
    # --- subset to people who have SEEN all six films (matches 538 figure) ---
    df = inpf.replace(r'^\s*$', np.nan, regex=True)       
    seen_cols = [f'seen_{ep}' for ep in eps]
    df = df.dropna(subset=seen_cols, how='any')
    total = len(df)                                       

    # --- compute % who ranked each movie as #1 ---
    rows = []
    for ep in eps:
        col = f'rank_{ep}'
        top1 = (pd.to_numeric(df[col], errors='coerce') == 1).sum()
        rows.append({'Name': names[ep], 'Percentage': 100.0 * top1 / total})

    best_df = pd.DataFrame(rows)

    # --- Altair chart: match 538-style as closely as practical ---
    base = alt.Chart(best_df)

    bars = base.mark_bar(size=22, color='#1f77b4').encode(
        x=alt.X('Percentage:Q',
                title=f'Percent of respondents (n = {total})',
                scale=alt.Scale(domain=[0,100]),
                axis=alt.Axis(format='.0f')),
        y=alt.Y('Name:N',
                sort='-x',
                title='')
    )

    labels = base.mark_text(align='left', baseline='middle', dx=4).encode(
        text=alt.Text('Percentage:Q', format='.0f')
    )

    chart = (bars + labels).properties(
        width=560, height=200,
        title=alt.TitleParams(
            text="What’s the Best ‘Star Wars’ Movie?",
            subtitle=f"Of {total} respondents who have seen all six films"
        )
    ).configure_view(
        strokeWidth=0
    ).configure_axis(
        grid=False, domain=False, labelColor='#444', titleColor='#444'
    ).configure_title(
        fontSize=16, subtitleFontSize=12
    )

    return chart
```

```python
# Recreate this image using altair here (10 POINTS)

#def genRateVis(inpf,eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above
    #raise NotImplementedError()
    # YOUR CODE HERE


def genRateVis(inpf, eps, names=namesDict, names_l=namesList):
    # --- subset: respondents who have SEEN all six films (matches 538 figure) ---
    df = inpf.replace(r'^\s*$', np.nan, regex=True)
    seen_cols = [f'seen_{ep}' for ep in eps]
    df = df.dropna(subset=seen_cols, how='any')

    # --- compute % in top/middle/bottom third for each movie ---
    rows = []
    for ep in eps:
        r = pd.to_numeric(df[f'rank_{ep}'], errors='coerce')
        top    = (r.isin([1, 2])).mean() * 100
        middle = (r.isin([3, 4])).mean() * 100
        bottom = (r.isin([5, 6])).mean() * 100
        full = names[ep]
        rows += [
            {"Name": full, "Tier": "Top third",    "Percentage": top,    "Label": f"{top:.0f}"},
            {"Name": full, "Tier": "Middle third", "Percentage": middle, "Label": f"{middle:.0f}"},
            {"Name": full, "Tier": "Bottom third", "Percentage": bottom, "Label": f"{bottom:.0f}"},
        ]
    rate_df = pd.DataFrame(rows)

    # helper to build one column per tier
    def tier_col(df_tier, color, title):
        base = alt.Chart(df_tier).encode(
            y=alt.Y('Name:N', sort=names_l, title='')
        )
        bars = base.mark_bar(size=16, color=color).encode(
            x=alt.X('Percentage:Q', axis=None)
        )
        labels = base.mark_text(align='left', baseline='middle', dx=3).encode(
            x='Percentage:Q', text='Label:N'
        )
        return (bars + labels).properties(width=180, height=230, title=title)

    top_ch    = tier_col(rate_df[rate_df.Tier == "Top third"],    '#2ca02c', 'Top third')
    middle_ch = tier_col(rate_df[rate_df.Tier == "Middle third"], '#1f77b4', 'Middle third')
    bottom_ch = tier_col(rate_df[rate_df.Tier == "Bottom third"], '#d62728', 'Bottom third')

    chart = alt.hconcat(top_ch, middle_ch, bottom_ch).properties(
        title=alt.TitleParams(
            text="How People Rate the ‘Star Wars’ Movies",
            subtitle="How often each film was rated in the top, middle and bottom third\n(by respondents who have seen all six films)"
        )
    ).configure_view(
        strokeWidth=0, fill='#f2f2f2'
    ).configure_title(
        fontSize=16, subtitleFontSize=12
    )

    return chart


```

### 2.3 Star Wars' Characters Favorability Ratings. Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

#def genFavorabilityVis(inpf):
    # input: inpf, the star wars dataset (as defined at top of file)
    # output: the Altair visualization as described above
    
    # YOUR CODE HERE
    #raise NotImplementedError()import numpy as np
import pandas as pd
import altair as alt

def genFavorabilityVis(inpf):
    # canonical names we want to show (order will follow 'Favorable' % later)
    canonical = [
        'Luke Skywalker','Han Solo','Princess Leia Organa','Obi Wan Kenobi','Yoda','R2-D2','C-3PO',
        'Anakin Skywalker','Darth Vader','Lando Calrissian','Padme Amidala','Boba Fett',
        'Emperor Palpatine','Jar Jar Binks'
    ]

    # known aliases -> canonical
    alias_map = {
        'R2 D2': 'R2-D2', 'R2D2': 'R2-D2',
        'C3PO': 'C-3PO', 'C 3PO': 'C-3PO',
        'Padne Amidala': 'Padme Amidala', 'Padmé Amidala': 'Padme Amidala',
        'Obi-Wan Kenobi': 'Obi Wan Kenobi'
    }

    # build a mapping from CANONICAL -> ACTUAL COLUMN NAME in dataframe
    cols = list(inpf.columns)
    # helper to normalize text for fuzzy match
    def norm(s): return str(s).lower().replace('-', '').replace(' ', '')
    canon_to_actual = {}
    for c in canonical:
        # exact
        if c in inpf.columns:
            canon_to_actual[c] = c
            continue
        # alias table
        aliased = [k for k, v in alias_map.items() if v == c and k in inpf.columns]
        if aliased:
            canon_to_actual[c] = aliased[0]
            continue
        # soft match
        target = norm(c)
        candidates = [col for col in cols if norm(col) == target]
        if candidates:
            canon_to_actual[c] = candidates[0]
            continue
     
    chars = [c for c in canonical if c in canon_to_actual]

    # buckets (adjust if your raw strings differ)
    favorable_vals   = {'Very favorably','Somewhat favorably'}
    neutral_vals     = {'Neither'}
    unfavorable_vals = {'Somewhat unfavorably','Very unfavorably'}
    unfamiliar_vals  = {'Unfamiliar (N/A)','Unfamiliar','Not familiar','Unfamiliar with this person'}

    rows = []
    for ch in chars:
        col = canon_to_actual[ch]
        s = inpf[col].astype('string')
        # denominator: respondents who answered THIS character (non-missing)
        denom = s.dropna().shape[0]
        if denom == 0:
            continue

        fav   = s.isin(favorable_vals).sum()
        neu   = s.isin(neutral_vals).sum()
        unfav = s.isin(unfavorable_vals).sum()
        unkn  = s.isin(unfamiliar_vals).sum()

        pct = lambda x: 100.0 * x / denom
        rows += [
            {'Character': ch, 'Bucket': 'Favorable',   'Percentage': pct(fav),   'Label': f'{pct(fav):.0f}'},
            {'Character': ch, 'Bucket': 'Neutral',     'Percentage': pct(neu),   'Label': f'{pct(neu):.0f}'},
            {'Character': ch, 'Bucket': 'Unfavorable', 'Percentage': pct(unfav), 'Label': f'{pct(unfav):.0f}'},
            {'Character': ch, 'Bucket': 'Unfamiliar',  'Percentage': pct(unkn),  'Label': f'{pct(unkn):.0f}'},
        ]

    df = pd.DataFrame(rows)

    # order by Favorable %
    order = (df[df['Bucket']=='Favorable']
             .sort_values('Percentage', ascending=False)['Character'].tolist())

    def bucket_col(bucket_name, color, title):
        sub = df[df['Bucket'] == bucket_name]
        base = alt.Chart(sub).encode(
            y=alt.Y('Character:N', sort=order, title='')
        )
        bars = base.mark_bar(size=16, color=color).encode(
            x=alt.X('Percentage:Q', axis=None)
        )
        labels = base.mark_text(align='left', baseline='middle', dx=3).encode(
            x='Percentage:Q', text='Label:N'
        )
        return (bars + labels).properties(width=160, height=285, title=title)

    favorable_ch   = bucket_col('Favorable',   '#2ca02c', 'Favorable')
    neutral_ch     = bucket_col('Neutral',     '#1f77b4', 'Neutral')
    unfavorable_ch = bucket_col('Unfavorable', '#d62728', 'Unfavorable')
    unfamiliar_ch  = bucket_col('Unfamiliar',  '#7f7f7f', 'Unfamiliar')

    chart = alt.hconcat(
        favorable_ch, neutral_ch, unfavorable_ch, unfamiliar_ch
    ).properties(
        title=alt.TitleParams(
            text="'Star Wars' Characters Favorability Ratings",
            subtitle="By 834 respondents"
        )
    ).configure_view(
        strokeWidth=0, fill='#f2f2f2'
    ).configure_title(
        fontSize=16, subtitleFontSize=12
    )

    return chart
```

### 2.4 Who shot first? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

#def genFirstShotVis(inpf):
    # input: inpf, the star wars dataset
    # output: the Altair visualization as described above
    
    # YOUR CODE HERE


import numpy as np
import pandas as pd
import altair as alt

def genFirstShotVis(inpf):
    # 1) find the column for the question (robust to wording)
    cols = list(inpf.columns)
    lc = [c.lower() for c in cols]
    cand_idx = [i for i, c in enumerate(lc)
                if ('shot' in c and 'first' in c) or ('who shot first' in c)]
    if not cand_idx:
        raise ValueError("Could not find the 'Who shot first?' column.")
    col = cols[cand_idx[0]]

    # 2) normalize responses to the three canonical options
    s = inpf[col].astype('string')
    s = s.str.strip()

    def canon(x):
        if x is None or pd.isna(x):
            return np.nan
        z = str(x).lower()
        if 'han' in z:
            return 'Han'
        if 'greedo' in z:
            return 'Greedo'
        if "don" in z and "understand" in z:
            return "I don't understand this question"
        # fallbacks sometimes used in datasets
        if 'not' in z and 'understand' in z:
            return "I don't understand this question"
        return np.nan

    canon_resp = s.map(canon)
    denom = canon_resp.dropna().shape[0]

    counts = canon_resp.value_counts(dropna=True)
    # ensure all three categories exist
    index = ["Han", "Greedo", "I don't understand this question"]
    counts = counts.reindex(index, fill_value=0)

    df = (counts / denom * 100.0).reset_index()
    df.columns = ['Answer', 'Percentage']

    # 3) build the chart
    base = alt.Chart(df)

    bars = base.mark_bar(size=22, color='#1f77b4').encode(
        x=alt.X('Percentage:Q',
                title=f'Percent of respondents (n = {denom})',
                scale=alt.Scale(domain=[0, 100]),
                axis=alt.Axis(format='.0f')),
        y=alt.Y('Answer:N', sort='-x', title='')
    )

    labels = base.mark_text(align='left', baseline='middle', dx=4).encode(
        text=alt.Text('Percentage:Q', format='.0f')
    )

    chart = (bars + labels).properties(
        width=560, height=160,
        title=alt.TitleParams(
            text="Who Shot First?",
            subtitle=f"According to {denom} respondents"
        )
    ).configure_view(
        strokeWidth=0, fill='#f2f2f2'
    ).configure_axis(
        grid=False, domain=False, tickColor='#888',
        labelColor='#444', titleColor='#444'
    ).configure_title(
        fontSize=16, subtitleFontSize=12
    )

    return chart



  
```
