```python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    # create the various mapping and order lists/dictionaries
    # return: list of episodes (in order), episode->name map, list of episode names (in order)
    episodes = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    names = {
        'EI' : 'The Phantom Menace', 'EII' : 'Attack of the Clones', 'EIII' : 'Revenge of the Sith', 
        'EIV': 'A New Hope', 'EV': 'The Empire Strikes Back', 'EVI' : 'The Return of the Jedi'
    }

    # we're also going to use this order to sort, so names_l will now have our sort order
    return episodes, names, [names[ep] for ep in episodes]

episodes, namesDict, namesList = genEpisodeNamesDF()
```

```python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
```

```python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN
    salo = indf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    return(salo)
```

```python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
```

```python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total = len(inpf)
    percs = []

    # loop over each column and calculate the number of people who have seen the movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count them
    # and divide by the percent
    for seen_ep in ['seen_' + ep for ep in episodes]:
        perc = len(inpf[~ pd.isna(inpf[seen_ep])]) / total
        percs.append(perc)

    # at this point percs is holding our percentages

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes],percs))
    seen_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    return(seen_per_df)
```

```python
seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
```

```python
# let's see what's inside
seenPerMovieDF
```

```python
def genPercentVis(indf, names_l=namesList):
    # input: indf, the dataframe as seen_per_df
    # input: names_l - a list of all the movies in series order (default namesList)
    # output: simple altair bar chart
    
    # ok, time to make the chart... let's make a bar chart (use mark_bar)
    bars = alt.Chart(indf).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        )
    )

    # at this point we don't really have a great plot (it's missing the annotations, titles, etc.)
    return(bars)

seenPerMovieVis = genPercentVis(seenPerMovieDF)
```

```python
# display it
seenPerMovieVis
```

```python
def augmentPercentVis(base):
    # input: base (the base vis, i.e., bars as above)
    # we're going to overlay the text with the percentages, so let's make another visualization
    # that's just text labels


    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    seen_movies = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=180
    ).properties(
        # add a title
        title="Which 'Star Wars' Movies Have you Seen?"
    )

    return(seen_movies)

# note that we are NOT formatting this in the Five Thirty Eight Style yet... we'll leave that to you to figure out
```

### 2.2 How people rate the 'Star Wars' movie? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genRateVis(inpf,eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above

    ##similar data filtering, except now I need percentages not of rank=1, but 1-2,3-4,5-6
    df = inpf.dropna(subset = ['seen_EI', 'seen_EII','seen_EIII', 'seen_EIV', 'seen_EV', 'seen_EVI'])
    ##now count total records, and divide the rank numbers to get %'s-
    seen_all_count = df.shape[0]
    target_cols = df.columns[9:15]
    ##turn the ranking datat from object to numeric
    for col in target_cols:
        df[col] = pd.to_numeric(df[col], errors = 'coerce')
    ## collect the top ratings and then divide by the total count of ppl who've seen all
    top_third = ((df[target_cols].isin([1,2])).sum() / seen_all_count).to_dict()
    ## now create our DF with the Names needed of the movies
    tuples = list(zip([names[ep] for ep in eps],[value for value in top_third.values()]))
    rate_df = pd.DataFrame(tuples, columns = ['Name', 'Top third'])
    ##now we need the other two categories, but I realized I need to have a mapping of namesDict, but the inverse
    invert_names ={v: k for k,v in namesDict.items()}
    ### now I'm adding the column that will match our mid and bottom third
    rate_df['Ep'] = rate_df['Name'].map(invert_names)
    rate_df['Ep'] = 'rank_' + rate_df['Ep'].astype(str)
    ###now get the mid and bottom third and map them into the dataframe
    mid_third = ((df[target_cols].isin([3,4])).sum() / seen_all_count).to_dict()
    but_third = ((df[target_cols].isin([5,6])).sum() / seen_all_count).to_dict()

    rate_df['Middle third'] = rate_df['Ep'].map(mid_third)
    rate_df['Bottom third'] = rate_df['Ep'].map(but_third)
    #print(rate_df.head(6))
    ##now transform the dataset into a 'long' version by collapsing the thirds into 1 column designation
    new_df = (
        rate_df.melt(
            id_vars = 'Name',
            value_vars = ['Top third','Middle third','Bottom third'],
            var_name = 'tier',
            value_name = 'pct'
        )
    )
    
    #print(new_df)
    
    ## it's a similar vertical bar chart, with the main differences being adding the grouped data separately, and the colors to the bars
    ## I'll start with our two layers, the bars and the text (i'mma recopy for my own benefit rather than re-using the above)
    
    base = alt.Chart(new_df).mark_bar().encode(
        y = alt.Y('Name:N', axis = alt.Axis(tickCount=5, title=''), sort = names_l),
        x = alt.X('pct:Q', axis=None),
        color = alt.Color('tier:O', scale =alt.Scale(range=['red','blue','green']),legend = None)
        
    )
    
        
    text = alt.Chart(new_df).mark_text(
        align='left',
        baseline='middle',
        dx=3,  # Nudges text to right so it doesn't appear on top of the bar
        fontSize = 14,
        color = 'black'
    ).encode(
        # we'll use the percentage as the text
        y = alt.Y('Name:N', axis = alt.Axis(tickCount=5, title=''), sort = names_l),
        x = alt.X('pct:Q', axis=None),
        text=alt.Text('pct:Q',format='.0%')
    )
    
    combo = (text + base)
    combo1 = combo.properties(
        # set the dimensions of the visualization
        width=60,
        height=200
    )
    
    new = combo1.facet(column = alt.Column(
        'tier:O',
        sort = ['Top third','Middle third','Bottom third'],
        title=None,
        header= alt.Header(labelFontSize = 14, labelFontWeight = 'bold')),
        spacing = 55
    
    ).configure_view(
        stroke=None    
    ).configure_facet(
        spacing=2
    ).resolve_scale(x='shared'
    ).configure_axisY(
        domain = False,
        ticks = False,
        labelFontSize = 16,
        labelPadding = 10
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.1
    ).properties(
        title={
         'text': "How Do People Rate the 'Star Wars' Movies",
        'subtitle': ['How often each film was rated in the top, middle, and bottom third', '(by 471 Respondents who have seen all six films)']},
        background = '#f5f5f5',
        padding = {"top": 10, "left": 10, "right": 10, "bottom": 10}    
    ).configure_title(
        anchor = 'start',
        fontSize = 24,
        subtitleFontSize = 18,
        offset=10
    )
    
    return new
    
```

### 2.3 Star Wars' Characters Favorability Ratings. Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genFavorabilityVis(inpf):
    # input: inpf, the star wars dataset (as defined at top of file)
    # output: the Altair visualization as described above
    ##now count total records, and divide the rank numbers to get %'s-
    df = inpf[inpf['seen_any_movie']=='Yes']
    df = df.dropna(subset = df.columns[15:29])
    total_responders = df.shape[0]
    ##we need to transform our dataset into 3 columns, the characters, the favorable/unfavorable etc., and then counts of those votes.  then we can divide by total count to get %s
    target_cols = df.columns[15:29]
    df = (
        df[target_cols]
        .melt(var_name = 'character', value_name = 'opinion')
        .value_counts(['character','opinion'])
        .reset_index(name='count')
               )
    ##annoyingly- the articles favorable mappings aren't the same as the dataset and the profs didn't provide a mapping so I will have to make one and collapse the data again
    fav_mapping = {'Very favorably': 'Favorable', 'Somewhat favorably': 'Favorable',
       'Neither favorably nor unfavorably (neutral)': 'Neutral', 'Unfamiliar (N/A)': 'Unfamiliar',
       'Very unfavorably': 'Unfavorable', 'Somewhat unfavorably': 'Unfavorable'}
    new_df = (
        df.assign(opinion = lambda d: d['opinion'].map(fav_mapping))
        .groupby(['character', 'opinion'], as_index=False)['count'].sum()
    )
    
    
    new_df['Pct'] = new_df['count']/ total_responders
    #return new_df.sample(10)
    ##alright I think the data is ready- chart is very similar to above example
    #I'm going to create a sort order first
    order = (new_df[new_df['opinion']== 'Favorable']
             .sort_values('Pct', ascending=False)['character']
             .tolist()
            )
       
    base = alt.Chart(new_df).mark_bar().encode(
        y = alt.Y('character:N', axis = alt.Axis(tickCount=5, title=''), sort = order),
        x = alt.X('Pct:Q', axis=None),
        color = alt.Color('opinion:O', scale =alt.Scale(range=['green','blue','lightgrey','red']),legend = None)
        
    )
    
        
    text = alt.Chart(new_df).mark_text(
        align='left',
        baseline='middle',
        dx=3,  # Nudges text to right so it doesn't appear on top of the bar
        fontSize = 14,
        color = 'black'
    ).encode(
        # we'll use the percentage as the text
        y = alt.Y('character:N', axis = alt.Axis(tickCount=5, title=''),sort = order),
        x = alt.X('Pct:Q', axis=None),
        text=alt.Text('Pct:Q',format='.0%')
    )
    
    combo = (text + base)
    combo1 = combo.properties(
        # set the dimensions of the visualization
        width=100,
        height=550
    )
    
    new = combo1.facet(column = alt.Column(
        'opinion:O',
        sort = ['Favorable','Neutral','Unfavorable','Unfamiliar'],
        title=None,
        header= alt.Header(labelFontSize = 14, labelFontWeight = 'bold')),
        spacing = 20
    
    ).configure_view(
        stroke=None    
    ).configure_facet(
        spacing=5
    ).resolve_scale(x='shared'
    ).configure_axisY(
        domain = False,
        ticks = False,
        labelFontSize = 16,
        labelPadding = 10
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.1
    ).properties(
        title={
        'text': "'Star Wars' Characters Favorability Ratings",
        'subtitle': [f"by {total_responders} Respondents"]},
        background = '#f5f5f5',
        padding = {"top": 10, "left": 10, "right": 10, "bottom": 10}    
    ).configure_title(
        anchor = 'start',
        fontSize = 24,
        subtitleFontSize = 18,
        offset=10
    )
    
    return new
  
```
