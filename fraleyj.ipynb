```python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    # create the various mapping and order lists/dictionaries
    # return: list of episodes (in order), episode->name map, list of episode names (in order)
    episodes = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    names = {
        'EI' : 'The Phantom Menace', 'EII' : 'Attack of the Clones', 'EIII' : 'Revenge of the Sith', 
        'EIV': 'A New Hope', 'EV': 'The Empire Strikes Back', 'EVI' : 'The Return of the Jedi'
    }

    # we're also going to use this order to sort, so names_l will now have our sort order
    return episodes, names, [names[ep] for ep in episodes]

episodes, namesDict, namesList = genEpisodeNamesDF()
```

```python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
```

```python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN
    salo = indf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    return(salo)
```

```python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
```

```python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total = len(inpf)
    percs = []

    # loop over each column and calculate the number of people who have seen the movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count them
    # and divide by the percent
    for seen_ep in ['seen_' + ep for ep in episodes]:
        perc = len(inpf[~ pd.isna(inpf[seen_ep])]) / total
        percs.append(perc)

    # at this point percs is holding our percentages

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes],percs))
    seen_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    return(seen_per_df)
```

```python
seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
```

```python
# let's see what's inside
seenPerMovieDF
```

```python
def genPercentVis(indf, names_l=namesList):
    # input: indf, the dataframe as seen_per_df
    # input: names_l - a list of all the movies in series order (default namesList)
    # output: simple altair bar chart
    
    # ok, time to make the chart... let's make a bar chart (use mark_bar)
    bars = alt.Chart(indf).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        )
    )

    # at this point we don't really have a great plot (it's missing the annotations, titles, etc.)
    return(bars)

seenPerMovieVis = genPercentVis(seenPerMovieDF)
```

```python
# display it
seenPerMovieVis
```

```python
def augmentPercentVis(base):
    # input: base (the base vis, i.e., bars as above)
    # we're going to overlay the text with the percentages, so let's make another visualization
    # that's just text labels


    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    seen_movies = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=180
    ).properties(
        # add a title
        title="Which 'Star Wars' Movies Have you Seen?"
    )

    return(seen_movies)

# note that we are NOT formatting this in the Five Thirty Eight Style yet... we'll leave that to you to figure out
```

### 2.2 How people rate the 'Star Wars' movie? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genRateVis(inpf,eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above
    ranko = inpf.dropna(subset=['rank_' + ep for ep in eps],how='all')
    top = {'rank_EI': 0, 'rank_EII': 0, 'rank_EIII': 0, 'rank_EIV': 0, 'rank_EV': 0, 'rank_EVI': 0}
    mid = {'rank_EI': 0, 'rank_EII': 0, 'rank_EIII': 0, 'rank_EIV': 0, 'rank_EV': 0, 'rank_EVI': 0}
    bot = {'rank_EI': 0, 'rank_EII': 0, 'rank_EIII': 0, 'rank_EIV': 0, 'rank_EV': 0, 'rank_EVI': 0}
    rank_cols = ['rank_EI', 'rank_EII', 'rank_EIII', 'rank_EIV', 'rank_EV', 'rank_EVI']
    total = len(ranko)

    for col in rank_cols:
        prop1 = (ranko[col] == '1').sum() / total
        prop2 = (ranko[col] == '2').sum() / total
        top[col] = prop1 + prop2 

        prop3 = (ranko[col] == '3').sum() / total
        prop4 = (ranko[col] == '4').sum() / total
        mid[col] = prop3 + prop4 

        prop5 = (ranko[col] == '5').sum() / total
        prop6 = (ranko[col] == '6').sum() / total
        bot[col] = prop5 + prop6 

    movies = ['The Phantom Menace', 'Attack of the Clones', 'Revenge of the Sith', 'A New Hope', 'The Empire Strikes Back',
              'Return of the Jedi']
    
    df = pd.DataFrame({'Movie': movies, 'Top Third': top.values(), 'Middle Third': mid.values(),
                      'Bottom Third': bot.values()})

    scale = alt.Scale(domain=[0,1])

    top_chart_base = alt.Chart(df).mark_bar(size=20, color='green').encode(
         y=alt.Y(
            'Movie:N',
             axis=alt.Axis(tickCount=5, title=''),
             sort=names_l),
        x = alt.X('Top Third:Q', scale=scale,
                  axis=None),
    ).properties(title='Top Third')
    
    top_text = alt.Chart(df).mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        y=alt.Y('Movie:N', sort=names_l),
        x=alt.X('Top Third:Q', scale=scale),
        # we'll use the percentage as the text
        text=alt.Text('Top Third:Q', format='.0%')
    )

    top_chart = (top_chart_base + top_text).properties(title = {'text':'Top Third',
                        'anchor':'middle'}).properties(height=150, width=200)

    middle_chart_base = alt.Chart(df).mark_bar(size=20, color='blue').encode(
         y=alt.Y(
            'Movie:N', title=None, axis=None),
        x = alt.X('Middle Third:Q', title = 'Middle Third', scale = scale,
                  axis=None),
    ).properties(title='Middle Third')
    
    middle_text = alt.Chart(df).mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        y=alt.Y('Movie:N', sort=names_l),
        x=alt.X('Middle Third:Q', scale=scale),
        text=alt.Text('Middle Third:Q', format='.0%')
    )

    middle_chart = (middle_chart_base + middle_text).properties(title = {'text':'Middle Third',
                        'anchor':'middle'}).properties(height=150, width=200)

    bottom_chart_base = alt.Chart(df).mark_bar(size=20, color='red').encode(
         y=alt.Y(
            'Movie:N', title=None, axis=None),
        x = alt.X('Bottom Third:Q', title = 'Bottom Third', scale=scale,
                  axis=None),
    ).properties(title='Bottom Third')
    
    bottom_text = alt.Chart(df).mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        y=alt.Y('Movie:N', sort=names_l),
        x=alt.X('Bottom Third:Q', scale=scale),
        text=alt.Text('Bottom Third:Q', format='.0%')
    )

    bottom_chart = (bottom_chart_base + bottom_text).properties(title = {'text':'Bottom Third',
                        'anchor':'middle'}).properties(height=150, width=200)

    alt.themes.enable('fivethirtyeight')
    
    thirds_chart = (top_chart | middle_chart | bottom_chart).configure_view(strokeWidth=0).properties(
        # add a title
        title={'text':"How People Rate the Star Wars Movies",
               'subtitle': 'How often each film was rated in the top, middle and bottom third (by 471 respondents who have seen all 6 films)'}
    )

    return thirds_chart
```

```python
# Recreate this image using altair here (10 POINTS)

def genFavorabilityVis(inpf):
    # input: inpf, the star wars dataset (as defined at top of file)
    # output: the Altair visualization as described above
    names = ['Luke Skywalker','Han Solo','Princess Leia Organa','Obi Wan Kenobi','Yoda','R2 D2','C-3P0',
             'Anakin Skywalker','Darth Vader','Lando Calrissian','Padme Amidala','Boba Fett','Emperor Palpatine','Jar Jar Binks']

    dropper = inpf.dropna(subset=names)
    total = len(dropper)
    
    Favorable = {}
    Neutral = {}
    Unfavorable = {}
    Unfamiliar = {}

    ratings = ['Favorable', 'Neutral', 'Unfavorable', 'Unfamiliar']
    
    for name in names:
        fav = (((dropper[name] == 'Very favorably').sum() / total) + ((dropper[name] == 'Somewhat favorably').sum() / total))
        neut = (dropper[name] == 'Neither favorably nor unfavorably (neutral)').sum() / total
        unfav = (((dropper[name] == 'Very unfavorably').sum() / total) + ((dropper[name] == 'Somewhat unfavorably').sum() / total))
        unfam = (dropper[name] == 'Unfamiliar (N/A)').sum() / total

        Favorable[name] = fav
        Neutral[name] = neut
        Unfavorable[name] = unfav
        Unfamiliar[name] = unfam

    structure = [Favorable, Neutral, Unfavorable, Unfamiliar]

    df1 = pd.DataFrame(structure, index=ratings)
    df = df1.T
    df= df.reset_index()

    scale = alt.Scale(domain=[0,1])

    fav_chart_base = alt.Chart(df).mark_bar(size=15, color='green').encode(
         y=alt.Y(
            'index:N',
             axis=alt.Axis(tickCount=5, title=''),
             sort=names),
        x = alt.X('Favorable:Q', title='Favorable', scale=scale,
                  axis=None),
    ).properties(title={'text':'Favorable',
                        'anchor':'middle'})
    
    fav_text = alt.Chart(df).mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        y=alt.Y('index:N', sort=names),
        x=alt.X('Favorable:Q', scale=scale),
        # we'll use the percentage as the text
        text=alt.Text('Favorable:Q', format='.0%')
    )

    fav_chart = (fav_chart_base + fav_text).properties(title = {'text':'Favorable',
                        'anchor':'middle'}).properties(height=300, width=175)

    neutral_chart_base = alt.Chart(df).mark_bar(size=15, color='blue').encode(
         y=alt.Y(
            'index:N', title=None, axis=None),
        x = alt.X('Neutral:Q', title = 'Neutral', scale = scale,
                  axis=None),
    ).properties(title='Neutral')
    
    neutral_text = alt.Chart(df).mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        y=alt.Y('index:N', sort=names),
        x=alt.X('Neutral:Q', scale=scale),
        text=alt.Text('Neutral:Q', format='.0%')
    )

    neutral_chart = (neutral_chart_base + neutral_text).properties(title = {'text':'Neutral',
                        'anchor':'middle'}).properties(height=300, width=175)

    unfavorable_chart_base = alt.Chart(df).mark_bar(size=15, color='red').encode(
         y=alt.Y(
            'index:N', title=None, axis=None),
        x = alt.X('Unfavorable:Q', title = 'Unfavorable', scale=scale,
                  axis=None),
    ).properties(title='Unfavorable')
    
    unfavorable_text = alt.Chart(df).mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        y=alt.Y('index:N', sort=names),
        x=alt.X('Unfavorable:Q', scale=scale),
        text=alt.Text('Unfavorable:Q', format='.0%')
    )

    unfavorable_chart = (unfavorable_chart_base + unfavorable_text).properties(title = {'text':'Unfavorable',
                        'anchor':'middle'}).properties(height=300, width=175)

    unfamiliar_chart_base = alt.Chart(df).mark_bar(size=15, color='gray').encode(
         y=alt.Y(
            'index:N', title=None, axis=None),
        x = alt.X('Unfamiliar:Q', title = 'Unfamiliar', scale=scale,
                  axis=None),
    ).properties(title='Unfamiliar')
    
    unfamiliar_text = alt.Chart(df).mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        y=alt.Y('index:N', sort=names),
        x=alt.X('Unfamiliar:Q', scale=scale),
        text=alt.Text('Unfamiliar:Q', format='.0%')
    )

    unfamiliar_chart = (unfamiliar_chart_base + unfamiliar_text).properties(title = {'text':'Unfamiliar',
                        'anchor':'middle'}).properties(height=300, width=175)
    
    alt.themes.enable('fivethirtyeight')
    
    char_chart = (fav_chart | neutral_chart | unfavorable_chart | unfamiliar_chart).configure_view(strokeWidth=0).properties(
        # add a title
        title={'text':"'Star Wars' Character Characters Favorability Ratings",
               'subtitle': 'by 760 respondents'}
    )
    
    return char_chart
    
    
```

### 2.4 Who shot first? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genFirstShotVis(inpf):
    # input: inpf, the star wars dataset
    # output: the Altair visualization as described above
    
    dropper = inpf.dropna(subset=['Which character shot first?'])
    answers = ['Han', 'Greedo', "I don't understand this question"]
    total = len(dropper)

    percents = []
    
    for ans in answers:
        prop = (dropper['Which character shot first?'] == ans).sum() / total
        percents.append(prop)

    df = pd.DataFrame({'Answer': answers, 'Percentage': percents})

    base = alt.Chart(df).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Answer:N',
             axis=alt.Axis(tickCount=5, title=''),
            sort=answers,
             # we give the sorting order to avoid alphabetical order
        )
    )

    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q', format='.0%')
    )

    alt.themes.enable('fivethirtyeight')
    
    who_shot = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=130
    ).properties(
        # add a title
        title={'text':"Who Shot First?",
               'subtitle': 'According to 750 respondents'}
    )
    
    return who_shot
    
```

```python
def preqGenerationData(inpf,eps, names=namesDict, names_l=namesList):
    # returns the dataframe I'll use for each visualization 
    seenDF = getSeenAtLeastOneDF(inpf, eps)
    Age_Groups = seenDF['Age'].dropna().unique()
    seen_cols = ['seen_EI', 'seen_EII', 'seen_EIII', 'seen_EIV', 'seen_EV', 'seen_EVI']
    storage = {}

    for movie in seen_cols:
        storage[movie] = []
        for age in Age_Groups:
            subset = seenDF[seenDF['Age'] == age]

            prop = subset[movie].notna().sum() / len(subset)
            storage[movie].append(prop)

    
    df = pd.DataFrame(storage, index=Age_Groups)
    df = df.rename(columns={'seen_EI': 'The Phantom Menace', 'seen_EII': 'Attack of the Clones', 'seen_EIII': 'Revenge of the Sith',
                            'seen_EIV': 'A New Hope', 'seen_EV': 'The Empire Strikes Back', 'seen_EVI': 'Return of the Jedi'})

    return df

    
```
