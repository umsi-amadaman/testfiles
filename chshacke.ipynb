### 1.2 A Pie Chart Version (15 points)


!["Drawing"](/voc/data/have_seen_resized.png)

Your colleague is arguing that the chart above could be made more *effective* by using one or more pie charts. In what situations are they right? In what situations would they be wrong? If it helps makes your case, feel free to create a picture (hand drawn is fine) and load it into the notebook.

Pie charts are effective when we want to show how one piece relates to the whole. They’re less effective (than bars) when we want to compare one piece to another, because our brains have a harder time judging angles and areas than comparing lengths along a shared baseline.

If our intent was to create a visual where the viewer could focus on ONE movie and note the percentage of ONE movie compared to the whole - e.g. looking strictly at phantom menace to see that 80% of viewers have seen it, and that's the end of the comparisons we want them to draw - then a pie chart would be more effective. It taps into our natural tendency to see the “whole picture,” something Gestalt psychology calls the principle of Prägnanz, or simplicity. 

In the circumstance that we want to invite the viewer to be able to compare the viewing rate of one movie versus another movie - i.e. which movie has stronger viewership, 'The Empire Strikes Back' or 'Return of the Jedi'? - then a bar chart will be more effective. Per Weber’s Law, people can detect smaller differences when comparing lengths than when comparing angles. That means a small difference like 88% versus 91% is hard to spot in a pie chart but obvious in a bar chart. Bar charts also make use of Gestalt principles like proximity, similarity, and good continuation. Because the bars share a common starting point and the same shape, our eyes can easily move across them and pick up on differences.

If the intent is to highlight part-to-whole relationships for a single category, the pie chart would be more effective. If our goal is to help the viewer make comparisons across categories or see changes in popularity between films, a bar chart is the better choice.

```python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    # create the various mapping and order lists/dictionaries
    # return: list of episodes (in order), episode->name map, list of episode names (in order)
    episodes = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    names = {
        'EI' : 'The Phantom Menace', 'EII' : 'Attack of the Clones', 'EIII' : 'Revenge of the Sith', 
        'EIV': 'A New Hope', 'EV': 'The Empire Strikes Back', 'EVI' : 'The Return of the Jedi'
    }

    # we're also going to use this order to sort, so names_l will now have our sort order
    return episodes, names, [names[ep] for ep in episodes]

episodes, namesDict, namesList = genEpisodeNamesDF()
```

```python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
```

```python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN
    salo = indf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    return(salo)
```

```python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
```

```python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total = len(inpf)
    percs = []

    # loop over each column and calculate the number of people who have seen the movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count them
    # and divide by the percent
    for seen_ep in ['seen_' + ep for ep in episodes]:
        perc = len(inpf[~ pd.isna(inpf[seen_ep])]) / total
        percs.append(perc)

    # at this point percs is holding our percentages

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes],percs))
    seen_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    return(seen_per_df)
```

```python
seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
```

```python
# let's see what's inside
seenPerMovieDF
```

```python
def genPercentVis(indf, names_l=namesList):
    # input: indf, the dataframe as seen_per_df
    # input: names_l - a list of all the movies in series order (default namesList)
    # output: simple altair bar chart
    
    # ok, time to make the chart... let's make a bar chart (use mark_bar)
    bars = alt.Chart(indf).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        )
    )

    # at this point we don't really have a great plot (it's missing the annotations, titles, etc.)
    return(bars)

seenPerMovieVis = genPercentVis(seenPerMovieDF)
```

```python
def augmentPercentVis(base):
    # input: base (the base vis, i.e., bars as above)
    # we're going to overlay the text with the percentages, so let's make another visualization
    # that's just text labels


    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    seen_movies = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=180
    ).properties(
        # add a title
        title="Which 'Star Wars' Movies Have you Seen?"
    )

    return(seen_movies)

# note that we are NOT formatting this in the Five Thirty Eight Style yet... we'll leave that to you to figure out
```

### 2.2 How people rate the 'Star Wars' movie? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genRateVis(inpf,eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above
    names = {**names, 'EVI': 'Return of the Jedi'}
    names_l = [names[e] for e in eps]

    df_all6 = inpf.dropna(subset=[f'seen_{e}' for e in eps], how='any').copy()

    rank_cols = {e: f'rank_{e}' for e in eps}
    rows = []
    for e in eps:
        r = pd.to_numeric(df_all6[rank_cols[e]], errors='coerce')
        film = names[e]
        rows += [
            (film, 'Top third',    r.isin([1,2]).mean()),
            (film, 'Middle third', r.isin([3,4]).mean()),
            (film, 'Bottom third', r.isin([5,6]).mean())
        ]

    ratings = pd.DataFrame(rows, columns=['Film','Category','Percent'])
    cat_color = {'Top third':'#60BD68','Middle third':'#4C78A8','Bottom third':'#E45756'}

    def col(cat):
        show_y = (cat == 'Top third')
        base = alt.Chart(ratings.query("Category == @cat")).encode(
            y=alt.Y(
                'Film:N',
                sort=names_l,
                title=None,
                axis=alt.Axis(labels=show_y, ticks=False, labelFontSize=16, labelPadding=6, labelLimit=180),
                scale=alt.Scale(paddingInner=0.001, paddingOuter=0.05)
            ),
            x=alt.X('Percent:Q', title=None, axis=None)
        )
        bars = base.mark_bar(size=26, color=cat_color[cat])
        labels = base.mark_text(align='left', baseline='middle', dx=6, color='#333'
                 ).encode(text=alt.Text('Percent:Q', format='.0%'))
        return (bars + labels).properties(title=cat, width=100, height=200)

    chart = col('Top third') | col('Middle third') | col('Bottom third')

    return chart.properties(
        background='#f7f7f7',
        title=alt.TitleParams(
            text="How People Rate the 'Star Wars' Movies",
            subtitle=[
                "How often each film was rated in the top, middle and bottom third",
                "(by 471 respondents who have seen all six films)"
            ],
            anchor='start'
        )
    ).configure_view(strokeWidth=0
    ).configure_title(fontSize=22, fontWeight='bold', subtitleFontSize=20
    ).configure_axis(domain=False, grid=False, labelColor='#6e6e6e', labelFontSize=16)
```

```python
# let's check our solution
genRateVis(sw,episodes)
```

```python
# Recreate this image using altair here (10 POINTS)

def genFavorabilityVis(inpf):
    # input: inpf, the star wars dataset (as defined at top of file)
    # output: the Altair visualization as described above
    
    # YOUR CODE HERE
    chars = [
        'Luke Skywalker','Han Solo','Princess Leia Organa','Obi Wan Kenobi','Yoda',
        'R2 D2','C-3P0','Anakin Skywalker','Darth Vader','Lando Calrissian',
        'Padme Amidala','Boba Fett','Emperor Palpatine','Jar Jar Binks'
    ]

    keep = (inpf['seen_any_movie'] == 'Yes')
    df = inpf.loc[keep, chars].copy()
    n = len(df)

    def bucket(v):
        if pd.isna(v): return 'Unfamiliar'
        s = str(v).strip().lower()
        if 'very favor' in s or 'somewhat favor' in s: return 'Favorable'
        if 'neither favorably nor unfavorably' in s or s == 'neither': return 'Neutral'
        if 'very unfavor' in s or 'somewhat unfavor' in s: return 'Unfavorable'
        if 'unfamiliar' in s or 'not familiar' in s: return 'Unfamiliar'
        return 'Unfamiliar'

    long = df.melt(var_name='Character', value_name='resp').dropna()
    long['Category'] = long['resp'].apply(bucket)

    pct = (
        long.groupby(['Character','Category'])
            .size()
            .groupby(level=0)
            .apply(lambda x: x/x.sum())
            .reset_index(name='Percent')
    )

    order = [
        'Luke Skywalker','Han Solo','Princess Leia Organa','Obi Wan Kenobi','Yoda',
        'R2 D2','C-3P0','Anakin Skywalker','Darth Vader','Lando Calrissian',
        'Padme Amidala','Boba Fett','Emperor Palpatine','Jar Jar Binks'
    ]

    colors = {'Favorable':'#60BD68','Neutral':'#4C78A8','Unfavorable':'#E45756','Unfamiliar':'#bab0ac'}
    text_color = '#333'
    bar_size = 22
    col_width = 100
    col_height = 360
    label_col_width = 10

    def bar_panel(cat, data):
        base = alt.Chart(data).encode(
            y=alt.Y('Character:N',
                    sort=order, title=None,
                    axis=alt.Axis(labels=False, ticks=False),
                    scale=alt.Scale(paddingInner=0.05, paddingOuter=0.05)),
            x=alt.X('Percent:Q', title=None, axis=None)
        )
        bars = base.mark_bar(size=bar_size, color=colors[cat], cornerRadius=2)
        labels = base.mark_text(
            align='left', baseline='middle', dx=6, color=text_color
        ).encode(
            text=alt.Text('Percent:Q', format='.0%')   # <- simple percent formatting
        )
        return (bars + labels).properties(
            title=alt.TitleParams(text=cat, anchor='start',
                                  fontSize=12, fontWeight='bold',
                                  color=text_color, dy=6),
            width=col_width, height=col_height
        )

    def first_column():
        data = pct.query("Category == 'Favorable'")
        labels_only = alt.Chart(data).encode(
            y=alt.Y('Character:N', sort=order, title=None,
                    axis=alt.Axis(labels=False, ticks=False),
                    scale=alt.Scale(paddingInner=0.05, paddingOuter=0.05))
        ).mark_text(
            align='right', baseline='middle', dx=-6, color=text_color, fontSize=16
        ).encode(text='Character:N').properties(width=label_col_width, height=col_height)

        bars = bar_panel('Favorable', data)
        return alt.hconcat(labels_only, bars, spacing=0)

    def column(cat):
        return bar_panel(cat, pct.query("Category == @cat"))

    c1 = first_column()
    c2 = column('Neutral')
    c3 = column('Unfavorable')
    c4 = column('Unfamiliar')

    title = alt.TitleParams(
        text="‘Star Wars’ Character Favorability Ratings",
        subtitle=[f"By {n} respondents"],
        anchor='start',
        color=text_color
    )

    chart = alt.hconcat(c1, c2, c3, c4, spacing=12).properties(
        background='#f7f7f7', title=title,
        padding={"left": 8, "right": 8, "top": 8, "bottom": 8}
    ).configure_view(
        strokeWidth=0
    ).configure_title(
        fontSize=24, fontWeight='bold',
        subtitleFontSize=20, subtitleFontWeight='normal',
        color=text_color, subtitleColor=text_color
    ).configure_axis(
        domain=False, grid=False, ticks=False, labelColor=text_color
    )

    return chart
```

This is both expressive (the visual form is the concept) and effective: length/position are the most accurate channels for quantitative judgment, so viewers can compare spreads quickly and read exact values from the end labels.
