```python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    # create the various mapping and order lists/dictionaries
    # return: list of episodes (in order), episode->name map, list of episode names (in order)
    episodes = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    names = {
        'EI' : 'The Phantom Menace', 'EII' : 'Attack of the Clones', 'EIII' : 'Revenge of the Sith', 
        'EIV': 'A New Hope', 'EV': 'The Empire Strikes Back', 'EVI' : 'The Return of the Jedi'
    }

    # we're also going to use this order to sort, so names_l will now have our sort order
    return episodes, names, [names[ep] for ep in episodes]

episodes, namesDict, namesList = genEpisodeNamesDF()
```

```python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
```

```python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN
    salo = indf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    return(salo)
```

```python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
```

```python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total = len(inpf)
    percs = []

    # loop over each column and calculate the number of people who have seen the movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count them
    # and divide by the percent
    for seen_ep in ['seen_' + ep for ep in episodes]:
        perc = len(inpf[~ pd.isna(inpf[seen_ep])]) / total
        percs.append(perc)

    # at this point percs is holding our percentages

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes],percs))
    seen_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    return(seen_per_df)
```

```python
seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
```

```python
# let's see what's inside
seenPerMovieDF
```

```python
def genPercentVis(indf, names_l=namesList):
    # input: indf, the dataframe as seen_per_df
    # input: names_l - a list of all the movies in series order (default namesList)
    # output: simple altair bar chart
    
    # ok, time to make the chart... let's make a bar chart (use mark_bar)
    bars = alt.Chart(indf).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        )
    )

    # at this point we don't really have a great plot (it's missing the annotations, titles, etc.)
    return(bars)

seenPerMovieVis = genPercentVis(seenPerMovieDF)
```

```python
def augmentPercentVis(base):
    # input: base (the base vis, i.e., bars as above)
    # we're going to overlay the text with the percentages, so let's make another visualization
    # that's just text labels


    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    seen_movies = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=180
    ).properties(
        # add a title
        title="Which 'Star Wars' Movies Have you Seen?"
    )

    return(seen_movies)

# note that we are NOT formatting this in the Five Thirty Eight Style yet... we'll leave that to you to figure out
```

```python
# Recreate this image using Altair
# match the "538 style" as best you can (hint: look at the altair lab at the start of the semester)

def genBestVis(inpf,eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above
    # YOUR CODE HERE
    
    # Keep rows with all 6 movies seen
    seen_all = inpf.dropna(subset=['seen_' + ep for ep in eps]).copy()

    total = len(seen_all)

    # Create a new column for favorites
    seen_all['Favorite'] = (seen_all[['rank_' + ep for ep in eps]] == '1').idxmax(axis=1).str.replace('rank_', '')

    # Group by favorite movie and output percentages
    results = seen_all.groupby('Favorite').apply(lambda x: (len(x) / len(seen_all))).reset_index(name="Percentage")
    results.replace(names, inplace=True)
    
    # Creating the graph
    bars = alt.Chart(results).mark_bar(size=20).encode(
        # Encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # Encode y using the name, use the name to label the axis, sort using the names_l
            'Favorite:N',
             axis=alt.Axis(tickCount=5, title='', labelColor='#9B9B9B'),
             sort=names_l
        )
    )

    text = bars.mark_text(
        align='left',
        baseline='middle',
        dx=3
    ).encode(
        text=alt.Text('Percentage:Q',format='.0%')
    )

    best_movies = (text + bars).configure_mark(
        color='#008fd5'
    ).configure_view(
        strokeWidth=0
    ).configure_scale(
        bandPaddingInner=0.2
    ).properties(
        width=500,
        height=180
    ).properties(
        title=alt.TitleParams(
            text="What's the Best 'Star Wars' Movie?",
            subtitle=f"Of {total} respondents who have seen all six films",
            align='left',
            anchor='start',
            fontSize=24,
            subtitleFontSize=15,
            offset=25,
        ),
        background='#F0F0F0'
    )

    return best_movies
    
```

### 2.2 How people rate the 'Star Wars' movie? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genRateVis(inpf,eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above
    
    # YOUR CODE HERE
    seen_all = inpf.dropna(subset=['seen_' + ep for ep in eps]).copy()
    total_length = len(seen_all)

    # Define dictionaries for each third
    top_third = {}
    middle_third = {}
    bottom_third = {}

    # Assign each dictionary a third percentage
    for col in ['rank_' + ep for ep in eps]:
        top_third[col[5:]] = len(seen_all.loc[seen_all[col].isin(['1', '2'])]) / total_length
        middle_third[col[5:]] = len(seen_all.loc[seen_all[col].isin(['3', '4'])]) / total_length
        bottom_third[col[5:]] = len(seen_all.loc[seen_all[col].isin(['5', '6'])]) / total_length

    # Convert Dictionarys to DataFrames
    top_third = pd.DataFrame({'Episode': list(top_third.keys()), 'Percent': list(top_third.values())}).replace(names)
    top_third = top_third.assign(Third='Top third')
    
    middle_third = pd.DataFrame({'Episode': list(middle_third.keys()), 'Percent': list(middle_third.values())}).replace(names)
    middle_third = middle_third.assign(Third='Middle third')
    
    bottom_third = pd.DataFrame({'Episode': list(bottom_third.keys()), 'Percent': list(bottom_third.values())}).replace(names)
    bottom_third = bottom_third.assign(Third='Bottom third')
    

    results = pd.concat([top_third, middle_third, bottom_third], ignore_index=True)

    base = alt.Chart(results).encode(
        # Encode x as the percent, and hide the axis
        x=alt.X(
            'Percent',
            axis=None),
        y=alt.Y(
            # Encode y using the epsiode, use the epsiode to label the axis, sort using the names_l
            'Episode:N',
            axis=alt.Axis(tickCount=5, title='', labelColor='#9B9B9B'),
            sort=names_l
        )
    )

    bars = base.mark_bar().encode(
        color=alt.Color(
            'Third:N', 
            scale=alt.Scale(
                domain=['Top third', 'Middle third', 'Bottom third'],
                range=['#4DA74A', '#008FD5', '#FF2600']
            ),
            legend=None
        )
    )

    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3
    ).encode(
        # Use percentage as text
        text=alt.Text('Percent:Q',format='.0%'),
        color=alt.Color(
            'Third:N',
            scale=alt.Scale(
                domain=['Top third', 'Middle third', 'Bottom third'],
                range=['#4DA74A', '#008FD5', '#FF2600']
            )
        )
    )

    # finally, we're going to combine the bars and the text and do some styling
    third = (bars + text).properties(
        width=80
    ).facet(
        column=alt.Column('Third:N', sort=['Top third', 'Middle third', 'Bottom third'], header=alt.Header(title=None)),
    ).configure_view(
        strokeWidth=0
    ).configure_scale(
        bandPaddingInner=0.2
    ).properties(
        # Add title, multiple subtitles, alignment / anchor, and style the font
        title=alt.TitleParams(
            text="How People Rate the 'Star Wars' Movies",
            subtitle=["How often each film was rated in the top, middle and bottom third", f"(by {total_length} respondents who have seen all six films)"],
            align='left',
            anchor='start',
            fontSize=24,
            subtitleFontSize=15,
            offset=25
        ),
        background='#F0F0F0'
    )
    
    return third
```

```python
# Recreate this image using altair here (10 POINTS)

def genFavorabilityVis(inpf):
    # input: inpf, the star wars dataset (as defined at top of file)
    # output: the Altair visualization as described above
    
    # YOUR CODE HERE
    characters = [
        'Han Solo', 'Luke Skywalker', 'Princess Leia Organa', 'Anakin Skywalker',
        'Obi Wan Kenobi', 'Emperor Palpatine', 'Darth Vader', 'Lando Calrissian',
        'Boba Fett', 'C-3P0', 'R2 D2', 'Jar Jar Binks', 'Padme Amidala', 'Yoda'
    ]

    # Filter out entries who do not rate any of the characters
    data = inpf.copy().dropna(subset=characters, how='all')
    total_length = len(data)
    
    favorable = {}
    neutral = {}
    unfavorable = {}
    unfamiliar = {}

    # Assign each dictionary a third percentage
    for col in characters:
        favorable[col] = len(data.loc[data[col].isin(['Very favorably', 'Somewhat favorably'])]) / total_length
        neutral[col] = len(data.loc[data[col] == 'Neither favorably nor unfavorably (neutral)']) / total_length
        unfavorable[col] = len(data.loc[data[col].isin(['Very unfavorably', 'Somewhat unfavorably'])]) / total_length
        unfamiliar[col] = len(data.loc[(data[col] == 'Unfamiliar (N/A)') | (data[col].isna())]) / total_length

    # Convert Dictionarys to DataFrames
    favorable = pd.DataFrame({'Character': list(favorable.keys()), 'Percent': list(favorable.values())})
    favorable = favorable.assign(Rating='Favorable')
    
    neutral = pd.DataFrame({'Character': list(neutral.keys()), 'Percent': list(neutral.values())})
    neutral = neutral.assign(Rating='Neutral')

    unfavorable = pd.DataFrame({'Character': list(unfavorable.keys()), 'Percent': list(unfavorable.values())})
    unfavorable = unfavorable.assign(Rating='Unfavorable')

    unfamiliar = pd.DataFrame({'Character': list(unfamiliar.keys()), 'Percent': list(unfamiliar.values())})
    unfamiliar = unfamiliar.assign(Rating='Unfamiliar')

    # Define a sort order
    sort_order = favorable.sort_values(by='Percent', ascending=False)['Character']

    # Join resulting dfs
    results = pd.concat([favorable, neutral, unfavorable, unfamiliar], ignore_index=True)

    base = alt.Chart(results).encode(
        # Encode x as the percent, and hide the axis
        x=alt.X(
            'Percent',
            axis=None),
        y=alt.Y(
            # Encode y using the character name, use the character name to label the axis, sort using predefined order
            'Character',
            axis=alt.Axis(tickCount=14, title='', labelColor='#9B9B9B'),
            sort=sort_order
        )
    )

    bars = base.mark_bar().encode(
        color=alt.Color(
            'Rating:N', 
            scale=alt.Scale(
                domain=['Favorable', 'Neutral', 'Unfavorable', 'Unfamiliar'],
                range=['#4DA74A', '#008FD5', '#FF2600', 'grey']
            ),
            legend=None
        )
    )

    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3
    ).encode(
        # Use the percent as the text, color by domain
        text=alt.Text('Percent:Q',format='.0%'),
        color=alt.Color(
            'Rating:N',
            scale=alt.Scale(
                domain=['Favorable', 'Neutral', 'Unfavorable', 'Unfamiliar'],
                range=['#4DA74A', '#008FD5', '#FF2600', 'grey']
            )
        )
    )

    # finally, we're going to combine the bars and the text and do some styling
    final = (bars + text).properties(
        width=80
    ).facet(
        column=alt.Column('Rating:N', sort=['Favorable', 'Neutral', 'Unfavorable', 'Unfamiliar'], header=alt.Header(title=None)),
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # add a title, subtitle
        title=alt.TitleParams(
            text="'Star Wars' Character Favorability Ratings",
            subtitle=f"By {total_length} respondents",
            align='left',
            anchor='start',
            fontSize=24,
            subtitleFontSize=15,
            offset=25
        ),
        background='#F0F0F0'
    )
    
    return final
    
```

### 2.4 Who shot first? Recreate the above image using altair (10 Points)

```python
# YOUR CODE HERE
# Favorite Movie by Age Group by those who have seen all the movies
# Visual #1
def generateVisualOne(inpf, eps, names=namesDict):
    # input: inpf, the star wars dataset
    # input: eps, episode list
    # input: names, {episode abbr: full name} dictionary
    # output: the Altair visualization of a bar chart
    
    data = inpf.copy()
    data.dropna(subset=['Age'] + ['seen_' + ep for ep in eps], inplace=True)

    total_length = len(data)

    # Filter out only the favorites
    data['Favorite'] = (data[['rank_' + ep for ep in eps]] == '1').idxmax(axis=1).str.replace('rank_', '')

    # Group data by Age and Favorite Episode
    results = data.groupby(['Age', 'Favorite']).size().reset_index(name='Size')
    results.replace(names, inplace=True)

    # Find the total number of responses per age group
    totals = data.groupby('Age').size().reset_index(name='Total')

    # Merge data together and form percentage column
    final_results = pd.merge(results, totals, on='Age')
    final_results['Percent'] = final_results['Size'] / final_results['Total']

    # Define sort order by percentage 
    sort_order = final_results.sort_values(by=['Percent'], ascending=False)['Favorite'].drop_duplicates().reset_index(drop=True)

    # Create a domain of age groups
    ages = final_results['Age'].drop_duplicates().reset_index(drop=True)

    base = alt.Chart(final_results).encode(
        # Encode x as the percent, and hide the axis
        x=alt.X(
            'Percent:Q',
            axis=None),
        y=alt.Y(
            # Encode y using episodes, sort by percentage
            'Favorite:N',
            axis=alt.Axis(tickCount=len(eps), title='', labelColor='#9B9B9B'),
            sort=sort_order
        )
    )

    # Create bar charts
    bars = base.mark_bar().encode(
        color=alt.Color(
            'Age:N', 
            scale=alt.Scale(
                domain=ages,
                scheme='rainbow'
            ),
            legend=None
        )
    )

    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3
    ).encode(
        text=alt.Text('Percent:Q',format='.0%'),
        color=alt.Color(
            'Age:N',
            scale=alt.Scale(
                domain=ages,
                scheme='rainbow'
            )
        )
    )

    # Combine bars with text and style the chart
    final = (bars + text).properties(
        width=80
    ).facet(
        column=alt.Column('Age:N', sort=ages, header=alt.Header(title=None)),
    ).configure_view(
        strokeWidth=0
    ).configure_scale(
        bandPaddingInner=0.2
    ).properties(
        title=alt.TitleParams(
            text="Favorite Episode by Age Group",
            subtitle=f"By {total_length} respondents",
            align='left',
            anchor='start',
            fontSize=24,
            subtitleFontSize=15,
            offset=25
        ),
        background='#F0F0F0'
    )
    
    return final
```

```python
# YOUR CODE HERE

# Visual #2
def generateVisualTwo(inpf, eps, names=namesDict):
    # input: inpf, the star wars dataset
    # input: eps, episode list
    # input: names, {episode abbr: full name} dictionary
    # output: the Altair visualization of a pie chart
    
    data = inpf.copy()
    data.dropna(subset=['Age'] + ['seen_' + ep for ep in eps], inplace=True)

    total_length = len(data)

    # Filter out only the favorites
    data['Favorite'] = (data[['rank_' + ep for ep in eps]] == '1').idxmax(axis=1).str.replace('rank_', '')

    # Group data by Age and Favorite Episode
    results = data.groupby(['Age', 'Favorite']).size().reset_index(name='Size')
    results.replace(names, inplace=True)

    # Find the total number of responses per age group
    totals = data.groupby('Age').size().reset_index(name='Total')

    # Merge data together and form percentage column
    final_results = pd.merge(results, totals, on='Age')
    final_results['Percent'] = final_results['Size'] / final_results['Total']

    base = alt.Chart(final_results).encode(
        # Encode theta as the percent, color as the episode, and order by percent
        theta=alt.Theta('Percent:Q'),
        color=alt.Color(
            'Favorite:N', 
            scale=alt.Scale(scheme='lightgreyteal')
        ),
        order=alt.Order('Percent:Q')
    )

    # Create pie charts
    pie_charts = base.mark_arc(outerRadius=60, innerRadius=0)

    # Join pie charts into a single visual
    final = pie_charts.properties(
        width=100,
        height=100
    ).facet(
        column=alt.Column('Age:N', header=alt.Header(title=None)),
        columns=6
    ).configure_view(
        stroke=None
    ).properties(
        title=alt.TitleParams(
            text="Favorite Episode by Age Group",
            subtitle=f"By {total_length} respondents",
            align='left',
            anchor='start',
            fontSize=24,
            subtitleFontSize=15,
            offset=25
        ),
        background='#F0F0F0'
    )
    
    return final
```
