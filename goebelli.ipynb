```python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    # create the various mapping and order lists/dictionaries
    # return: list of episodes (in order), episode->name map, list of episode names (in order)
    episodes = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    names = {
        'EI' : 'The Phantom Menace', 'EII' : 'Attack of the Clones', 'EIII' : 'Revenge of the Sith', 
        'EIV': 'A New Hope', 'EV': 'The Empire Strikes Back', 'EVI' : 'The Return of the Jedi'
    }

    # we're also going to use this order to sort, so names_l will now have our sort order
    return episodes, names, [names[ep] for ep in episodes]

episodes, namesDict, namesList = genEpisodeNamesDF()
```

```python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
```

```python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN
    salo = indf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    return(salo)
```

```python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
```

```python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total = len(inpf)
    percs = []

    # loop over each column and calculate the number of people who have seen the movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count them
    # and divide by the percent
    for seen_ep in ['seen_' + ep for ep in episodes]:
        perc = len(inpf[~ pd.isna(inpf[seen_ep])]) / total
        percs.append(perc)

    # at this point percs is holding our percentages

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes],percs))
    seen_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    return(seen_per_df)
```

```python
seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
```

```python
# let's see what's inside
seenPerMovieDF
```

```python
def genPercentVis(indf, names_l=namesList):
    # input: indf, the dataframe as seen_per_df
    # input: names_l - a list of all the movies in series order (default namesList)
    # output: simple altair bar chart
    
    # ok, time to make the chart... let's make a bar chart (use mark_bar)
    bars = alt.Chart(indf).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        )
    )

    # at this point we don't really have a great plot (it's missing the annotations, titles, etc.)
    return(bars)

seenPerMovieVis = genPercentVis(seenPerMovieDF)
```

```python
# display it
seenPerMovieVis
```

```python
def augmentPercentVis(base):
    # input: base (the base vis, i.e., bars as above)
    # we're going to overlay the text with the percentages, so let's make another visualization
    # that's just text labels


    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    seen_movies = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=180
    ).properties(
        # add a title
        title="Which 'Star Wars' Movies Have you Seen?"
    )

    return(seen_movies)

# note that we are NOT formatting this in the Five Thirty Eight Style yet... we'll leave that to you to figure out
```

### 2.2 How people rate the 'Star Wars' movie? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genRateVis(inpf,eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above
    
    # YOUR CODE HERE
    seenAllSix = inpf.copy().dropna(subset=['seen_' + ep for ep in eps], how='any')
    rank_cols = ['rank_' + ep for ep in eps]
    seenAllSix[rank_cols] = seenAllSix[rank_cols].apply(pd.to_numeric)
    complete_ranks = seenAllSix.dropna(subset=rank_cols).copy()

    def rank_bucket(rank):
        if rank in [1, 2]:
            return 'Top third'
        elif rank in [3, 4]:
            return 'Middle third'
        elif rank in [5, 6]:
            return 'Bottom third'

    df = complete_ranks.melt(id_vars=[], value_vars=rank_cols, var_name='movie', value_name='rank')
    df['movie'] = df['movie'].str.replace('rank_', '')
    df['bucket'] = df['rank'].apply(rank_bucket)
    df['movie_name'] = df['movie'].map(names)
    
    bucket_counts = df.groupby(['bucket', 'movie_name']).size().reset_index(name='count')
    bucket_counts['percentage'] = bucket_counts['count'] / len(complete_ranks)

    bucket_order = ['Top third', 'Middle third', 'Bottom third']

    color_scale = alt.Scale(
        domain=bucket_order,
        range=['#3bb013', '#008fd5', '#d62728']
    )
    
    base = alt.Chart(bucket_counts)
    bars = base.mark_bar().encode(
        x=alt.X('percentage:Q', axis=alt.Axis(format='', grid=False, title='', labels=False, ticks=False, domain=False)),
        y=alt.Y('movie_name:N', sort=names_l, axis=alt.Axis(title='')),
        color=alt.Color('bucket:N', legend=None, scale=color_scale)
    )
    
    text = base.mark_text(
        align='left', dx=3, fontWeight='bold').encode(
        x='percentage:Q',
        y=alt.Y('movie_name:N', sort=bucket_order),
        text=alt.Text('percentage:Q', format='.0%'),
        color=alt.Color('bucket:N', legend=None, scale=color_scale)
    )
    
    layered = alt.layer(bars, text).properties(width=80)
    faceted = layered.facet(
        column=alt.Column('bucket:N', header=alt.Header(title=""), sort=bucket_order)
    ).properties(
        title={
            "text": "How People Rate the 'Star Wars' Movies by Bucket",
            "subtitle": "How often each film was rated in top, middle and bottom third (by "  + str(len(complete_ranks)) + " respondents who have seen all six films)",
            "anchor": "start",  "fontSize": 20},
        background='#f5f5f5').configure_axis(
        labelColor='#888888', tickColor='#cccccc').configure_view(stroke=None)
    
    return faceted
    
```

### 2.3 Star Wars' Characters Favorability Ratings. Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genFavorabilityVis(inpf):
    # input: inpf, the star wars dataset (as defined at top of file)
    # output: the Altair visualization as described above
    
    # YOUR CODE HERE
    df = getSeenAtLeastOneDF(inpf, episodes)
    total = len(df)

    char_list = ['Luke Skywalker', 'Han Solo', 'Princess Leia Organa','Obi Wan Kenobi','Yoda','R2 D2','C-3P0',
                 'Anakin Skywalker','Darth Vader','Lando Calrissian','Padme Amidala','Boba Fett','Emperor Palpatine','Jar Jar Binks']

    data = []
    for character in char_list:
        char_series = df[character]
        
        pct_fav = char_series.isin(['Very favorably', 'Somewhat favorably']).sum() / total
        pct_neu = char_series.isin(['Neither favorably nor unfavorably (neutral)']).sum() / total
        pct_unfav = char_series.isin(['Very unfavorably', 'Somewhat unfavorably']).sum() / total
        pct_unfam = char_series.isin(['Unfamiliar (N/A)']).sum() / total
        
        data.append({'Name': character, 'Bucket': 'Favorably', 'Percentage': pct_fav})
        data.append({'Name': character, 'Bucket': 'Neutral', 'Percentage': pct_neu})
        data.append({'Name': character, 'Bucket': 'Unfavorably', 'Percentage': pct_unfav})
        data.append({'Name': character, 'Bucket': 'Unfamiliar', 'Percentage': pct_unfam})
    
    result_df = pd.DataFrame(data)

    bucket_order = ['Favorably', 'Neutral', 'Unfavorably', 'Unfamiliar']
    color_scale = alt.Scale( domain=bucket_order, range=['#3bb013', '#008fd5', '#d62728', '#aaaaaa'])
    
    base = alt.Chart(result_df)
    bars = base.mark_bar().encode(
        x=alt.X('Percentage:Q', axis=alt.Axis(format='', grid=False, title='', labels=False, ticks=False, domain=False)),
        y=alt.Y('Name:N', axis=alt.Axis(title=''), sort=char_list),
        color=alt.Color('Bucket:N', legend=None, scale=color_scale)
    )
 
    text = base.mark_text(
        align='left', dx=3, fontWeight='bold').encode(
        x='Percentage:Q',
        y=alt.Y('Name:N', sort=bucket_order),
        text=alt.Text('Percentage:Q', format='.0%'),
        color=alt.Color('Bucket:N', legend=None, scale=color_scale)
    )
    
    layered = alt.layer(bars, text).properties(width=80)
    faceted = layered.facet(
        column=alt.Column('Bucket:N', header=alt.Header(title=""), sort=bucket_order)
    ).properties(
        title={
            "text": "Star Wars' Characters Favorability Ratings",
            "subtitle": "By "  + str(total) + " respondents",
            "anchor": "start", "fontSize": 20},
        background='#f5f5f5').configure_axis(labelColor='#888888', tickColor='#cccccc').configure_view(stroke=None)
    
    return faceted
  
```

```python
# My data cleaning

rank_cols = ['rank_EI', 'rank_EII', 'rank_EIII', 'rank_EIV', 'rank_EV', 'rank_EVI']

episode_labels = {f'rank_{ep}': namesDict[ep] for ep in episodes}
df_long = sw.melt(id_vars='Age', value_vars=rank_cols, var_name='Episode', value_name='Ranking')

df_long['Episode'] = df_long['Episode'].map(episode_labels)

df_long = df_long.dropna()
# display(df_long.head())
```

```python
# YOUR CODE HERE
def WorseVis(df):
    
    bar_chart = alt.Chart(df).mark_bar().encode(
        x=alt.X('Episode:N', sort=list(episode_labels.values())),
        y='mean(Ranking):Q',
        color='Age:N',
        tooltip=['Age:N', 'mean(Ranking):Q']
    ).properties(
        width=600, height=400,
        title='Average Movie Ranking by Age Group (Lower is Better)'
    ).configure_axis(labelAngle=-45)

    return bar_chart
```

```python
WorseVis(df_long)
```

```python
# YOUR CODE HERE
def BetterVis(df):
    
    heatmap = alt.Chart(df).mark_rect().encode(
        x=alt.X('Episode:N', sort=list(episode_labels.values())),
        y=alt.Y('Age:N'),
        color=alt.Color('mean(Ranking):Q',
            scale=alt.Scale(scheme='blues'),
            legend=alt.Legend(
                title='Average Ranking',
                direction='vertical',
                orient='right',
                gradientLength=100,
                labelExpr="datum.label"
            )
        ),
        tooltip=['Age:N', 'Episode:N', 'mean(Ranking):Q']
    ).properties(
        width=600,
        height=300,
        title='Heatmap of Average Rankings by Age Group and Episode'
    ).configure_axis(labelAngle=-45)

    return heatmap
```

```python
BetterVis(df_long)
```

#### Justification
*Provide your justification here* to argue why your "better" version is actually better. Make sure to defend this based on expressiveness and effectiveness as well as perceptual arguments.

The heatmap is a better visualization because it more effectively represents the nature of the data (average movie rankings across age groups) using color intensity instead of bar height. Rankings are ordinal, not quantitative, so using a single hue color scale (ex. dark blue for best rankings) better matches the data type without implying misleading proportional differences. The grid layout also allows for rapid comparisons both across age groups and across episodes, making large scale patterns, like generational preferences, much easier to spot.

In contrast, the bar chart introduces visual clutter and uses length to represent rank, which can be misleading (especially since lower rankings are better). Small differences in rankings become harder to perceive, and the grouped format forces the reader to scan back and forth. The heatmap is more compact, intuitive, and expressive. It highlights the most important insights with minimal cognitive effort.
