```python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    # create the various mapping and order lists/dictionaries
    # return: list of episodes (in order), episode->name map, list of episode names (in order)
    episodes = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    names = {
        'EI' : 'The Phantom Menace', 'EII' : 'Attack of the Clones', 'EIII' : 'Revenge of the Sith', 
        'EIV': 'A New Hope', 'EV': 'The Empire Strikes Back', 'EVI' : 'The Return of the Jedi'
    }

    # we're also going to use this order to sort, so names_l will now have our sort order
    return episodes, names, [names[ep] for ep in episodes]

episodes, namesDict, namesList = genEpisodeNamesDF()
```

```python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
```

```python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN
    salo = indf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    return(salo)
```

```python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
```

```python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total = len(inpf)
    percs = []

    # loop over each column and calculate the number of people who have seen the movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count them
    # and divide by the percent
    for seen_ep in ['seen_' + ep for ep in episodes]:
        perc = len(inpf[~ pd.isna(inpf[seen_ep])]) / total
        percs.append(perc)

    # at this point percs is holding our percentages

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes],percs))
    seen_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    return(seen_per_df)
```

```python
seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
```

```python
# let's see what's inside
seenPerMovieDF
```

```python
def genPercentVis(indf, names_l=namesList):
    # input: indf, the dataframe as seen_per_df
    # input: names_l - a list of all the movies in series order (default namesList)
    # output: simple altair bar chart
    
    # ok, time to make the chart... let's make a bar chart (use mark_bar)
    bars = alt.Chart(indf).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        )
    )

    # at this point we don't really have a great plot (it's missing the annotations, titles, etc.)
    return(bars)

seenPerMovieVis = genPercentVis(seenPerMovieDF)
```

```python
def augmentPercentVis(base):
    # input: base (the base vis, i.e., bars as above)
    # we're going to overlay the text with the percentages, so let's make another visualization
    # that's just text labels


    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    seen_movies = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=180
    ).properties(
        # add a title
        title="Which 'Star Wars' Movies Have you Seen?"
    )

    return(seen_movies)

# note that we are NOT formatting this in the Five Thirty Eight Style yet... we'll leave that to you to figure out
```

```python
# Recreate this image using Altair
# match the "538 style" as best you can (hint: look at the altair lab at the start of the semester)

def genBestVis(inpf, eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above

    rank_cols = [f"rank_{e}" for e in eps]
    seen_cols = [f"seen_{e}" for e in eps]

    df_seen6 = inpf.dropna(subset=seen_cols).copy()
    n_seen = len(df_seen6)  # should be 471

    ranks_stripped = (
        df_seen6[rank_cols]
        .astype(str)
        .applymap(lambda x: re.sub(r"[^\d\.]", "", x).strip())
        .replace({"": np.nan})
    )
    ranks = ranks_stripped.apply(pd.to_numeric, errors="coerce")

    arr = ranks.to_numpy()
    valid_row = ~np.isnan(arr).all(axis=1)
    arr_valid = arr[valid_row]
    if arr_valid.size == 0:
        raise ValueError("No valid ranks found after cleaning.")

    best_pos = np.nanargmin(arr_valid, axis=1)
    kept_index = ranks.index[valid_row]
    best_eps = pd.Series(np.array(eps)[best_pos], index=kept_index)

    counts = (
        best_eps.value_counts()
        .rename_axis("ep")
        .reset_index(name="count")
        .sort_values("ep", key=lambda s: s.map({e:i for i,e in enumerate(eps)}))
    )
    counts["pct"] = counts["count"] / n_seen
    counts["movie"] = counts["ep"].map(names)

    all_rows = pd.DataFrame({"ep": eps})
    counts = all_rows.merge(counts, on="ep", how="left").fillna({"count": 0, "pct": 0})
    counts["movie"] = counts["ep"].map(names)
    counts["movie"] = pd.Categorical(counts["movie"], categories=[names[e] for e in eps], ordered=True)

    title_main = "What's the best 'Star Wars' Movie?"
    title_sub  = f"Of {n_seen} respondents who have seen all six films"

    base = alt.Chart(counts).encode(
        y=alt.Y(
            "movie:N",
            title=None,
            sort=[names[e] for e in eps],
            axis=alt.Axis(labelPadding=8, labelColor="gray")
        )
    )

    bars = base.mark_bar(size=18, color="#2a9fd6").encode(
        x=alt.X("pct:Q", title=None, axis=None)
    )

    labels = base.mark_text(
        align="left",
        baseline="middle",
        dx=6,
        font="Helvetica Neue",
        fontSize=12,
        color="black"  
    ).encode(
        x="pct:Q",
        text=alt.Text("pct:Q", format=".0%")
    )
    
    chart = (bars + labels).properties(
        title=alt.TitleParams(
            text=title_main,
            subtitle=title_sub,
            font="sans-serif",
            fontSize=24,
            subtitleFontSize=12,
            anchor="start",
            offset=10
        ),
        width=550,
        height=10 + 26 * (len(eps) - 1)
    ).configure(
        background="white"
    ).configure_view(
        fill="white",
        stroke=None
    ).configure_axis(
        grid=False,
        domain=False,
        titleColor="black",
        labelColor="gray"
    ).configure_title(
        color="black",
        subtitleColor="black"
    )

    return chart
```

### 2.2 How people rate the 'Star Wars' movie? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genRateVis(inpf,eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above
    
    rank_cols = [f"rank_{e}" for e in eps]
    seen_cols = [f"seen_{e}" for e in eps]

    df_seen6 = inpf.dropna(subset=seen_cols).copy()
    n_seen = len(df_seen6)  # expected 471

    ranks_stripped = (
        df_seen6[rank_cols]
        .astype(str)
        .applymap(lambda x: re.sub(r"[^\d\.]", "", x).strip())
        .replace({"": np.nan})
    )
    ranks = ranks_stripped.apply(pd.to_numeric, errors="coerce")

    rows = []
    for e in eps:
        mv = names[e]
        r  = ranks[f"rank_{e}"]
        rows += [
            {"movie": mv, "tier": "Top third",    "pct": (r <= 2).sum(skipna=True) / n_seen},
            {"movie": mv, "tier": "Middle third", "pct": ((r >= 3) & (r <= 4)).sum(skipna=True) / n_seen},
            {"movie": mv, "tier": "Bottom third", "pct": ((r >= 5) & (r <= 6)).sum(skipna=True) / n_seen},
        ]
    long_df = pd.DataFrame(rows)
    long_df["movie"] = pd.Categorical(long_df["movie"],
                                      categories=[names[e] for e in eps], ordered=True)

    tier_color = {
        "Top third":    "#5cb85c",
        "Middle third": "#2a9fd6",
        "Bottom third": "#ff1133",
    }

    base = alt.Chart(long_df).encode(
        y=alt.Y("movie:N",
                sort=[names[e] for e in eps],
                title=None,
                axis=alt.Axis(labelPadding=8, labelColor="gray")),
        color=alt.Color("tier:N",
                        scale=alt.Scale(domain=list(tier_color.keys()),
                                        range=[tier_color[k] for k in ["Top third","Middle third","Bottom third"]]),
                        legend=None)
    )

    bars = base.mark_bar(size=18).encode(
        x=alt.X("pct:Q", title=None, axis=None)
    )

    labels = base.mark_text(
        align="left", baseline="middle", dx=6,
        font="sans-serif", fontSize=12, color="black", 
    ).encode(
        x="pct:Q",
        text=alt.Text("pct:Q", format=".0%")
    )

    layered = alt.layer(bars, labels, data=long_df).properties(
        width=80,
        height=10 + 26 * (len(eps) - 1)
    )

    chart = layered.facet(
        column=alt.Column(
            "tier:N",
            sort=["Top third", "Middle third", "Bottom third"],
            title=None, 
            header=alt.Header(
                labelOrient="top",
                labelFont="sans-serif",
                labelFontSize=10,
                labelColor="black",
                labelAngle=0,
                labelPadding=0,
                labelLimit=0 
            )
        )
    ).properties(
        title=alt.TitleParams(
            text="How People Rate the 'Star Wars' Movies",
            subtitle=["How often each film was rated in the top, middle and bottom third",
                      f"(by {n_seen} respondents who have seen all six films)"],
            anchor="start",
            offset=10,
            font="sans-serif", fontSize=24,
            subtitleFont="sans-serif", subtitleFontSize=12
        )
    ).resolve_scale(
        x="shared", y="shared", color="independent"
    ).configure(
        background="white"
    ).configure_view(
        fill="white", stroke=None
    ).configure_axis(
        grid=False, domain=False,
        labelFont="sans-serif", titleFont="sans-serif",
        titleColor="black", labelColor="gray"
    ).configure_title(
        color="black", subtitleColor="black",
        font="sans-serif", subtitleFont="sans-serif"
    ).configure_text(
        font="sans-serif", color="black"
    )

    return chart
```

```python
# Recreate this image using altair here (10 POINTS)

def genFavorabilityVis(inpf):
    # input: inpf, the star wars dataset (as defined at top of file)
    # output: the Altair visualization as described above
    
    characters = ['Han Solo', 'Luke Skywalker', 'Princess Leia Organa', 'Anakin Skywalker', 'Obi Wan Kenobi', 
                  'Emperor Palpatine', 'Darth Vader', 'Lando Calrissian', 'Boba Fett', 'C-3P0', 
                  'R2 D2', 'Jar Jar Binks', 'Padme Amidala', 'Yoda']

    mask_any = inpf[characters].notna().any(axis=1)
    df_clean = inpf.loc[mask_any, characters].copy()
    n_resp = len(df_clean)         # subtitle denominator

    long = df_clean.melt(var_name="character", value_name="resp")

    favorable_set   = {"Very favorably", "Somewhat favorably"}
    neutral_set     = {"Neither favorably nor unfavorably (neutral)"}
    unfavorable_set = {"Somewhat unfavorably", "Very unfavorably"}
    unfamiliar_set  = {"Unfamiliar (N/A)"}

    def bucketize(x):
        x = str(x).strip()
        if x in favorable_set:   return "Favorable"
        if x in neutral_set:     return "Neutral"
        if x in unfavorable_set: return "Unfavorable"
        if x in unfamiliar_set:  return "Unfamiliar"
        return np.nan

    long["bucket"] = long["resp"].map(bucketize)
    long = long.dropna(subset=["bucket"])

    counts = (
        long.groupby(["character", "bucket"], as_index=False)
            .size().rename(columns={"size": "count"})
    )

    all_pairs = pd.MultiIndex.from_product(
        [characters, ["Favorable","Neutral","Unfavorable","Unfamiliar"]],
        names=["character","bucket"]
    ).to_frame(index=False)
    counts = all_pairs.merge(counts, on=["character","bucket"], how="left").fillna({"count": 0})

    counts["pct"] = counts["count"] / n_resp

    fav_order = (counts[counts["bucket"]=="Favorable"]
                 .sort_values("pct", ascending=False)["character"].tolist())
    counts["character"] = pd.Categorical(counts["character"], categories=fav_order, ordered=True)

    palette = {"Favorable":"#5cb85c","Neutral":"#2a9fd6","Unfavorable":"#e34a33","Unfamiliar":"#b3b3b3"}
    bucket_order = ["Favorable","Neutral","Unfavorable","Unfamiliar"]

    base = alt.Chart(counts).encode(
        y=alt.Y("character:N", sort=fav_order, title=None,
                axis=alt.Axis(labelPadding=8, labelColor="gray")),
        color=alt.Color("bucket:N",
                        scale=alt.Scale(domain=bucket_order, range=[palette[b] for b in bucket_order]),
                        legend=None)
    )

    bars = base.mark_bar(size=18).encode(x=alt.X("pct:Q", title=None, axis=None))
    labels = base.mark_text(align="left", baseline="middle", dx=6,
                            font="sans-serif", fontSize=12, color="black"
                           ).encode(x="pct:Q", text=alt.Text("pct:Q", format=".0%"))

    inner = alt.layer(bars, labels, data=counts).properties(
        width=75, height=12 + 20 * (len(fav_order) - 1)
    )

    chart = inner.facet(
        column=alt.Column("bucket:N", sort=bucket_order, title=None,
                          header=alt.Header(labelOrient="top", labelFont="sans-serif",
                                            labelFontSize=12, labelColor="black",
                                            labelAngle=0, labelPadding=4, labelLimit=0))
    ).properties(
        title=alt.TitleParams(
            text="'Star Wars' Characters Favorability Ratings",
            subtitle=f"By {n_resp} respondents",
            anchor="start", offset=10,
            font="sans-serif", fontSize=24,
            subtitleFont="sans-serif", subtitleFontSize=12
        )
    ).resolve_scale(x="shared", y="shared", color="independent").configure(
        background="white"
    ).configure_view(fill="white", stroke=None).configure_axis(
        grid=False, domain=False, labelFont="sans-serif", titleFont="sans-serif",
        titleColor="black", labelColor="gray"
    ).configure_title(
        color="black", subtitleColor="black",
        font="sans-serif", subtitleFont="sans-serif"
    ).configure_text(font="sans-serif", color="black")

    return chart
```

```python
def best_by_age_prepare(inpf, eps=('EI','EII','EIII','EIV','EV','EVI'),
                        age_order=("18-29","30-44","45-60","> 60")):
    rank_cols = [f"rank_{e}" for e in eps]
    seen_cols = [f"seen_{e}" for e in eps]

    df = inpf.dropna(subset=seen_cols).copy()  # saw all six
    # Coerce ranks → numeric
    ranks = df[rank_cols].apply(pd.to_numeric, errors="coerce")
    # Find best episode per respondent
    best_pos = np.argmin(ranks.to_numpy(), axis=1)
    best_eps = pd.Series(np.array(eps)[best_pos], index=df.index, name="episode")
    out = pd.concat([df["Age"], best_eps], axis=1).dropna(subset=["Age"])

    # Percent per Age x Episode
    grp = out.groupby(["Age","episode"]).size().rename("count").reset_index()
    denom = out.groupby("Age").size().rename("n_age").reset_index()
    counts = grp.merge(denom, on="Age")
    counts["pct"] = counts["count"] / counts["n_age"]

    # Fill missing episode rows with zero
    all_pairs = pd.MultiIndex.from_product([denom["Age"].unique(), eps], names=["Age","episode"]).to_frame(index=False)
    counts = all_pairs.merge(counts, on=["Age","episode"], how="left").fillna({"count":0,"n_age":np.nan,"pct":0})
    # Keep original denominators for subtitle
    n_total = int(denom["n_age"].sum())

    # Trilogy color helper
    tril_map = {"EI":"Prequel","EII":"Prequel","EIII":"Prequel",
                "EIV":"Original","EV":"Original","EVI":"Original"}
    counts["trilogy"] = counts["episode"].map(tril_map)

    # Ordering
    counts["episode"] = pd.Categorical(counts["episode"], categories=list(eps), ordered=True)
    counts["Age"] = pd.Categorical(counts["Age"], categories=list(age_order), ordered=True)

    return counts, n_total
```

# Which is better?

I recommend the bar chart as the primary figure. Use the heatmap as a compact overview.

## Bar chart (primary)
- Bar length (position on a shared x-axis) supports quick, precise comparisons within and across age groups.
- Color encodes trilogy (categorical) while length encodes percentage (quantitative), avoiding overloaded color scales.
-End labels fit cleanly at bar ends; no deciphering of color intensity.
- Reading down a list per facet matches how people scan lists; clearer than a grid for rank/magnitude judgments.

## Heatmap (companion)
- One grid shows the whole landscape—broad trends pop (e.g., OT stronger in older cohorts).
- Color intensity is less precise than position; small differences require reading labels.
- Harder for color-blind users; adjacent cells can mask subtle contrasts.

## When to use which?
- **Bar chart:** When you want readers to compare magnitudes and rank episodes across age groups to make an argument.
- **Heatmap:** When you want readers to scan for patterns across many categories at once to get the lay of the land.
