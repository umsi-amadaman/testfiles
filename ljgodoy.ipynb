Seventy-nine percent of those respondents said they had watched at least one of the “Star Wars” films.
```python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    # create the various mapping and order lists/dictionaries
    # return: list of episodes (in order), episode->name map, list of episode names (in order)
    episodes = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    names = {
        'EI' : 'The Phantom Menace', 'EII' : 'Attack of the Clones', 'EIII' : 'Revenge of the Sith', 
        'EIV': 'A New Hope', 'EV': 'The Empire Strikes Back', 'EVI' : 'The Return of the Jedi'
    }

    # we're also going to use this order to sort, so names_l will now have our sort order
    return episodes, names, [names[ep] for ep in episodes]

episodes, namesDict, namesList = genEpisodeNamesDF()
```

```python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
```

```python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN
    salo = indf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    return(salo)
```

```python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
```

```python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total = len(inpf)
    percs = []

    # loop over each column and calculate the number of people who have seen the movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count them
    # and divide by the percent
    for seen_ep in ['seen_' + ep for ep in episodes]:
        perc = len(inpf[~ pd.isna(inpf[seen_ep])]) / total
        percs.append(perc)

    # at this point percs is holding our percentages

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes],percs))
    seen_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    return(seen_per_df)
```

```python
seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
```

```python
# let's see what's inside
seenPerMovieDF
```

```python
def genPercentVis(indf, names_l=namesList):
    # input: indf, the dataframe as seen_per_df
    # input: names_l - a list of all the movies in series order (default namesList)
    # output: simple altair bar chart
    
    # ok, time to make the chart... let's make a bar chart (use mark_bar)
    bars = alt.Chart(indf).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        )
    )

    # at this point we don't really have a great plot (it's missing the annotations, titles, etc.)
    return(bars)

seenPerMovieVis = genPercentVis(seenPerMovieDF)
```

```python
# display it
seenPerMovieVis
```

```python
def augmentPercentVis(base):
    # input: base (the base vis, i.e., bars as above)
    # we're going to overlay the text with the percentages, so let's make another visualization
    # that's just text labels


    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    seen_movies = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=180
    ).properties(
        # add a title
        title="Which 'Star Wars' Movies Have you Seen?"
    )

    return(seen_movies)

# note that we are NOT formatting this in the Five Thirty Eight Style yet... we'll leave that to you to figure out
```

```python
# Recreate this image using Altair
# match the "538 style" as best you can (hint: look at the altair lab at the start of the semester)

def genBestVis(inpf,eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above
    # YOUR CODE HERE

    rank_cols = ['rank_' + ep for ep in eps]

    ranks_clean = inpf.copy()
    for col in rank_cols:
        ranks_clean[col] = pd.to_numeric(ranks_clean[col], errors='coerce')
    qualified = ranks_clean.dropna(subset=rank_cols, how='any')
    total = len(qualified)

    counts = [(qualified['rank_' + ep] == 1).sum() for ep in eps]
    percentages = [cnt / total if total > 0 else 0 for cnt in counts]
    result = pd.DataFrame({
        'Name': [names[ep] for ep in eps],
        'Percentage': percentages
    })

    bars = alt.Chart(result).mark_bar(
        size=22,
        color='#008fd5'
    ).encode(
        x=alt.X('Percentage:Q', axis=None, scale=alt.Scale(domain=[0, max(0.01, max(percentages) * 1.1)])),
        y=alt.Y('Name:N',
                axis=alt.Axis(tickCount=6, title=''),
                sort=names_l)
    )
    text = alt.Chart(result).mark_text(
        align='left',
        baseline='middle',
        dx=5,
        fontSize=16,
        font='Source Sans Pro',
        color='black'
    ).encode(
        x='Percentage:Q',
        y=alt.Y('Name:N', sort=names_l),
        text=alt.Text('Percentage:Q', format='.0%')
    )
    chart = alt.layer(bars, text).configure_view(
        strokeWidth=0,
        fill='white'
    ).properties(
        width=520,
        height=220,
        title={
            "text": ["What's the best 'Star Wars' Movie?"],
            "subtitle": [f'Of {total} respondents who have seen all six films'],
            "align": "left",
            "fontSize": 27,
            "subtitleFontSize": 15,
            "subtitleColor": "Black",
            "font": "Source Sans Pro",
            "color": "Black"
        },
        background = "white"
    ).configure_axis(
        labelFontSize=13,
        labelFont='Source Sans Pro',
        labelColor='Black'
    ).configure_title(
        anchor='start'
    )
    return chart
```

### 2.2 How people rate the 'Star Wars' movie? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genRateVis(inpf,eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above
    
    # YOUR CODE HERE

    rank_cols = ['rank_' + ep for ep in eps]
    ranks_clean = inpf.copy()
    for col in rank_cols:
        ranks_clean[col] = pd.to_numeric(ranks_clean[col], errors='coerce')
    qualified = ranks_clean.dropna(subset=rank_cols, how='any')
    total = len(qualified)
    thirds_dict = {'Top third': [1,2], 'Middle third': [3,4], 'Bottom third': [5,6]}
    chart_colors = {'Top third': '#73c16c', 'Middle third': '#37a6dd', 'Bottom third': '#df4b37'}
    
    percent_data = []
    for ep in eps:
        col = 'rank_' + ep
        for label, ranks in thirds_dict.items():
            count = qualified[col].isin(ranks).sum()
            percent = count / total if total > 0 else 0
            percent_data.append({'Name': names[ep], 'Third': label, 'Percentage': percent})
    percent_df = pd.DataFrame(percent_data)
    
    charts = []
    thirds_order = ['Top third','Middle third','Bottom third']
    for third in thirds_order:
        color = chart_colors[third]
        label_color = 'black' if third != 'Bottom third' else '#990000'
        chart_data = percent_df[percent_df['Third'] == third]
        header = alt.Chart(pd.DataFrame({'Header': [third]})).mark_text(
            fontSize=18, font='Source Sans Pro', fontWeight="bold", color='Black', dy=-12
        ).encode(
            text='Header:N'
        ).properties(
            width=120,
            height=20
        )
        base = alt.Chart(chart_data).mark_bar(
            size=20, color=color
        ).encode(
            x=alt.X('Percentage:Q', axis=None),
            y=alt.Y('Name:N', title='', sort=names_l)
        ).properties(
            width=120,
            height=200
        )
        text = alt.Chart(chart_data).mark_text(
            align='left',
            baseline='middle',
            dx=3,
            fontSize=16,
            font='Source Sans Pro',
            color=label_color
        ).encode(
            x='Percentage:Q',
            y=alt.Y('Name:N', sort=names_l),
            text=alt.Text('Percentage:Q', format='.0%')
        ).properties(
            width=120,
            height=200
        )
        charts.append(header & (base + text))
    

    final = alt.concat(*charts, spacing=18).configure_view(
        strokeWidth=0,
        fill='white'
    ).properties(
        title={
            "text": ["How People Rate the 'Star Wars' Movies"],
            "subtitle": [
                f"How often each film was rated in the top, middle and bottom third",
                f"(by {total} respondents who have seen all six films)"
            ],
            "align": "left",
            "fontSize": 27,
            "subtitleFontSize": 15,
            "subtitleColor": "Black",
            "font": "Source Sans Pro",
            "color": "Black"
        },
        background = "white"
    ).configure_axis(
        labelFontSize=13,
        labelFont='Source Sans Pro',
        labelColor='Black'
    ).configure_title(
        anchor='start'
    )
    return final
```

```python
# Recreate this image using altair here (10 POINTS)

def genFavorabilityVis(inpf):
    # input: inpf, the star wars dataset (as defined at top of file)
    # output: the Altair visualization as described above
    
    # YOUR CODE HERE
    # Character columns in required chart order
    character_cols = [
        'Luke Skywalker', 'Han Solo', 'Princess Leia Organa', 'Obi Wan Kenobi',
        'Yoda', 'R2 D2', 'C-3P0', 'Anakin Skywalker', 'Darth Vader',
        'Lando Calrissian', 'Padme Amidala', 'Boba Fett',
        'Emperor Palpatine', 'Jar Jar Binks'
    ]

    norm_map = {
        'very favorably': 'Favorable',
        'somewhat favorably': 'Favorable',
        'neither favorably nor unfavorably (neutral)': 'Neutral',
        'somewhat unfavorably': 'Unfavorable',
        'very unfavorably': 'Unfavorable',
        'unfamiliar': 'Unfamiliar'
    }
    group_order = ['Favorable', 'Neutral', 'Unfavorable', 'Unfamiliar']
    chart_colors = {
        'Favorable': '#73c16c',
        'Neutral': '#37a6dd',
        'Unfavorable': '#df4b37',
        'Unfamiliar': '#888888'
    }

    clean = inpf.copy()
    data = []
    total = 0

    valid = clean.dropna(subset=character_cols, how='all')
    total = len(valid)

    for char in character_cols:
        groups = {g: 0 for g in group_order}
        vals = valid[char].astype(str).str.lower().str.strip().replace('nan', np.nan)
        for v in vals:
            group = norm_map.get(v, None)
            if group is None:
                if pd.isna(v) or v in ['nan', '']:
                    groups['Unfamiliar'] += 1
                else:
                    found = False
                    for k,g in norm_map.items():
                        if k in str(v):
                            groups[g] += 1
                            found = True
                            break
                    if not found:
                        groups['Unfamiliar'] += 1
            else:
                groups[group] += 1
        for g in group_order:
            percent = groups[g] / total if total > 0 else 0
            data.append({'Character': char, 'Group': g, 'Percentage': percent})

    df = pd.DataFrame(data)

    charts = []
    for group in group_order:
        color = chart_colors[group]
        header = alt.Chart(pd.DataFrame({'Header': [group]})).mark_text(
            dy=-12, fontSize=18, font='Source Sans Pro', fontWeight='bold', color='black'
        ).encode(
            text='Header:N'
        ).properties(width=110)
        subset = df[df['Group'] == group]

        axis_label_config = alt.Axis(
            labelColor='#888888',
            labelFont='Source Sans Pro',
            labelFontSize=13,
            title=''
        )

        bars = alt.Chart(subset).mark_bar(
            size=20, color=color
        ).encode(
            x=alt.X('Percentage:Q', axis=None),
            y=alt.Y('Character:N', sort=character_cols, axis=axis_label_config)
        ).properties(
            width=110,
            height=310
        )
        text = alt.Chart(subset).mark_text(
            align='left', baseline='middle', dx=4, fontSize=16,
            font='Source Sans Pro', color=color
        ).encode(
            x='Percentage:Q',
            y=alt.Y('Character:N', sort=character_cols),
            text=alt.Text('Percentage:Q', format='.0%')
        ).properties(
            width=110,
            height=310
        )
        charts.append(header & (bars + text))

    final = alt.concat(*charts, spacing=18).configure_view(
        strokeWidth=0,
        fill='white'
    ).properties(
        title={
            "text": ["'Star Wars' Characters Favorability Ratings"],
            "subtitle": [f"By {total} respondents"],
            "align": "left",
            "fontSize": 27,
            "subtitleFontSize": 17,
            "subtitleColor": "Black",
            "font": "Source Sans Pro",
            "color": "black"
        },
        background = "white"
    ).configure_axis(
        labelFontSize=13,
        labelFont='Source Sans Pro',
        labelColor='#888888'
    ).configure_title(
        anchor='start',
        color='black'
    )
    return final
```

```python
# Recreate this image using altair here (10 POINTS)

def genFirstShotVis(inpf):
    # input: inpf, the star wars dataset
    # output: the Altair visualization as described above
    
    # YOUR CODE HERE
    col = 'Which character shot first?'

    mapping = {
        'han': 'Han',
        'greedo': 'Greedo',
        "i dont understand this question": "I don't understand\nthis question",
        "i don't understand this question": "I don't understand\nthis question",
        "I dont understand this question": "I don't understand\nthis question",
        "I don't understand this question": "I don't understand\nthis question"
    }
    allowed = ['Han', 'Greedo', "I don't understand\nthis question"]
    data = inpf.copy()
    responses = data[col].dropna().astype(str).str.lower().str.strip()
    mapped = responses.map(lambda x: mapping.get(x, None))
    mapped = mapped[mapped.isin(allowed)]
    total = len(mapped)
    percentages = [round((mapped == label).sum() / total * 100) for label in allowed]
    result = pd.DataFrame({"Response": allowed, "Percentage": percentages})

    result["Response"] = result["Response"].replace("I don't understand\nthis question", "I don't understand\nthis question")

    bars = alt.Chart(result).mark_bar(size=28, color='#2196F3').encode(
        x=alt.X("Percentage:Q", axis=None, scale=alt.Scale(domain=[0, max(percentages) + 10])),
        y=alt.Y("Response:N",
                axis=alt.Axis(
                    title="", 
                    labelFontSize=18, 
                    labelFont='Source Sans Pro',
                    labelLimit=300,
                    labelColor='#000000'
                ),
                sort=allowed)
    ).properties(
        width=420,
        height=190
    )

    text = alt.Chart(result).mark_text(
        align="left", baseline="middle", dx=6, font="Source Sans Pro", fontSize=21, color="black"
    ).encode(
        x="Percentage:Q",
        y=alt.Y("Response:N", sort=allowed),
        text=alt.Text("Percentage:Q", format='.0f')
    ).properties(
        width=420,
        height=190
    )

    chart = (bars + text).configure_view(
        strokeWidth=0,
        fill="white"
    ).properties(
        title={
            "text": ["Who Shot First?"],
            "subtitle": [f"According to {total} respondents"],
            "align": "left",
            "fontSize": 30,
            "subtitleFontSize": 21,
            "subtitleColor": "Black",
            "font": "Source Sans Pro",
            "color": "black"
        },
        background = "white"
    ).configure_axis(
        labelFontSize=18,
        labelFont="Source Sans Pro"
    ).configure_title(
        anchor="start",
        color="black"
    )
    return chart

```

```python
# YOUR CODE HERE
def characterFavorabilityStacked(sw):

    characters = [
        'Luke Skywalker', 'Han Solo', 'Princess Leia Organa', 'Obi Wan Kenobi',
        'Yoda', 'R2 D2', 'C-3P0', 'Anakin Skywalker', 'Darth Vader',
        'Lando Calrissian', 'Padme Amidala', 'Boba Fett',
        'Emperor Palpatine', 'Jar Jar Binks'
    ]
    favor_map = {
        'very favorably': 'Favorable',
        'somewhat favorably': 'Favorable',
        'neither favorably nor unfavorably (neutral)': 'Neutral',
        'somewhat unfavorably': 'Unfavorable',
        'very unfavorably': 'Unfavorable',
        'unfamiliar': None
    }
    group_order = ['Favorable', 'Neutral', 'Unfavorable']
    colors = {'Favorable': '#45b36e', 'Neutral': '#549dd3', 'Unfavorable': '#e23b3b'}
    data = []

    valid = sw.dropna(subset=characters, how='all')
    total = len(valid)

    for char in characters:
        votes = valid[char].astype(str).str.lower().str.strip().replace('nan', np.nan)
        counts = {g: 0 for g in group_order}
        for v in votes:
            norm = favor_map.get(v, None)
            if norm is None:
                if any(k in str(v) for k in favor_map.keys() if favor_map[k] is not None):
                    for k in favor_map.keys():
                        if k in str(v):
                            norm2 = favor_map[k]
                            if norm2 is not None:
                                counts[norm2] += 1
            else:
                counts[norm] += 1

        for g in group_order:
            data.append({
                'Character': char,
                'Group': g,
                'Percentage': counts[g]/total if total else 0,
                'Net': (counts['Favorable'] - counts['Unfavorable'])/total if total else 0
            })

    df = pd.DataFrame(data)
    net_df = df[df['Group'] == 'Favorable'][['Character','Net']]
    sort_order = net_df.sort_values('Net', ascending=False)['Character'].tolist()

    chart = alt.Chart(df).mark_bar().encode(
        x=alt.X('Percentage:Q', stack='normalize', axis=alt.Axis(format='%'), title="Share of respondents"),
        y=alt.Y('Character:N', sort=sort_order, axis=alt.Axis(title='', labelFontSize=13, labelFont='Source Sans Pro', labelColor='#888888')),
        color=alt.Color('Group:N',
                        scale=alt.Scale(domain=group_order, range=[colors[g] for g in group_order]),
                        legend=alt.Legend(title="")),
        order=alt.Order('Group', sort='descending'),   # Favorable at left
        tooltip=['Character:N', 'Group:N', alt.Tooltip('Percentage:Q', format='.0%')]
    ).properties(
        width=540,
        height=380,
        title="Star Wars Character Favorability: Sorted by Net Favorability",
        background='White'
    ).configure_view(
        strokeWidth=0,
        fill='white'
    ).configure_axis(
        grid=False,
        labelFontSize=13,
        labelFont='Source Sans Pro',
        labelColor="#888888"
    ).configure_title(
        anchor='start',
        fontSize=27,
        font='Source Sans Pro',
        color='black'
    )
    return chart

characterFavorabilityStacked(sw)
```
