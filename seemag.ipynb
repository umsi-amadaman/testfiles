A pie chart can be used to depict the percentage of respondents who have seen each of the Star Wars movies, but it would be less effective. A pie chart is best used to illustrate part-to-whole relationships or when there are only a few categories to show proportional dominance. In this case, the percentage variation between films is very small, and the data is not being compared as a whole but rather as an illustration of how many of the 835 respondents have seen each of the six films. Furthermore, the bar aspect of the chart is easier for humans to interpret because people are better at judging aligned positions than angles or areas. It can also become cognitively overwhelming if a viewer has to distinguish between too many slices or colors, and small differences may go unnoticed.

The horizontal bar chart, on the other hand, is both expressive and effective. Quantitative data is most effectively represented through length and position encodings, allowing for faster and more accurate magnitude comparisons. According to Gestalt principles, alignment, proximity, and similarity make the chart visually coherent and easier to understand. This design supports perceptual clarity, reduces cognitive load, and allows viewers to immediately recognize relative differences between films making the horizontal bar chart the superior choice for conveying this information.

In the example of the pie chart attached, the first pie chart replicates the bar chart values directly into pie slices. This approach is misleading because the purpose of a pie chart is to show each category as a part of a single whole. When all the percentages are added together, they exceed 100%, creating a false impression that the categories are mutually exclusive parts of one dataset. This misrepresents the data and can easily confuse viewers about what the visualization actually conveys. The corrected pie chart below shows the proportional relationship of movies to each other, but that still fails to answer the question, “Which Star Wars movies have you seen?” since viewers could have seen multiple films, making a part-to-whole representation inappropriate for this type of data.

```python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    # create the various mapping and order lists/dictionaries
    # return: list of episodes (in order), episode->name map, list of episode names (in order)
    episodes = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    names = {
        'EI' : 'The Phantom Menace', 'EII' : 'Attack of the Clones', 'EIII' : 'Revenge of the Sith', 
        'EIV': 'A New Hope', 'EV': 'The Empire Strikes Back', 'EVI' : 'The Return of the Jedi'
    }

    # we're also going to use this order to sort, so names_l will now have our sort order
    return episodes, names, [names[ep] for ep in episodes]

episodes, namesDict, namesList = genEpisodeNamesDF()
```

```python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
```

```python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN
    salo = indf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    return(salo)
```

```python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
```

```python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total = len(inpf)
    percs = []

    # loop over each column and calculate the number of people who have seen the movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count them
    # and divide by the percent
    for seen_ep in ['seen_' + ep for ep in episodes]:
        perc = len(inpf[~ pd.isna(inpf[seen_ep])]) / total
        percs.append(perc)

    # at this point percs is holding our percentages

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes],percs))
    seen_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    return(seen_per_df)
```

```python
seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
```

```python
def genPercentVis(indf, names_l=namesList):
    # input: indf, the dataframe as seen_per_df
    # input: names_l - a list of all the movies in series order (default namesList)
    # output: simple altair bar chart
    
    # ok, time to make the chart... let's make a bar chart (use mark_bar)
    bars = alt.Chart(indf).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        )
    )

    # at this point we don't really have a great plot (it's missing the annotations, titles, etc.)
    return(bars)

seenPerMovieVis = genPercentVis(seenPerMovieDF)
```

```python
def augmentPercentVis(base):
    # input: base (the base vis, i.e., bars as above)
    # we're going to overlay the text with the percentages, so let's make another visualization
    # that's just text labels


    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    seen_movies = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=180
    ).properties(
        # add a title
        title="Which 'Star Wars' Movies Have you Seen?"
    )

    return(seen_movies)

# note that we are NOT formatting this in the Five Thirty Eight Style yet... we'll leave that to you to figure out
```

### 2.2 How people rate the 'Star Wars' movie? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genRateVis(inpf,eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above
    
    # YOUR CODE HERE
    #### PART 1 ###
    rate_movies = inpf.dropna(subset=['rank_' + ep for ep in eps], how='any').copy()
    for ep in eps:
        rate_movies['rank_' + ep] = pd.to_numeric(rate_movies['rank_' + ep], errors='coerce')

    total = len(rate_movies)
    count = []
    for ep in eps:
        col = 'rank_' + ep
        ranks = pd.to_numeric(inpf[col], errors='coerce')
        valid = ranks.dropna()
        total = len(valid)
    
        top = ((valid == 1) | (valid == 2)).sum() / total
        middle = ((valid == 3) | (valid == 4)).sum() / total
        bottom = ((valid == 5) | (valid == 6)).sum() / total
    
        count.append((names[ep], 'Top third', top))
        count.append((names[ep], 'Middle third', middle))
        count.append((names[ep], 'Bottom third', bottom))

    df = pd.DataFrame(count, columns=['Name', 'RatingGroup', 'Percentage'])
    df['Name'] = pd.Categorical(df['Name'], categories=names_l, ordered=True)
    df['RatingGroup'] = pd.Categorical(
        df['RatingGroup'],
        categories=['Top third', 'Middle third', 'Bottom third'],
        ordered=True
    )


    #### PART 2 ###
    base = (
        alt.Chart(df, width=90, height=120)
        .mark_bar(size=16)   
        .encode(
            x=alt.X('Percentage:Q', axis=None, scale=alt.Scale(domain=[0, 0.8])),
            y=alt.Y('Name:N', sort=names_l, axis=alt.Axis(title='', labelColor='#999', labelFontSize=11)
            ),
            color=alt.Color(
                'RatingGroup:N',
                scale=alt.Scale(
                    domain=['Top third', 'Middle third', 'Bottom third'],
                    range=['#8AC146', '#008fd5', '#FF0000']
                ),
                legend=None
            ),
            tooltip=['Name', 'RatingGroup', alt.Tooltip('Percentage:Q', format='.0%')]
        )
    )

    text = (
        base.mark_text(
            align='left', 
            baseline='middle',
            dx=3,            
            fontWeight='bold',
            fontSize=10
        )
        .encode(
            text=alt.Text('Percentage:Q', format='.0%'),
            color=alt.Color(    
                'RatingGroup:N',
                scale=alt.Scale(
                    domain=['Top third', 'Middle third', 'Bottom third'],
                    range=['#8AC146', '#4169E1', '#E45756']
                ),
                legend=None
            )
        )
    )
    
    layered = (base + text).properties(
    width=40,
    height=80
    )
    
    chart = (
        (base + text)
        .facet(
            facet=alt.Facet(
                'RatingGroup:N',
                sort=['Top third', 'Middle third', 'Bottom third'],
                header=alt.Header(
                    labelFontSize=13,
                    labelFontWeight='bold',
                    labelColor='black',
                    title=None,
                    labelAlign='center'
                )
            ),
            columns=3
        )
        .resolve_scale(x='shared')
        .properties(
            title={
                "text": "How People Rate the 'Star Wars' Movies",
                "subtitle": [
                "How often each film was rated in the top, middle, and bottom third",
                "(by 471 respondents who have seen all six films)"
                ],
                "subtitleColor": "#555",
                "fontSize": 22,
                "fontWeight": "bold",
                "subtitleFontSize": 12,
                "subtitleFontStyle": "bold",
                "anchor": "start"
            }
        )
        .configure_view(strokeWidth=0)
        .configure_facet(spacing=10)
        .configure_axis(grid=False, domain=False)
        .configure_axisX(labels=False, ticks=False, title=None)
    )

    return chart
    
```

```python
# Recreate this image using altair here (10 POINTS)

def genFavorabilityVis(inpf):
    # input: inpf, the star wars dataset (as defined at top of file)
    # output: the Altair visualization as described above
    
    # YOUR CODE HERE
    ### ------------------ Part 1 ------------------###
    chars = [
        'Luke Skywalker', 'Han Solo', 'Princess Leia Organa', 'Obi Wan Kenobi', 'Yoda',
        'R2 D2', 'C-3P0', 'Anakin Skywalker', 'Darth Vader', 'Lando Calrissian',
        'Padme Amidala', 'Boba Fett', 'Emperor Palpatine', 'Jar Jar Binks'
    ]

    df = inpf.melt(
        value_vars=chars,
        var_name='Character',
        value_name='Favorability'
    ).dropna(subset=['Favorability'])

    df['Favorability'] = df['Favorability'].str.strip()

    mapping = {
        'Very favorably': 'Favorable',
        'Somewhat favorably': 'Favorable',
        'Neither favorably nor unfavorably (neutral)': 'Neutral',
        'Somewhat unfavorably': 'Unfavorable',
        'Very unfavorably': 'Unfavorable',
        'Unfamiliar (N/A)': 'Unfamiliar'
    }

    df['Favorability'] = df['Favorability'].replace(mapping)


    order_fav = ['Favorable', 'Neutral', 'Unfavorable', 'Unfamiliar']
    df = df[df['Favorability'].isin(order_fav)]

    grouped = (
        df.groupby(['Character', 'Favorability'])
        .size()
        .reset_index(name='Count')
    )

    totals = grouped.groupby('Character')['Count'].sum().reset_index(name='Total')
    merged = grouped.merge(totals, on='Character')
    merged['Percentage'] = merged['Count'] / merged['Total']

    merged['Favorability'] = pd.Categorical(
        merged['Favorability'], categories=order_fav, ordered=True
    )
    merged['Character'] = pd.Categorical(merged['Character'], categories=chars, ordered=True)

    ### ------------------ Part 2 ------------------###
    base = (
        alt.Chart(merged, width=120, height=200)
        .mark_bar(size=12)
        .encode(
        x=alt.X('Percentage:Q', axis=None, scale=alt.Scale(domain=[0, 1])),
        y=alt.Y(
            'Character:N',
            sort=chars,
            axis=alt.Axis(
                title='',
                labelColor='#666',
                labelFontSize=11,
                labelPadding=6          
            ),
            scale=alt.Scale(
                paddingInner=0.35,
                paddingOuter=0.1
            )
        ),
            color=alt.Color(
                'Favorability:N',
                scale=alt.Scale(
                    domain=order_fav,
                    range=['#73AF48', '#008fd5', '#E45756', '#989898']
                ),
                legend=None
            ),
            tooltip=['Character', 'Favorability', alt.Tooltip('Percentage:Q', format='.0%')]
        )
    )

    text = (
        base.mark_text(
            align='left',
            baseline='middle',
            dx=3,
            fontSize=10,
            fontWeight='bold'
        )
        .encode(
            text=alt.Text('Percentage:Q', format='.0%'),
            color=alt.value('black')
        )
    )

    chart = (
        (base + text)
        .facet(
            facet=alt.Facet(
                'Favorability:N',
                sort=order_fav,
                header=alt.Header(
                    labelFontSize=13,
                    labelFontWeight='bold',
                    title=None,
                    labelColor='black',
                    labelAlign='center'
                )
            ),
            columns=4
        )
        .resolve_scale(x='independent')
        .properties(
            title={
                "text": "‘Star Wars’ Character Favorability Ratings",
                "subtitle": ["By 834 respondents"],
                "fontSize": 22,
                "fontWeight": "bold",
                "subtitleFontSize": 13,
                "subtitleFontStyle": "bold",
                "subtitleColor": "#555"
                
            }
        )
        .configure_view(strokeWidth=0)
        .configure_facet(spacing=10)
        .configure_axis(grid=False, domain=False)
        .configure_axisX(labels=False, ticks=False, title=None)
    )

    return chart
```

```python
# Recreate this image using altair here (10 POINTS)

def genFirstShotVis(inpf):
    # input: inpf, the star wars dataset
    # output: the Altair visualization as described above
    
    # YOUR CODE HERE
    colname = [c for c in inpf.columns if "shot first" in c.lower()][0]

    df = (
        inpf[[colname]]
        .dropna()
        .rename(columns={colname: "Response"})
    )
    
    df["Response"] = df["Response"].str.strip().str.replace("  ", " ")

    grouped = (
        df["Response"]
        .value_counts(normalize=True)
        .reset_index()
        .rename(columns={"index": "Response", "Response": "Percentage"})
    )

    order = ["Han", "Greedo", "I don't understand this question"]
    grouped["Response"] = pd.Categorical(grouped["Response"], categories=order, ordered=True)
    grouped = grouped.sort_values("Response")


    base = (
        alt.Chart(grouped, width=450, height=180)
        .mark_bar(size=35)
        .encode(
            x=alt.X(
                "Percentage:Q",
                axis=None,
                scale=alt.Scale(domain=[0, 0.45])
            ),
            y=alt.Y(
                "Response:N",
                sort=order,
                axis=alt.Axis(title=None, labelFontSize=14, labelColor="#444", labelLimit = 0)
            ),
            color=alt.value("#008fd5"), 
            tooltip=[
                alt.Tooltip("Response:N", title="Answer"),
                alt.Tooltip("Percentage:Q", format=".0%")
            ]
        )
    )

    text = (
        alt.Chart(grouped)
        .mark_text(
            align="left",
            baseline="middle",
            dx=5,
            fontSize=13,
            fontWeight="bold"
        )
        .encode(
            x=alt.X("Percentage:Q"),
            y=alt.Y("Response:N", sort=order),
            text=alt.Text("Percentage:Q", format=".0%"),
            color=alt.value("black")
        )
    )

    chart = (
        (base + text)
        .properties(
            title={
                "text": "Who Shot First?",
                "subtitle": ["According to 834 respondents"],
                "fontSize": 24,
                "fontWeight": "bold",
                "subtitleFontSize": 14,
                "subtitleFontStyle": "bold",
                "subtitleColor": "#555",
                "anchor": "start"
            }
        )
        .configure_view(strokeWidth=0)
        .configure_axis(grid=False, domain=False)
    )

    return chart
```
