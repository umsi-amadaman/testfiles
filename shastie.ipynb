```python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    # create the various mapping and order lists/dictionaries
    # return: list of episodes (in order), episode->name map, list of episode names (in order)
    episodes = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    names = {
        'EI' : 'The Phantom Menace', 'EII' : 'Attack of the Clones', 'EIII' : 'Revenge of the Sith', 
        'EIV': 'A New Hope', 'EV': 'The Empire Strikes Back', 'EVI' : 'The Return of the Jedi'
    }

    # we're also going to use this order to sort, so names_l will now have our sort order
    return episodes, names, [names[ep] for ep in episodes]

episodes, namesDict, namesList = genEpisodeNamesDF()
```

```python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
```

```python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN
    salo = indf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    return(salo)
```

```python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
```

```python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total = len(inpf)
    percs = []

    # loop over each column and calculate the number of people who have seen the movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count them
    # and divide by the percent
    for seen_ep in ['seen_' + ep for ep in episodes]:
        perc = len(inpf[~ pd.isna(inpf[seen_ep])]) / total
        percs.append(perc)

    # at this point percs is holding our percentages

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes],percs))
    seen_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    return(seen_per_df)
```

```python
seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
```

```python
# let's see what's inside
seenPerMovieDF
```

```python
def genPercentVis(indf, names_l=namesList):
    # input: indf, the dataframe as seen_per_df
    # input: names_l - a list of all the movies in series order (default namesList)
    # output: simple altair bar chart
    
    # ok, time to make the chart... let's make a bar chart (use mark_bar)
    bars = alt.Chart(indf).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        )
    )

    # at this point we don't really have a great plot (it's missing the annotations, titles, etc.)
    return(bars)

seenPerMovieVis = genPercentVis(seenPerMovieDF)
```

```python
# display it
seenPerMovieVis
```

```python
def augmentPercentVis(base):
    # input: base (the base vis, i.e., bars as above)
    # we're going to overlay the text with the percentages, so let's make another visualization
    # that's just text labels


    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    seen_movies = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=180
    ).properties(
        # add a title
        title="Which 'Star Wars' Movies Have you Seen?"
    )

    return(seen_movies)

# note that we are NOT formatting this in the Five Thirty Eight Style yet... we'll leave that to you to figure out
```

```python
# Recreate this image using altair here (10 POINTS)

def genRateVis(inpf,eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above
    
    # YOUR CODE HERE

    all_movies_seen_df = inpf[
        (inpf['seen_EI'] == 'Star Wars: Episode I  The Phantom Menace') &
        (inpf['seen_EII'] == 'Star Wars: Episode II  Attack of the Clones') &
        (inpf['seen_EIII'] == 'Star Wars: Episode III  Revenge of the Sith') &
        (inpf['seen_EIV'] == 'Star Wars: Episode IV  A New Hope') &
        (inpf['seen_EV'] == 'Star Wars: Episode V The Empire Strikes Back') &
        (inpf['seen_EVI'] == 'Star Wars: Episode VI Return of the Jedi')
    ]

    total = len(all_movies_seen_df)
    percsTop = []
    percsMid = []
    percsBot = []


    # loop over each column and calculate the number of people who have rated each movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count up the ones that 
    # rank each movie as first
    # and divide by the percent
    for rank_ep in ['rank_' + ep for ep in eps]:
        # the rank of the episode is equal to one
        perctop = len(all_movies_seen_df[(all_movies_seen_df[rank_ep] == '1') | (all_movies_seen_df[rank_ep] == '2')]) / total
        percsTop.append(perctop)
        percmid = len(all_movies_seen_df[(all_movies_seen_df[rank_ep] == '3') | (all_movies_seen_df[rank_ep] == '4')]) / total
        percsMid.append(percmid)
        percbot = len(all_movies_seen_df[(all_movies_seen_df[rank_ep] == '5') | (all_movies_seen_df[rank_ep] == '6')]) / total
        percsBot.append(percbot)

         

    # at this point percs is holding our percentages

    # tuple pairing for each top, mid, bottom
    df_top = pd.DataFrame({'Name': [names[ep] for ep in eps], 'Percentage': percsTop})
    df_mid = pd.DataFrame({'Name': [names[ep] for ep in eps], 'Percentage': percsMid})
    df_bot = pd.DataFrame({'Name': [names[ep] for ep in eps], 'Percentage': percsBot})
    
    color_top, color_mid, color_bot = '#8CC152', '#4A89DC', "red"

    bars_top = alt.Chart(df_top).mark_bar(size=26, color=color_top).encode(
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            'Name:N',
             axis=alt.Axis(tickCount=5, title='',labelColor='gray'),
             sort=names_l,
            scale=alt.Scale(paddingInner=0.1, paddingOuter=0.05))
    )

    bars_mid = alt.Chart(df_mid).mark_bar(size=26, color=color_mid).encode(
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            'Name:N',
             axis=alt.Axis(title=None, ticks=False, labels=False),
             sort=names_l,
        )
    )

    bars_bot = alt.Chart(df_bot).mark_bar(size=26, color=color_bot).encode(
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            'Name:N',
             axis=alt.Axis(title=None, ticks=False, labels=False),
             sort=names_l
        )
    )



    text_top = bars_top.mark_text(
        align='left', baseline='middle', dx=3, fontSize=12, color=color_top
    ).encode(
        text=alt.Text('Percentage:Q', format='.0%')
    )
    text_mid = bars_mid.mark_text(
        align='left', baseline='middle', dx=3, fontSize=12, color=color_mid
    ).encode(
        text=alt.Text('Percentage:Q', format='.0%')
    )
    text_bot = bars_bot.mark_text(
        align='left', baseline='middle', dx=3, fontSize=12, color=color_bot
    ).encode(
        text=alt.Text('Percentage:Q', format='.0%')
    )

    chart_top = (bars_top + text_top).properties(
        title=alt.TitleParams("Top third", anchor='middle', fontSize=14, fontWeight='normal'),
        width=120,
        height=180
    )
    chart_mid = (bars_mid + text_mid).properties(
        title=alt.TitleParams("Middle third", anchor='middle', fontSize=14, fontWeight='normal'),
        width=120,
        height=180
    )
    chart_bot = (bars_bot + text_bot).properties(
        title=alt.TitleParams("Bottom third", anchor='middle', fontSize=14, fontWeight='normal'),
        width=120,
        height=180
    )
    
    combined = (alt.hconcat(chart_top, chart_mid, chart_bot).resolve_scale(y='shared').properties(
        title={
            "text": "How People Rate the 'Star Wars' Movies",
            "subtitle": [
                "How often each film was rated in the top, middle and bottom third",
                "(by 471 respondents who have seen all six films)"
            ],
            "anchor": "start",
            "offset": 30
        }
    ).configure_view(
        strokeWidth=0  
    ).configure_title(
        fontSize=24,
        subtitleFontSize=14,
        subtitlePadding=4
    ))
    return combined
```

### 2.3 Star Wars' Characters Favorability Ratings. Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genFavorabilityVis(inpf):
    # input: inpf, the star wars dataset (as defined at top of file)
    # output: the Altair visualization as described above
    
    # YOUR CODE HERE
    seenAtLeastOneDF = getSeenAtLeastOneDF(inpf, episodes)
    total = len(seenAtLeastOneDF)

    favorable = []
    neutral = []
    unfavorable = []
    unfamiliar = []
    
    characters = ['Han Solo',
       'Luke Skywalker', 'Princess Leia Organa', 'Anakin Skywalker',
       'Obi Wan Kenobi', 'Emperor Palpatine', 'Darth Vader',
       'Lando Calrissian', 'Boba Fett', 'C-3P0', 'R2 D2', 'Jar Jar Binks',
       'Padme Amidala', 'Yoda']

    for c in characters:
            percfavorable = len(seenAtLeastOneDF[(seenAtLeastOneDF[c] == 'Very favorably') | (seenAtLeastOneDF[c] == 'Somewhat favorably')]) / total
            favorable.append(percfavorable)
            
            percneutral = len(seenAtLeastOneDF[(seenAtLeastOneDF[c] == 'Neither favorably nor unfavorably (neutral)') | pd.isna(seenAtLeastOneDF[c])]) / total
            neutral.append(percneutral)
            
            percunfavorable = len(seenAtLeastOneDF[(seenAtLeastOneDF[c] == 'Somewhat unfavorably') | (seenAtLeastOneDF[c] == 'Very unfavorably')]) / total
            unfavorable.append(percunfavorable)
            
            percunfamiliar = len(seenAtLeastOneDF[seenAtLeastOneDF[c] == 'Unfamiliar (N/A)']) / total
            unfamiliar.append(percunfamiliar)

    df_fav = pd.DataFrame({'Name': characters, 'Percentage': favorable})
    df_neut = pd.DataFrame({'Name': characters, 'Percentage': neutral})
    df_unfav = pd.DataFrame({'Name': characters, 'Percentage': unfavorable})
    df_unfam = pd.DataFrame({'Name': characters, 'Percentage': unfamiliar})

    color_fav = '#8CC152'
    color_neut = '#4A89DC'
    color_unfav = 'red'
    color_unfam = 'gray'
    
    sorted_characters = [x for _, x in sorted(zip(favorable, characters), reverse=True)]


    bars_fav = alt.Chart(df_fav).mark_bar(size=30, color=color_fav).encode(
        x=alt.X('Percentage:Q', axis=None),
        y=alt.Y('Name:N', sort=sorted_characters, axis=alt.Axis(title='', labelColor='gray'))
    )

    bars_neut = alt.Chart(df_neut).mark_bar(size=30, color=color_neut).encode(
        x=alt.X('Percentage:Q', axis=None),
        y=alt.Y('Name:N', sort=sorted_characters, axis=None)
    )

    bars_unfav = alt.Chart(df_unfav).mark_bar(size=30, color=color_unfav).encode(
        x=alt.X('Percentage:Q', axis=None),
        y=alt.Y('Name:N', sort=sorted_characters, axis=None)
    )

    bars_unfam = alt.Chart(df_unfam).mark_bar(size=30, color=color_unfam).encode(
        x=alt.X('Percentage:Q', axis=None),
        y=alt.Y('Name:N', sort=sorted_characters, axis=None)
    )

    text_fav = bars_fav.mark_text(align='left', baseline='middle', dx=3, fontSize=12, color=color_fav).encode(
        text=alt.Text('Percentage:Q', format='.0%')
    )
    text_neut = bars_neut.mark_text(align='left', baseline='middle', dx=3, fontSize=12, color=color_neut).encode(
        text=alt.Text('Percentage:Q', format='.0%')
    )
    text_unfav = bars_unfav.mark_text(align='left', baseline='middle', dx=3, fontSize=12, color=color_unfav).encode(
        text=alt.Text('Percentage:Q', format='.0%')
    )
    text_unfam = bars_unfam.mark_text(align='left', baseline='middle', dx=3, fontSize=12, color=color_unfam).encode(
        text=alt.Text('Percentage:Q', format='.0%')
    )
    chart_fav = (bars_fav + text_fav).properties(
        title=alt.TitleParams("Favorable", anchor='middle', fontSize=14, fontWeight='normal'),
        width=150,
        height=500
    )
    chart_neut = (bars_neut + text_neut).properties(
        title=alt.TitleParams("Neutral", anchor='middle', fontSize=14, fontWeight='normal'),
        width=150,
        height=500
    )
    chart_unfav = (bars_unfav + text_unfav).properties(
        title=alt.TitleParams("Unfavorable", anchor='middle', fontSize=14, fontWeight='normal'),
        width=150,
        height=500
    )
    chart_unfam = (bars_unfam + text_unfam).properties(
        title=alt.TitleParams("Unfamiliar", anchor='middle', fontSize=14, fontWeight='normal'),
        width=150,
        height=500
    )

    combined = (alt.hconcat(chart_fav, chart_neut, chart_unfav, chart_unfam)
                .resolve_scale(y='shared')
                .properties(
                    title={
                        "text": "'Star Wars' Character Favorability Ratings",
                        "subtitle": [
                            f"By {total} respondents"
                        ],
                        "anchor": "start",
                        "offset": 20
                    }
                )
                .configure_view(strokeWidth=0)
                .configure_title(
                    fontSize=25,
                    subtitleFontSize=14,
                    subtitlePadding=4
                )
    )

    return combined

    
            
        
```

### 2.4 Who shot first? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genFirstShotVis(inpf):
    # input: inpf, the star wars dataset
    # output: the Altair visualization as described above
    
    # YOUR CODE HERE
    seenAtLeastOneDF = getSeenAtLeastOneDF(inpf, episodes)
    total = len(seenAtLeastOneDF)
    opts = ['Han', 'Greedo', "I don't understand this question"]
    percs = []

    han = (seenAtLeastOneDF['Which character shot first?'] == 'Han').sum() / total
    percs.append(han)
    greedo = (seenAtLeastOneDF['Which character shot first?'] == 'Greedo').sum() / total
    percs.append(greedo)
    na = ((seenAtLeastOneDF['Which character shot first?'] == "I don't understand this question") | pd.isna(seenAtLeastOneDF['Which character shot first?'])).sum() / total
    percs.append(na)

    tuples = list(zip(opts,percs))
    rank_per_df = pd.DataFrame(tuples, columns = ['Option', 'Percentage'])

    bars = alt.Chart(rank_per_df).mark_bar(size=28).encode(
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            'Option:N',
             axis=alt.Axis(tickCount=5, title='', labelColor='gray'),
             sort=opts
        )
    )

    text = bars.mark_text(
        align='right',
        baseline='middle',
        dx=25
    ).encode(
        text=alt.Text('Percentage:Q',format='.0%')
    )

    best = (text + bars).configure_mark(
        color='#008fd5'
    ).configure_view(
        strokeWidth=0
    ).configure_scale(
        bandPaddingInner=0.1
    ).properties(
        width=500,
        height=180
    )
    return (best).properties(title={
            "text": "Who Shot First?",
            "subtitle": [
                f"According to {total} respondents"
            ],
            "anchor": "start",
            "offset": 30
        }
    ).configure_view(
        strokeWidth=0  
    ).configure_title(
        fontSize=24,
        subtitleFontSize=14,
        subtitlePadding=4
    )
    

    
```
