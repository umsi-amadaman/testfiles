We ran a poll through [SurveyMonkey Audience](https://www.surveymonkey.com/mp/audience/), surveying 1,186 respondents from June 3 to 6 (the [data](https://github.com/fivethirtyeight/data/tree/master/star-wars-survey) is available [on GitHub](https://github.com/fivethirtyeight/data)). Seventy-nine percent of those respondents said they had watched at least one of the “Star Wars” films.
```python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    # create the various mapping and order lists/dictionaries
    # return: list of episodes (in order), episode->name map, list of episode names (in order)
    episodes = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    names = {
        'EI' : 'The Phantom Menace', 'EII' : 'Attack of the Clones', 'EIII' : 'Revenge of the Sith', 
        'EIV': 'A New Hope', 'EV': 'The Empire Strikes Back', 'EVI' : 'The Return of the Jedi'
    }

    # we're also going to use this order to sort, so names_l will now have our sort order
    return episodes, names, [names[ep] for ep in episodes]

episodes, namesDict, namesList = genEpisodeNamesDF()
```

```python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
```

```python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN
    salo = indf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    return(salo)
```

```python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
```

```python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total = len(inpf)
    percs = []

    # loop over each column and calculate the number of people who have seen the movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count them
    # and divide by the percent
    for seen_ep in ['seen_' + ep for ep in episodes]:
        perc = len(inpf[~ pd.isna(inpf[seen_ep])]) / total
        percs.append(perc)

    # at this point percs is holding our percentages

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes],percs))
    seen_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    return(seen_per_df)
```

```python
seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
```

```python
# let's see what's inside
seenPerMovieDF
```

```python
def genPercentVis(indf, names_l=namesList):
    # input: indf, the dataframe as seen_per_df
    # input: names_l - a list of all the movies in series order (default namesList)
    # output: simple altair bar chart
    
    # ok, time to make the chart... let's make a bar chart (use mark_bar)
    bars = alt.Chart(indf).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        )
    )

    # at this point we don't really have a great plot (it's missing the annotations, titles, etc.)
    return(bars)

seenPerMovieVis = genPercentVis(seenPerMovieDF)
```

```python
# display it
seenPerMovieVis
```

```python
def augmentPercentVis(base):
    # input: base (the base vis, i.e., bars as above)
    # we're going to overlay the text with the percentages, so let's make another visualization
    # that's just text labels


    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    seen_movies = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=180
    ).properties(
        # add a title
        title="Which 'Star Wars' Movies Have you Seen?"
    )

    return(seen_movies)

# note that we are NOT formatting this in the Five Thirty Eight Style yet... we'll leave that to you to figure out
```

```python
# Recreate this image using Altair
# match the "538 style" as best you can (hint: look at the altair lab at the start of the semester)

def genBestVis(inpf, eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above
    salo = inpf.dropna(subset=['seen_' + ep for ep in eps],how='any')

    total = len(salo)
    percs = []

    for rank_ep in ['rank_' + ep for ep in eps]:
        perc = len(salo[salo[rank_ep] == '1']) / total
        percs.append(perc)

    tuples = list(zip([names[ep] for ep in eps],percs))
    rank_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])

    bars = alt.Chart(rank_per_df).mark_bar(size=24).encode(
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             sort=names_l
        )
    )

    text = bars.mark_text(
        align='left',
        baseline='middle',
        dx=3  
    ).encode(
        text=alt.Text('Percentage:Q',format='.0%')
    )

    rank_movies = (text + bars).configure_mark(
        color='#008fd5'
    ).configure_view(
        strokeWidth=0
    ).configure_scale(
        bandPaddingInner=0.1
    ).properties(
        width=600,
        height=180
    ).properties(
        title=alt.TitleParams(
            text="What's the best 'Star Wars' Movie?",
            subtitle=["of " + str(total) + " respondents who have seen all six films","                                    "],
            subtitleColor='black'
    )).configure_title(
    anchor='start',
    fontSize=24,
    subtitleFontSize=13,
    font='Times New Roman',
    color = 'black'
    ).configure_view(
    fill="white"
    ).configure_axis(
        labelFontSize=13,
        labelFont='Times New Roman',
        orient='left',
        labelColor = 'gray'
    )
    
    return rank_movies
```

### 2.2 How people rate the 'Star Wars' movie? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genRateVis(inpf,eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above
    salo = inpf.dropna(subset=['seen_' + ep for ep in eps],how='any')
    
    total = len(salo)
    top_percs = []
    middle_percs = []
    bottom_percs = []

    for top_rank_ep in ['rank_' + ep for ep in eps]:
        top_perc = len(salo[salo[top_rank_ep] == '1'] + salo[salo[top_rank_ep] == '2']) / total
        top_percs.append(top_perc)

    for middle_rank_ep in ['rank_' + ep for ep in eps]:
        middle_perc = len(salo[salo[middle_rank_ep] == '3'] + salo[salo[middle_rank_ep] == '4']) / total
        middle_percs.append(middle_perc)

    for bottom_rank_ep in ['rank_' + ep for ep in eps]:
        bottom_perc = len(salo[salo[bottom_rank_ep] == '5'] + salo[salo[bottom_rank_ep] == '6']) / total
        bottom_percs.append(bottom_perc)

    top_tuples = list(zip([names[ep] for ep in eps],top_percs))
    top_rank_per_df = pd.DataFrame(top_tuples, columns = ['Name', 'Top third'])

    middle_tuples = list(zip([names[ep] for ep in eps],middle_percs))
    middle_rank_per_df = pd.DataFrame(middle_tuples, columns = ['Name', 'Middle third'])

    bottom_tuples = list(zip([names[ep] for ep in eps],bottom_percs))
    bottom_rank_per_df = pd.DataFrame(bottom_tuples, columns = ['Name', 'Bottom third'])
    
    top_bars = alt.Chart(top_rank_per_df).mark_bar(size=20, color='green').encode(
        x=alt.X(
            'Top third',
            axis=None),
        y=alt.Y(
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             sort=names_l
                ) 
    )

    middle_bars = alt.Chart(middle_rank_per_df).mark_bar(size=20, color='#008fd5').encode(
        x=alt.X(
            'Middle third',
            axis=None),
        y=alt.Y(
            'Name:N',
             axis=None,
             sort=names_l
                ) 
    )

    bottom_bars = alt.Chart(bottom_rank_per_df).mark_bar(size=20, color='red').encode(
        x=alt.X(
            'Bottom third',
            axis=None),
        y=alt.Y(
            'Name:N',
             axis=None,
             sort=names_l
                ) 
    )

    top_text = top_bars.mark_text(
        align='left',
        baseline='middle',
        dx=3,
        color='green'
    ).encode(
        text=alt.Text('Top third:Q',format='.0%')
    )

    middle_text = middle_bars.mark_text(
        align='left',
        baseline='middle',
        dx=3,
        color='#008fd5'
    ).encode(
        text=alt.Text('Middle third:Q',format='.0%')
    )

    bottom_text = bottom_bars.mark_text(
        align='left',
        baseline='middle',
        dx=3,
        color='red'
    ).encode(
        text=alt.Text('Bottom third:Q',format='.0%')
    )

    top_chart = (top_text + top_bars).properties(
        # set the dimensions of the visualization
        width=100,
        height=180
    ).properties(
        # add a title
        title="Top third"
    )

    middle_chart = (middle_text + middle_bars).properties(
        # set the dimensions of the visualization
        width=100,
        height=180
    ).properties(
        # add a title
        title="Middle third"
    )

    bottom_chart = (bottom_text + bottom_bars).properties(
        # set the dimensions of the visualization
        width=100,
        height=180
    ).properties(
        # add a title
        title="Bottom third"
    )

    all_charts = alt.hconcat(top_chart, middle_chart, bottom_chart).resolve_scale(y='shared').configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        title=alt.TitleParams(
            text="How People Rate the 'Star Wars' Movies",
            subtitle=["How often each film was rated in the top, middle and bottom third","(by " + str(total) + " respondents who have seen all six films)", "                                     "],
            subtitleColor='black'
    )).configure_title(
    anchor='start',
    fontSize=24,
    subtitleFontSize=12,
    font='Times New Roman',
    color='black'
    ).configure_view(
    fill="white"
    ).configure_axis(
        labelFontSize=13,
        labelFont='Times New Roman',
        orient='left',
        labelColor = 'gray'
    )

    return all_charts
```

```python
# Recreate this image using altair here (10 POINTS)

def genFavorabilityVis(inpf):
    # input: inpf, the star wars dataset (as defined at top of file)
    # output: the Altair visualization as described above

    characterList = ['Luke Skywalker', 'Han Solo', 'Princess Leia Organa', 'Obi Wan Kenobi', 'Yoda', 'R2 D2', 'C-3P0', 'Anakin Skywalker', 'Darth Vader', 'Lando Calrissian', 'Padme Amidala', 
                     'Boba Fett', 'Emperor Palpatine', 'Jar Jar Binks']

    salo = inpf.dropna(subset=characterList,how='all')

    fave_percs = []
    neutral_percs = []
    unfave_percs = []
    unfamiliar_percs = []

    for fave_char in characterList:
        fave_perc = (len(salo[salo[fave_char] == 'Very favorably']) + len(salo[salo[fave_char] == 'Somewhat favorably'])) / len(salo[fave_char])
        fave_percs.append(fave_perc)

    for neutral_char in characterList:
        neutral_perc = len(salo[salo[neutral_char] == 'Neither favorably nor unfavorably (neutral)']) / len(salo[neutral_char])
        neutral_percs.append(neutral_perc)

    for unfave_char in characterList:
        unfave_perc = (len(salo[salo[unfave_char] == 'Very unfavorably']) + len(salo[salo[unfave_char] == 'Somewhat unfavorably'])) / len(salo[unfave_char])
        unfave_percs.append(unfave_perc)

    for unfamiliar_char in characterList:
        unfamiliar_perc = len(salo[salo[unfamiliar_char] == 'Unfamiliar (N/A)']) / len(salo[unfamiliar_char])
        unfamiliar_percs.append(unfamiliar_perc)

    fave_tuples = list(zip([fave_char for fave_char in characterList],fave_percs))
    fave_per_df = pd.DataFrame(fave_tuples, columns = ['Name', 'Favorable'])

    neutral_tuples = list(zip([neutral_char for neutral_char in characterList],neutral_percs))
    neutral_per_df = pd.DataFrame(neutral_tuples, columns = ['Name', 'Neutral'])

    unfave_tuples = list(zip([unfave_char for unfave_char in characterList],unfave_percs))
    unfave_per_df = pd.DataFrame(unfave_tuples, columns = ['Name', 'Unfavorable'])

    unfamiliar_tuples = list(zip([unfamiliar_char for unfamiliar_char in characterList],unfamiliar_percs))
    unfamiliar_per_df = pd.DataFrame(unfamiliar_tuples, columns = ['Name', 'Unfamiliar'])

    fave_bars = alt.Chart(fave_per_df).mark_bar(size=18, color='green').encode(
        x=alt.X(
            'Favorable',
            axis=None),
        y=alt.Y(
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             sort=characterList
                ) 
    )

    neutral_bars = alt.Chart(neutral_per_df).mark_bar(size=18, color='#008fd5').encode(
        x=alt.X(
            'Neutral',
            axis=None),
        y=alt.Y(
            'Name:N',
             axis=None,
             sort=characterList
                ) 
    )

    unfave_bars = alt.Chart(unfave_per_df).mark_bar(size=18, color='red').encode(
        x=alt.X(
            'Unfavorable',
            axis=None),
        y=alt.Y(
            'Name:N',
             axis=None,
             sort=characterList
                ) 
    )

    unfamiliar_bars = alt.Chart(unfamiliar_per_df).mark_bar(size=18, color='gray').encode(
        x=alt.X(
            'Unfamiliar',
            axis=None),
        y=alt.Y(
            'Name:N',
             axis=None,
             sort=characterList
                ) 
    )

    fave_text = fave_bars.mark_text(
        align='left',
        baseline='middle',
        dx=3,
        color='green'
    ).encode(
        text=alt.Text('Favorable:Q',format='.0%')
    )

    neutral_text = neutral_bars.mark_text(
        align='left',
        baseline='middle',
        dx=3,
        color='#008fd5'
    ).encode(
        text=alt.Text('Neutral:Q',format='.0%')
    )

    unfave_text = unfave_bars.mark_text(
        align='left',
        baseline='middle',
        dx=3,
        color='red'
    ).encode(
        text=alt.Text('Unfavorable:Q',format='.0%')
    )

    unfamiliar_text = unfamiliar_bars.mark_text(
        align='left',
        baseline='middle',
        dx=3,
        color='gray'
    ).encode(
        text=alt.Text('Unfamiliar:Q',format='.0%')
    )

    fave_chart = (fave_text + fave_bars).properties(
        # set the dimensions of the visualization
        width=75,
        height=300
    ).properties(
        # add a title
        title="Favorable"
    )

    neutral_chart = (neutral_text + neutral_bars).properties(
        # set the dimensions of the visualization
        width=75,
        height=300
    ).properties(
        # add a title
        title="Neutral"
    )

    unfave_chart = (unfave_text + unfave_bars).properties(
        # set the dimensions of the visualization
        width=75,
        height=300
    ).properties(
        # add a title
        title="Unfavorable"
    )

    unfamiliar_chart = (unfamiliar_text + unfamiliar_bars).properties(
        # set the dimensions of the visualization
        width=75,
        height=300
    ).properties(
        # add a title
        title="Unfamiliar"
    )

    all_charts = alt.hconcat(fave_chart, neutral_chart, unfave_chart, unfamiliar_chart).resolve_scale(y='shared').configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        title=alt.TitleParams(
            text="'Star Wars' Characters Favorability Ratings",
            subtitle=["By " + str(len(salo)) + " respondents", "                                       "],
            subtitleColor = 'black'
    )).configure_title(
    anchor='start',
    fontSize=24,
    subtitleFontSize=13,
    font='Times New Roman',
    color='black'
    ).configure_view(
    fill="white"
    ).configure_axis(
        labelFontSize=13,
        labelFont='Times New Roman',
        orient='left',
        labelColor = 'gray'
    )
    
    return all_charts
```

### 2.4 Who shot first? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genFirstShotVis(inpf):
    # input: inpf, the star wars dataset
    # output: the Altair visualization as described above
    
    salo = inpf.dropna(subset=['Which character shot first?'],how='all')

    han_count = 0
    greedo_count = 0
    idk_count = 0

    for ans in salo['Which character shot first?']:
        if ans == 'Han':
            han_count += 1
        elif ans == 'Greedo':
            greedo_count += 1
        else:
            idk_count += 1

    han_perc = han_count / len(salo)
    greedo_perc = greedo_count / len(salo)
    idk_perc = idk_count / len(salo)

    data = {
        'Answer': ['Han', 'Greedo', 'I don\'t understand this question'],
        'Percentage': [han_perc, greedo_perc, idk_perc]
    }

    df = pd.DataFrame(data)

    answerList = ['Han', 'Greedo', 'I don\'t understand this question']

    bars = alt.Chart(df).mark_bar(size=30).encode(
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            'Answer:N',
             title=None,
             sort=answerList,
             axis=alt.Axis(labelLimit=500)
        )
    )

    text = bars.mark_text(
        align='left',
        baseline='middle',
        dx=3  
    ).encode(
        text=alt.Text('Percentage:Q',format='.0%')
    )

    who_shot = (text + bars).configure_mark(
        color='#008fd5'
    ).configure_view(
        strokeWidth=0
    ).configure_scale(
        bandPaddingInner=0.2
    ).properties(
        width=600,
        height=150
    ).properties(
        title=alt.TitleParams(
            text="Who Shot First?",
            subtitle=["According to " + str(len(salo)) + " respondents", "                                 "]
    )).configure_title(
    anchor='start',
    fontSize=35,
    subtitleFontSize=18,
    font='Times New Roman'
    ).configure_view(
    fill="white"
    ).configure_axis(
        labelFontSize=20,
        labelFont='Times New Roman',
        orient='left' 
    )
    
    return who_shot
```
