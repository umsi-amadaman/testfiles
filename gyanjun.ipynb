```python
# We're going to fix the labels a bit so will create a mapping to the full names
# and have the full sort order
def genEpisodeNamesDF():
    # create the various mapping and order lists/dictionaries
    # return: list of episodes (in order), episode->name map, list of episode names (in order)
    episodes = ['EI', 'EII', 'EIII', 'EIV', 'EV', 'EVI']
    names = {
        'EI' : 'The Phantom Menace', 'EII' : 'Attack of the Clones', 'EIII' : 'Revenge of the Sith', 
        'EIV': 'A New Hope', 'EV': 'The Empire Strikes Back', 'EVI' : 'The Return of the Jedi'
    }

    # we're also going to use this order to sort, so names_l will now have our sort order
    return episodes, names, [names[ep] for ep in episodes]

episodes, namesDict, namesList = genEpisodeNamesDF()
```

```python
# let's inspect what we've generated. These will be useful to you below
print("abbreviated list (sorted):\n  ",episodes)
print("\nmapping between abberviated names and full titles:\n  ",namesDict)
print("\nfull titles, sorted:\n  ",namesList)
```

```python
# let's do some data pre-processing... recall that sw (star wars) has everything

def getSeenAtLeastOneDF(indf, eps):
    # input: indf the data file as formatted above
    # input: eps a list of episodes (movies)

    # returns a substet of the dataset
    
    # We want to only use those people who have seen at least one movie, let's get the people, toss NAs
    # and get the total count

    # find people who have at least on of the columns (seen_*) not NaN
    salo = indf.dropna(subset=['seen_' + ep for ep in eps],how='all')
    
    return(salo)
```

```python
seenAtLeastOneDF = getSeenAtLeastOneDF(sw, episodes)
print("total who have seen at least one: ", len(seenAtLeastOneDF),"\nSample:")
display(seenAtLeastOneDF.sample(5))
```

```python
# for each movie, we're going to calculate the percents and generate a new data frame

def genSeenPercentDf(inpf, names=namesDict, names_l=namesList):
    # input: inpf - an input frame of the form output by getSeenAtLeastOneDF()
    # input: names - a dictionary of abbreviations to names
    # input: names_l - a list of all the movies in series order
    
    total = len(inpf)
    percs = []

    # loop over each column and calculate the number of people who have seen the movie
    # specifically, filter out the people who are *NaN* for a specific episode (e.g., ep_EII), count them
    # and divide by the percent
    for seen_ep in ['seen_' + ep for ep in episodes]:
        perc = len(inpf[~ pd.isna(inpf[seen_ep])]) / total
        percs.append(perc)

    # at this point percs is holding our percentages

    # now we're going use a trick to make tuples--pairing names with percents--using "zip" and then make a dataframe
    tuples = list(zip([names[ep] for ep in episodes],percs))
    seen_per_df = pd.DataFrame(tuples, columns = ['Name', 'Percentage'])
    return(seen_per_df)
```

```python
seenPerMovieDF = genSeenPercentDf(seenAtLeastOneDF)
```

```python
def genPercentVis(indf, names_l=namesList):
    # input: indf, the dataframe as seen_per_df
    # input: names_l - a list of all the movies in series order (default namesList)
    # output: simple altair bar chart
    
    # ok, time to make the chart... let's make a bar chart (use mark_bar)
    bars = alt.Chart(indf).mark_bar(size=20).encode(
        # encode x as the percent, and hide the axis
        x=alt.X(
            'Percentage',
            axis=None),
        y=alt.Y(
            # encode y using the name, use the movie name to label the axis, sort using the names_l
            'Name:N',
             axis=alt.Axis(tickCount=5, title=''),
             # we give the sorting order to avoid alphabetical order
             sort=names_l
        )
    )

    # at this point we don't really have a great plot (it's missing the annotations, titles, etc.)
    return(bars)

seenPerMovieVis = genPercentVis(seenPerMovieDF)
```

```python
def augmentPercentVis(base):
    # input: base (the base vis, i.e., bars as above)
    # we're going to overlay the text with the percentages, so let's make another visualization
    # that's just text labels


    text = base.mark_text(
        align='left',
        baseline='middle',
        dx=3  # Nudges text to right so it doesn't appear on top of the bar
    ).encode(
        # we'll use the percentage as the text
        text=alt.Text('Percentage:Q',format='.0%')
    )

    # finally, we're going to combine the bars and the text and do some styling
    seen_movies = (text + base).configure_mark(
        # we don't love the blue
        color='#008fd5'
    ).configure_view(
        # we don't want a stroke around the bars
        strokeWidth=0
    ).configure_scale(
        # add some padding
        bandPaddingInner=0.2
    ).properties(
        # set the dimensions of the visualization
        width=500,
        height=180
    ).properties(
        # add a title
        title="Which 'Star Wars' Movies Have you Seen?"
    )

    return(seen_movies)

# note that we are NOT formatting this in the Five Thirty Eight Style yet... we'll leave that to you to figure out
```

### 2.2 How people rate the 'Star Wars' movie? Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genRateVis(inpf,eps, names=namesDict, names_l=namesList):
    # input: inpf, the star wars dataset
    # input: eps, the list of episodes
    # input: names - a dictionary of abbreviations to names (default: namesDict)
    # input: names_l - a list of all the movies in series order (default: namesList)
    # output: the Altair visualization as described above

     # Generate first a subdataframe with respondents who have seen all star wars films
    seenAllDF = inpf.dropna(subset=['seen_' + ep for ep in eps],how='any')
    
    # Generate new grouping dataframe
    Top_list = []
    Middle_list = []
    Bottom_list = []
    
    for seen_ep in ['rank_' + ep for ep in eps]:
        Top = seenAllDF[seen_ep].isin(["1", "2"]).sum()
        Top_list.append(Top)
        Middle = seenAllDF[seen_ep].isin(["3", "4"]).sum()
        Middle_list.append(Middle)
        Bottom = seenAllDF[seen_ep].isin(["5", "6"]).sum()
        Bottom_list.append(Bottom)
    
    movieRankDF = pd.DataFrame({
            'Name': [names[ep] for ep in eps],
            'Top count': Top_list,
            'Middle count': Middle_list,
            'Bottom count': Bottom_list
        })
        
    # loop over each rank column and calculate the number of people who rated a given film as 1
    movieRankDF['Total count'] = movieRankDF.sum(axis=1)
    movieRankDF['Top third'] = movieRankDF['Top count']/movieRankDF['Total count']
    movieRankDF['Middle third'] = movieRankDF['Middle count']/movieRankDF['Total count']
    movieRankDF['Bottom third'] = movieRankDF['Bottom count']/movieRankDF['Total count']
    
    # Clean the dataframes
    movieRankDF.drop(movieRankDF.filter(regex = 'count$').columns, axis = 1, inplace = True)
    # reshape for visualization
    movieRankPerc = movieRankDF.melt(id_vars=["Name"], var_name="Group", value_name="Percentage")
    # define colour to be used for each group
    color_scale = alt.Scale(
        domain=["Top third", "Middle third", "Bottom third"],  
        range=["green", "blue", "red"] 
    )
    # Generate base bar chart
    base = alt.Chart(movieRankPerc).encode(
        x=alt.X('Percentage:Q', axis=None),
        y=alt.Y('Name:N', title=None, sort=names_l)
    ).properties(
        width=70)
    
    bars = base.mark_bar().encode(
        color = alt.Color('Group:N', scale=color_scale, legend=None)
    )
    
    text = base.mark_text(dx=15).encode(
        text=alt.Text('Percentage:Q', format=".0%"),
        color=alt.Color('Group:N', scale=color_scale, legend=None)
    )
    
    group_order = ["Top third", "Middle third", "Bottom third"]
    
    # Faceting charts
    chart = (bars + text).facet(
        column=alt.Column('Group:N', title=None, sort=group_order)
    ).properties(
        title={
            "text": "How People Rate 'Star Wars' Movies",
            "subtitle": ["How often each film was rated in the top, middle, and bottom third","(by 471 respondents who have seen the film)"]
        }
    ).configure(
        background='#F0F0F0'
    ).configure_view(
        stroke='transparent'
    )
    
    return(chart)
```

### 2.3 Star Wars' Characters Favorability Ratings. Recreate the above image using altair (10 Points)

```python
# Recreate this image using altair here (10 POINTS)

def genFavorabilityVis(inpf):
    # input: inpf, the star wars dataset (as defined at top of file)
    # output: the Altair visualization as described above
    
    # Create a list of characters
    char =  ['Han Solo', 'Luke Skywalker','Princess Leia Organa','Anakin Skywalker','Obi Wan Kenobi','Emperor Palpatine','Darth Vader','Lando Calrissian','Boba Fett','C-3P0','R2 D2','Jar Jar Binks','Padme Amidala','Yoda']
    
    # Get the subdataset where the respondents have rated at least one character
    charDF = sw[char].dropna(how = "all")
    
    # Create a mapping of the grouped rating
    rating_mapping = {
      "Very favorably": "Favorable",
      "Somewhat favorably": "Favorable",
      "Somewhat unfavorably": "Unfavorable",
      "Very unfavorably": "Unfavorable",
      "Neither favorably nor unfavorably (neutral)": "Neutral",
      "Unfamiliar (N/A)": "Unfamiliar"
    }
    
    charDF = charDF.replace(rating_mapping) 
    
    # Compute percentage of each grouped rating for every character
    rating_percentages = charDF.apply(lambda x: x.value_counts(normalize=True))
    
    # reshape for visualization
    df_transposed = rating_percentages.T.reset_index()
    
    # Convert to long format
    df_long = df_transposed.melt(id_vars=["index"], var_name="Rating Category", value_name="Percentage").rename(columns={"index": "Character"})
    
    # define colour to be used for each group
    color_mapping = {
        "Favorable": "green",
        "Neutral": "blue",
        "Unfavorable": "red",
        "Unfamiliar": "grey"
    }
    
    # Order characters based on their favorability
    favorable_sorted_list = rating_percentages.loc["Favorable"].sort_values(ascending=False).index.tolist()

    # Write a function to create the base chart
    def create_chart(category, color, sort_list):
        return alt.Chart(df_long[df_long["Rating Category"] == category]).mark_bar(size=20).encode(
            x=alt.X('Percentage:Q', title="", axis=None),
            y=alt.Y('Character:N', title=None, sort=sort_list),
            color=alt.value(color)
        ).properties(
            title=category,
            width=200,
            height=500
        )
    def create_chart(category, color, sort_list):
        base = alt.Chart(df_long[df_long["Rating Category"] == category]).encode(
          x=alt.X('Percentage:Q', title="", axis=None),
          y=alt.Y('Character:N', title=None, sort=sort_list)
        ).properties(
          title=category,
          width=240,
          height=500
        )

        bars = base.mark_bar(size=20).encode(
            color=alt.value(color)
        )
        
        
        text = base.mark_text(
          align='left',     
          baseline='middle',
          dx=6
        ).encode(
            text=alt.Text('Percentage:Q', format='.0%'),
            color=alt.value('black')
        )
        return bars + text

    # Create four separate charts
    favorable_chart = create_chart("Favorable", color_mapping["Favorable"],favorable_sorted_list)
    neutral_chart = create_chart("Neutral", color_mapping["Neutral"],favorable_sorted_list)
    unfavorable_chart = create_chart("Unfavorable", color_mapping["Unfavorable"],favorable_sorted_list)
    unfamiliar_chart = create_chart("Unfamiliar", color_mapping["Unfamiliar"],favorable_sorted_list)
    
    # Concatenate all four charts horizontally
    final_chart = alt.hconcat(favorable_chart, neutral_chart, unfavorable_chart, unfamiliar_chart).properties(
        title={
            "text": "'Star Wars' Characters Favorability Ratings",
            "subtitle": "By 834 respondents"}
    ).configure(
        background='#F0F0F0'
    ).configure_view(
        stroke='transparent'
    )

    return(final_chart)
```

```python
def generate_movie_rank_by_gender(inpf, eps, names=namesDict, names_l=namesList):
    # Remove rows where any seen episode is missing
    seenAllDF = inpf.dropna(subset=['seen_' + ep for ep in eps], how='any')
    
    # Initialize lists for overall ranking
    Top_list, Middle_list, Bottom_list = [], [], []
    
    # Initialize lists for gender-specific ranking
    Top_male_list, Middle_male_list, Bottom_male_list = [], [], []
    Top_female_list, Middle_female_list, Bottom_female_list = [], [], []
    
    # Process each movie ranking
    for seen_ep in ['rank_' + ep for ep in eps]:
        # Compute overall counts
        Top = seenAllDF[seen_ep].isin(["1", "2"]).sum()
        Middle = seenAllDF[seen_ep].isin(["3", "4"]).sum()
        Bottom = seenAllDF[seen_ep].isin(["5", "6"]).sum()
        # Append to lists
        Top_list.append(Top)
        Middle_list.append(Middle)
        Bottom_list.append(Bottom)
        
        # Compute counts by gender
        male_df = seenAllDF[seenAllDF['Gender'] == 'Male']
        female_df = seenAllDF[seenAllDF['Gender'] == 'Female']
        Top_male_list.append(male_df[seen_ep].isin(["1", "2"]).sum())
        Middle_male_list.append(male_df[seen_ep].isin(["3", "4"]).sum())
        Bottom_male_list.append(male_df[seen_ep].isin(["5", "6"]).sum())
        Top_female_list.append(female_df[seen_ep].isin(["1", "2"]).sum())
        Middle_female_list.append(female_df[seen_ep].isin(["3", "4"]).sum())  # <-- if your column is 'seen_ep', keep it lowercase; otherwise fix here
        Bottom_female_list.append(female_df[seen_ep].isin(["5", "6"]).sum())

    # Create DataFrames for overall and gender-specific rankings
    movieRankDF = pd.DataFrame({
        'Name': [names[ep] for ep in eps],
        'Top count': Top_list,
        'Middle count': Middle_list,
        'Bottom count': Bottom_list
    })
    movieRankDF_male = pd.DataFrame({
        'Name': [names[ep] for ep in eps],
        'Top count': Top_male_list,
        'Middle count': Middle_male_list,
        'Bottom count': Bottom_male_list
    })
    movieRankDF_female = pd.DataFrame({
        'Name': [names[ep] for ep in eps],
        'Top count': Top_female_list,
        'Middle count': Middle_female_list,
        'Bottom count': Bottom_female_list
    })

    # Compute total counts
    movieRankDF['Total count'] = movieRankDF.sum(axis=1)
    movieRankDF_male['Total count'] = movieRankDF_male.sum(axis=1)
    movieRankDF_female['Total count'] = movieRankDF_female.sum(axis=1)

    # Compute proportions
    for df in [movieRankDF, movieRankDF_male, movieRankDF_female]:
        df['Top third'] = df['Top count'] / df['Total count'] * 100
        df['Middle third'] = df['Middle count'] / df['Total count'] * 100
        df['Bottom third'] = df['Bottom count'] / df['Total count'] * 100

    # Remove raw count columns
    for df in [movieRankDF, movieRankDF_male, movieRankDF_female]:
        df.drop(df.filter(regex='count$').columns, axis=1, inplace=True)

    # Reshape for visualization
    movieRankPerc = movieRankDF.melt(id_vars=["Name"], var_name="Group", value_name="Percentage")
    movieRankPerc_male = movieRankDF_male.melt(id_vars=["Name"], var_name="Group", value_name="Percentage")
    movieRankPerc_female = movieRankDF_female.melt(id_vars=["Name"], var_name="Group", value_name="Percentage")

    # Add gender labels for better visualization
    movieRankPerc['Gender'] = 'All'
    movieRankPerc_male['Gender'] = 'Male'
    movieRankPerc_female['Gender'] = 'Female'

    # Concatenate all three datasets
    final_movieRankPerc = pd.concat([movieRankPerc, movieRankPerc_male, movieRankPerc_female])

    # Define color scale (optional; not used in the return)
    color_scale = alt.Scale(
        domain=["Top third", "Middle third", "Bottom third"],
        range=["green", "blue", "red"]
    )

    # Return the final structured DataFrame
    return final_movieRankPerc
```
